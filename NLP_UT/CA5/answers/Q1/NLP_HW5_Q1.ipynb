```python

docs = []
metadata = []
doc_id_counter = 0

file_to_title = {
    'نسخه چاپی قانون تأمين اجتماعي.pdf': 'قانون تأمین اجتماعی',
    'نسخه چاپی قانون ثبت اسناد و املاك.pdf': 'قانون ثبت اسناد و املاک',
    'نسخه چاپی قانون روابط موجر و مستأجر_page-0001.pdf': 'قانون روابط موجر و مستأجر',
    'نسخه چاپی قانون صدور چك_page-0001.pdf': 'قانون صدور چک',
    'نسخه چاپی قانون كار.pdf': 'قانون کار',
    'نسخه چاپی قانون ماليات بر ارزش افزوده.pdf': 'قانون مالیات بر ارزش افزوده',
    'نسخه چاپی قانون مبارزه با قاچاق كالا و ارز.pdf': 'قانون مبارزه با قاچاق کالا و ارز',
}

```

```python
pdf_files = [f for f in os.listdir(PDF_DIR) if f.endswith(".pdf")]
pdf_files.sort()
```

```python


for filename in pdf_files:
    filepath = os.path.join(PDF_DIR, filename)
    try:
        reader = PdfReader(filepath)
        text_content = ""
        for page in reader.pages:
            text_content += page.extract_text() + "\n"

        text_content = " ".join(text_content.split())

        if text_content:
            docs.append(text_content)
            title = file_to_title.get(filename, filename.replace('.pdf', ''))
            metadata.append({
                "doc_id": doc_id_counter,
                "title": title,
                "filename": filename,
                "length": len(text_content)
            })
            doc_id_counter += 1
    except Exception as e:
        print(f"Error processing {filename}: {e}")
```

```python
vectorizer = TfidfVectorizer(token_pattern=r'\b\w+\b', min_df=1)
X = vectorizer.fit_transform(docs)

```

```python

try:
    plt.rcParams['font.family'] = 'Tahoma'
except:
    pass
```

```python
df_metadata = pd.DataFrame(metadata).sort_values(by='length', ascending=False)

plt.figure(figsize=(12, 6))
plt.bar(df_metadata['title'], df_metadata['length'], color='skyblue')
plt.ylabel('تعداد کاراکتر (طول سند)')
plt.title('توزیع طول اسناد قانونی')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig("document_length_distribution.png")
print("Visualiztion saved to document_length_distribution.png")

```

```python


print(f"\n Number of documents: {X.shape[0]:,}")
print(f"Vocabulary size: {len(vectorizer.get_feature_names_out()):,}")

sums = X.sum(axis=0)
data_scores = []
for col, term in enumerate(vectorizer.get_feature_names_out()):
    data_scores.append((term, sums[0, col]))

df_scores = pd.DataFrame(data_scores, columns=['term', 'score']).sort_values(by='score', ascending=False).head(15)
```

```python

plt.figure(figsize=(12, 7))
plt.barh(df_scores['term'][::-1], df_scores['score'][::-1], color='lightcoral')
plt.xlabel('Total TF-IDF Score')
plt.title('Top 15 Terms by Total TF-IDF Score in the Corpus')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig("top_tfidf_terms.png")
print("Visualization saved to top_tfidf_terms.png")

```

## <div style="text-align: center; direction: rtl; font-family: Vazir;">طراحی عامل(۲۵ امتیاز)</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
تمامی کد های شما می بایست با استفاده از <a href="https://www.langchain.com/" target="_blank" rel="noopener noreferrer">langchain</a>
 , <a href="https://www.langchain.com/langgraph" target="_blank" rel="noopener noreferrer">langgraph</a>
 پیاده سازی شده باشد

قابلیت های کلی Agent باید به صورت زیر باشد :

یک نود با نام rewrite_query طراحی کنید که ورودی آن، query اصلی کاربر و خروجی آن، یک rewritten_query به‌عنوان نسخه‌ای بازنویسی‌شده و شفاف‌تر از سؤال باشد. وظیفه این ماژول آن است که ابهام‌های زبانی موجود در پرسش را برطرف کند، لحن را استاندارد و یکنواخت سازد و در نهایت پرسش کاربر را به فرمی تبدیل کند که برای مرحله‌ی بازیابی اطلاعات (retrieval) مناسب‌تر باشد.

نودی با نام classify_intent طراحی کنید که وظیفه آن تشخیص نوع پرسش کاربر بر اساس query دریافتی است. این نود باید تعیین کند که پرسش کاربر در کدام یک از دسته‌های «Greeting / Small Talk» (مانند سلام، احوال‌پرسی، تبریک و عبارات مشابه)، «Abusive / Insult / بی‌احترامی» یا «Law Question» (سؤال مرتبط با قانون و نیازمند مراجعه به پایگاه دانش) قرار می‌گیرد. در صورت تشخیص دسته‌ی Greeting، نود باید یک پاسخ کوتاه، محترمانه و دوستانه تولید کند. در صورت تشخیص دسته‌ی Abusive، نود موظف است پاسخی حرفه‌ای، محترمانه و بدون هرگونه توهین ارائه دهد و در آن توضیح دهد که مدل برای کمک طراحی شده است و وارد جدل یا بحث تنش‌زا نمی‌شود. اگر خروجی نود در دسته‌ی Law Question قرار گرفت، جریان پردازش باید به نود بعدی، یعنی نود «Metadata Extraction»، منتقل شود تا مراحل بعدی بازیابی اطلاعات انجام گیرد.

یک نود با نام extract_metadata طراحی کنید که ورودی آن rewritten_query و خروجی آن یک ساختار metadata_filter باشد؛ این ساختار در مرحله‌ی بازیابی اطلاعات از LanceDB برای اعمال فیلترهای مرتبط استفاده خواهد شد. در این نود، با استفاده از یک LLM، از روی rewritten_query اطلاعات ساختارمند و قابل‌استفاده به‌عنوان فیلتر استخراج می‌شود تا در قالب metadata_filter به LanceDB ارسال شده و بازیابی اسناد با دقت و ارتباط بیشتری انجام پذیرد.

نودی با نام context_retrieve طراحی کنید که با استفاده از rewritten_query در LanceDB عملیات جستجو را انجام دهد. این نود باید از ساختار metadata_filter برای محدود کردن فضای جستجو و اعمال فیلترهای موردنظر بهره بگیرد و در نهایت، تعداد مشخصی سند (chunk) مرتبط با پرسش کاربر، مثلاً به تعداد K سند (برای نمونه K = 10)، را به‌عنوان خروجی بازگرداند.

یک نود با نام rerank طراحی کنید که روی خروجی مرحله‌ی context_retrieve یک لایه‌ی بازمرتب‌سازی (Reranking) اعمال کند؛ به‌عنوان مثال می‌توان از یک LLM-based reranker یا یک معیار شباهت ثانویه برای امتیازدهی مجدد به اسناد استفاده کرد. این نود باید بر اساس مقادیر similarity یا score، تعداد محدودی از بهترین سندها (top-N، مثلاً N = 3) را برای مرحله‌ی تولید پاسخ (generation) انتخاب کند. علاوه بر این، در همین نود باید یک مرحله‌ی بررسی ارتباط (Relevance Check) انجام شود؛ به این صورت که اگر هیچ‌یک از contextهای انتخاب‌شده در top-N، بر اساس یک آستانه‌ی تعریف‌شده یا قضاوت LLM، ارتباط کافی با query نداشتند، فرآیند retrieval یک‌بار دیگر تکرار شده و تلاش مجدد برای بازیابی contextهای مرتبط‌تر انجام گیرد. دقت شود که نهایتا یک بار می بایست این مرحله انجام شود و Agent در loop بی نهایت قرار نگیرد پس این موضوع باید توسط شما کنترل شود.

نودی با نام generate_answer طراحی کنید که ورودی آن شامل rewritten_query به‌همراه contextهای نهایی انتخاب‌شده از مراحل قبل باشد و بر اساس این ورودی، وظیفه دارد یک پاسخ متنی نهایی، منسجم، دقیق و قابل‌فهم برای کاربر تولید و به‌عنوان خروجی برگرداند.

به طور کلی معماری پیشنهادی یک حالت از نحوه پیاده سازی این Agent هست و می توان نوع نگاه متفاوتی در طراحی معماری Agent داشت پس شما مجاز هستید که با پیاده سازی همه مواردی که بیان شد با معماری متفاوت عمل کنید معماری پیشنهادی شما می بایست تمامی این موارد را شامل شده باشد:

query rewriting, extract metadata , context retrieve , reranking , generate answer

```python

def rewrite_query(query: str) -> str:
    q = query.strip()
    q = re.sub(r"\s+", " ", q)
    return q

```

```python
GREET_KEYWORDS = ["سلام", "درود", "صبح بخیر", "وقت بخیر" , "خوش آمدید", "خوش اومدید", "سلامتی"]
ABUSE_KEYWORDS = ["مسخره", "بی‌ادب", "احمق", "نادان", "بی‌فکر", "بی‌عقل", "بی‌شعور" , "بی‌خرد"]
```

```python

def classify_intent(query: str) -> dict:
    q = query.lower()
    for w in GREET_KEYWORDS:
        if w in q:
            return {"intent": "greeting", "response": "سلام! چطور می‌توانم کمک کنم؟"}
    for w in ABUSE_KEYWORDS:
        if w in q:
            return {
                "intent": "abuse",
                "response": "من اینجا هستم تا کمک کنم؛ لطفاً مودبانه سوالاتتان را مطرح کنید.",
            }
    return {"intent": "law_question"}

```

```python
def extract_metadata(rewritten_query: str) -> dict:
    m = {}
    # try to extract year
    years = re.findall(r"13\d{2}|14\d{2}|\d{4}", rewritten_query)
    if years:
        try:
            m["year"] = int(years[0])
        except:
            pass

    # Key extraction for filtering
    keywords = []
    if "کار" in rewritten_query:
        keywords.append("کار")
    if "مدنی" in rewritten_query:
        keywords.append("مدنی")
    if "جرایم" in rewritten_query or "رایانه" in rewritten_query:
        keywords.append("جرایم")
    if keywords:
        m["keywords"] = keywords
    return m
```

```python

def context_retrieve(rewritten_query: str, metadata_filter: dict = None, K: int = 10):
    q_vec = vectorizer.transform([rewritten_query])
    sims = cosine_similarity(q_vec, X).flatten()

    idxs = np.argsort(-sims)

    if metadata_filter:
        filtered = []
        for i in idxs:
            meta = metadata[i]
            ok = True
            if "year" in metadata_filter and "year" in meta:
                ok = ok and (meta.get("year") == metadata_filter["year"])

            if "keywords" in metadata_filter:
                found_k = any(
                    k in meta.get("title", "") or k in docs[i]
                    for k in metadata_filter["keywords"]
                )
                ok = ok and found_k

            if ok:
                filtered.append((i, sims[i]))
        results = filtered[:K]
    else:
        results = [(int(i), float(sims[int(i)])) for i in idxs[:K]]

    contexts = []
    for i, score in results:
        contexts.append(
            {"doc_id": i, "text": docs[i], "metadata": metadata[i], "score": score}
        )
    return contexts
```

```python
def rerank(contexts, top_n=3, relevance_threshold=0.01):
    if not contexts:
        return []
    contexts = sorted(contexts, key=lambda x: -x["score"])
    top = contexts[:top_n]

    # relevance check
    if all(c["score"] < relevance_threshold for c in top):
        return []
    return top

```

```python


def generate_answer(rewritten_query: str, contexts):
    if not contexts:
        return "متاسفانه نتوانستم پاسخی مرتبط در اسناد پیدا کنم. لطفاً سوال را با کلمات کلیدی دقیق‌تر مطرح کنید."

    parts = []
    for c in contexts:
        snippet = c["text"][:400] + "..."
        source = c["metadata"]["title"]
        parts.append(f"منبع: {source} (امتیاز: {c['score']:.2f})\nمتن: {snippet}")

    answer_body = "\n\n---\n\n".join(parts)
    final = f"پاسخ پیشنهادی (یافته شده در منابع):\n\n{answer_body}"
    return final
```

```python

def run_pipeline(query: str, K=10, top_n=3):
    timings = {}
    t0 = time.time()
    timings["start"] = t0

    # Rewrite
    t = time.time()
    rewritten = rewrite_query(query)
    timings["rewrite"] = time.time() - t

    # Intent
    t = time.time()
    intent = classify_intent(rewritten)
    timings["classify_intent"] = time.time() - t

    if intent.get("intent") != "law_question":
        return {"answer": intent.get("response"), "timings": timings, "contexts": []}

    # Metadata
    t = time.time()
    meta_filter = extract_metadata(rewritten)
    timings["extract_metadata"] = time.time() - t

    # Retrieve
    t = time.time()
    contexts = context_retrieve(rewritten, meta_filter, K=K)
    timings["context_retrieve"] = time.time() - t

    # Rerank
    t = time.time()
    top = rerank(contexts, top_n=top_n)

    # Fallback: if empty, try wider search without filters
    if not top:
        contexts = context_retrieve(rewritten, None, K=min(len(docs), K * 2))
        top = rerank(contexts, top_n=top_n)
    timings["rerank"] = time.time() - t

    # Generate
    t = time.time()
    answer = generate_answer(rewritten, top)
    timings["generate_answer"] = time.time() - t

    timings["total"] = time.time() - t0
    return {"answer": answer, "timings": timings, "contexts": top}
```

```python
print("Loading index for benchmarking...")

try:
    with open("data/indexed_dataset.pkl", "rb") as f:
        store = pickle.load(f)
    docs = store["docs"]
    metadata = store["metadata"]
    vectorizer = store["vectorizer"]
    X = store["X"]
    print(f"Index loaded successfully. Number of documents: {len(docs):,}")
except FileNotFoundError:
    print("Error: The index file 'data/indexed_dataset.pkl' was not found. Please run the indexing steps first.")
    raise
```

```python


queries = []
for meta in metadata:
    doc_id = meta["doc_id"]
    title = meta["title"]
    for i in range(3):
        queries.append(
            {
                "query": f"قانون {title} در مورد بند {i+1} چه می‌گوید؟",
                "doc_id": doc_id,
                "title": title
            }
        )

print(f"\nTotal number of test queries: {len(queries):,}")
```

```python

records = []
for i, q in enumerate(queries):
    if i % 10 == 0:
        print(f"In processing query {i+1}/{len(queries)}...")

    try:
        out = run_pipeline(q["query"])
    except Exception as e:
        print(f"Error running run_pipeline for query {q['query']}: {e}")
        continue

    rec = {"query": q["query"], "doc_id": q["doc_id"], "title": q["title"]}

    for k, v in out["timings"].items():
        rec[k] = v

    found = any(c["doc_id"] == q["doc_id"] for c in out.get("contexts", []))
    rec["hit_at_3"] = int(found)

    records.append(rec)
```

```python

df = pd.DataFrame(records)
print("\n--- Sample performance data ---")
print(df.head())

agg = df.mean(numeric_only=True).to_dict()
print("\n--- Average timings (in seconds) ---")
print("Average timings (s):", {k: f"{v:.6f}" for k, v in agg.items() if k in df.columns and k != 'doc_id'})

```

```python
nodes = [
    "rewrite",
    "classify_intent",
    "extract_metadata",
    "context_retrieve",
    "rerank",
    "generate_answer",
]
avg_nodes = {n: agg.get(n, 0) for n in nodes}
bottleneck = max(avg_nodes, key=avg_nodes.get)
print(f"\n**Bottleneck node: {bottleneck} with an average of {avg_nodes[bottleneck]:.6f} seconds**")
print(f"Hit@3 rate: {agg.get('hit_at_3', 0) * 100:.2f}%")


df.to_csv("data/pipeline_timings.csv", index=False)
print("\nPipeline timings saved to 'data/pipeline_timings.csv'.")

```
