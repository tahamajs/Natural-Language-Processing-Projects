```python
from dataclasses import dataclass
from pathlib import Path
import pickle

@dataclass
class IndexManager:
    index_path: Path = Path("data/indexed_dataset.pkl")

    def build(self, docs: list, metadata: list, vectorizer=None):
        """ساخت ایندکس از اسناد و متادیتا"""
        from sklearn.feature_extraction.text import TfidfVectorizer
        if vectorizer is None:
            # استفاده از تنظیمات بهینه برای متن فارسی
            vectorizer = TfidfVectorizer(token_pattern=r"\b\w+\b", min_df=1)

        X = vectorizer.fit_transform(docs)
        self.store = {
            "docs": docs,
            "metadata": metadata,
            "vectorizer": vectorizer,
            "X": X
        }
        return self.store

    def save(self):
        """ذخیره دیتای ایندکس شده در فایل pkl"""
        if not hasattr(self, "store"):
            raise RuntimeError("ابتدا باید متد build را اجرا کنید تا دیتایی برای ذخیره وجود داشته باشد.")

        # مطمئن شدن از وجود پوشه data
        self.index_path.parent.mkdir(parents=True, exist_ok=True)

        with open(self.index_path, "wb") as f:
            pickle.dump(self.store, f)
        print(f"✅ ایندکس با موفقیت در مسیر {self.index_path} ذخیره شد.")

    def load(self):
        """بارگذاری ایندکس از فایل"""
        if not self.index_path.exists():
            raise FileNotFoundError(f"فایلی در مسیر {self.index_path} یافت نشد.")
        with open(self.index_path, "rb") as f:
            self.store = pickle.load(f)
        return self.store

# مقداردهی مجدد شیء مدیریت ایندکس
_index_mgr = IndexManager()
```

```python
import pdfplumber
import os
from hazm import Normalizer
from bidi.algorithm import get_display

# اگر اینها را ندارید: !pip install python-bidi arabic-reshaper
normalizer = Normalizer()

def fix_text_direction(text):
    """اصلاح جهت متن فارسی از حالت بصری به منطقی"""
    if not text: return ""
    # این تابع متن را از حالت 'نوناق' به 'قانون' برمی‌گرداند
    return get_display(text)

def chunk_text(text, chunk_size=1000, overlap=100):
    """تقسیم متن بزرگ به قطعات کوچک‌تر برای جستجوی دقیق‌تر"""
    chunks = []
    for i in range(0, len(text), chunk_size - overlap):
        chunks.append(text[i : i + chunk_size])
    return chunks

for filename in pdf_files:
    filepath = os.path.join(PDF_DIR, filename)
    try:
        with pdfplumber.open(filepath) as pdf:
            full_text = ""
            for page in pdf.pages:
                page_text = page.extract_text()
                if page_text:
                    # ۱. ابتدا اصلاح جهت متن (بسیار مهم)
                    fixed_text = fix_text_direction(page_text)
                    # ۲. سپس نرمال‌سازی (اصلاح ی/ک و نیم‌فاصله)
                    normalized_text = normalizer.normalize(fixed_text)
                    full_text += normalized_text + "\n"

            # ۳. تمیزکاری نهایی فواصل
            full_text = " ".join(full_text.split())

            if full_text:
                # ۴. قطعه‌بندی متن (Chunking)
                # این کار باعث می‌شود امتیاز شباهت (Score) به شدت بالا برود
                text_chunks = chunk_text(full_text)

                title = file_to_title.get(filename, filename.replace('.pdf', ''))

                for i, chunk in enumerate(text_chunks):
                    docs.append(chunk)
                    metadata.append({
                        "doc_id": doc_id_counter,
                        "title": f"{title} - Part {i+1}",
                        "filename": filename,
                        "chunk_index": i,
                        "length": len(chunk)
                    })
                    doc_id_counter += 1

                print(f"✅ {filename} processed and split into {len(text_chunks)} chunks.")

    except Exception as e:
        print(f"Error processing {filename}: {e}")
```

```python
# Robust extraction + cleaning (PyMuPDF + OCR fallback) and sentence-based chunking
# Installs (only if missing) and imports
try:
    import fitz
except Exception:
    # Use %pip to ensure it installs into the notebook kernel
    get_ipython().system("pip install PyMuPDF arabic-reshaper pdf2image pytesseract")
    import fitz

try:
    import arabic_reshaper
except Exception:
    get_ipython().system("pip install arabic-reshaper")
    import arabic_reshaper

try:
    import pytesseract
except Exception:
    get_ipython().system("pip install pytesseract pdf2image")
    import pytesseract

from pdf2image import convert_from_path
from bidi.algorithm import get_display
from hazm import Normalizer, sent_tokenize
import logging, warnings

# Suppress the noisy pdfminer FontBBox warnings (they don't always mean fatal errors)
logging.getLogger("pdfminer").setLevel(logging.ERROR)
logging.getLogger("pdfminer.pdffont").setLevel(logging.ERROR)
warnings.filterwarnings("ignore", message="Could get FontBBox from font descriptor")

_normalizer = Normalizer()

import re

def _clean_text(text: str) -> str:
    """Basic cleaning: remove URLs, IDS/font fragments, repeated slashes and control chars."""
    if not text:
        return ""
    # Remove long http/https tokens / broken URL-like fragments
    text = re.sub(r'https?://\S+|www\.\S+', ' ', text)
    # Remove patterns like IDS=..., &font=...
    text = re.sub(r'IDS=[^\s&]+', ' ', text, flags=re.I)
    text = re.sub(r'&\w+=\S+', ' ', text)
    # Replace multiple slashes and stray punctuation groups
    text = re.sub(r'[\x00-\x1f\r\t]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()
    return text


def _shape_persian(text: str) -> str:
    """Apply arabic_reshaper + bidi to get correct visual order for Persian text."""
    if not text:
        return ""
    try:
        reshaped = arabic_reshaper.reshape(text)
        return get_display(reshaped)
    except Exception:
        # Fallback to get_display of original text
        try:
            return get_display(text)
        except Exception:
            return text


def extract_text_with_fitz(filepath: str) -> str:
    """Extract text using PyMuPDF (fast, often cleaner than pdfminer)."""
    doc = fitz.open(filepath)
    pages = []
    for p in doc:
        txt = p.get_text("text")
        if txt:
            pages.append(txt)
    out = "\n".join(pages)
    out = _clean_text(out)
    return out


def ocr_pdf(filepath: str, dpi: int = 300) -> str:
    """Fallback OCR using pdf2image + pytesseract (set up Tesseract with Persian support first)."""
    pages = convert_from_path(filepath, dpi=dpi)
    texts = []
    for img in pages:
        try:
            txt = pytesseract.image_to_string(img, lang='fas')
        except Exception:
            txt = pytesseract.image_to_string(img)
        texts.append(txt)
    out = "\n".join(texts)
    out = _clean_text(out)
    return out


def chunk_by_sentences(text: str, max_chunk: int = 1000, overlap: int = 100):
    """Split text into sentence-aware chunks (hazm.sent_tokenize) so chunks stay coherent."""
    if not text:
        return []
    sents = sent_tokenize(text)
    chunks = []
    cur = ""
    for s in sents:
        s = s.strip()
        if not s:
            continue
        # if adding sentence keeps under budget -> append
        if len(cur) + len(s) + 1 <= max_chunk:
            cur = (cur + " " + s).strip() if cur else s
        else:
            chunks.append(cur)
            # start new chunk with the sentence (allow overlap: keep last N chars optional)
            cur = s
    if cur:
        chunks.append(cur)
    # Optionally add small overlaps
    return chunks


# Re-process PDFs with the new extraction pipeline
new_docs = []
new_metadata = []
new_id = 0

for filename in pdf_files:
    filepath = os.path.join(PDF_DIR, filename)

    text = extract_text_with_fitz(filepath)
    if not text or len(text) < 50:
        # fallback to OCR
        print(f"Falling back to OCR for {filename} (no usable text from fitz).")
        text = ocr_pdf(filepath)

    # Normalize and clean
    text = _normalizer.normalize(text)
    text = _clean_text(text)

    if not text:
        print(f"⚠️ No text could be extracted from {filename}")
        continue

    # Optional: shape for readability when inspecting in notebook
    # But store plain normalized text to avoid display-only transformations
    shaped_preview = _shape_persian(text[:500])

    # Sentence-based chunking
    chunks = chunk_by_sentences(text, max_chunk=1200)
    if not chunks:
        chunks = [text]

    title = file_to_title.get(filename, filename.replace('.pdf', ''))

    for i, chunk in enumerate(chunks):
        new_docs.append(chunk)
        new_metadata.append({
            "doc_id": new_id,
            "title": f"{title} - Part {i+1}",
            "filename": filename,
            "chunk_index": i,
            "length": len(chunk),
        })
        new_id += 1

    print(f"✅ {filename}: extracted {len(chunks)} chunks (preview: {shaped_preview[:200]})")

# Replace current docs/metadata with cleaned ones
docs = new_docs
metadata = new_metadata

# Rebuild index and persist
store = _index_mgr.build(docs, metadata)
vectorizer = store['vectorizer']
X = store['X']
_index_mgr.save()
print('✅ Extraction + chunking complete, index rebuilt.')
```

```python
# Load or build index using consolidated IndexManager helpers
INDEX_PATH = Path("data/indexed_dataset.pkl")
INDEX_PATH.parent.mkdir(parents=True, exist_ok=True)
# مطمئن شوید که ابتدا شیء منیجر ساخته شده است
if '_index_mgr' not in globals():
    _index_mgr = IndexManager()

def load_index():
    """فراخوانی متد لود از کلاس منیجر"""
    return _index_mgr.load()

def save_index():
    """فراخوانی متد ذخیره از کلاس منیجر"""
    return _index_mgr.save()

def load_index_or_build_from_memory(docs=None, metadata=None, vectorizer=None):
    """مدیریت هوشمند بارگذاری یا ساخت ایندکس"""
    return _index_mgr.ensure(docs=docs, metadata=metadata, vectorizer=vectorizer)
try:
    store = load_index()
    docs = store["docs"]
    metadata = store["metadata"]
    vectorizer = store["vectorizer"]
    X = store["X"]
    print("Index loaded successfully.")
except FileNotFoundError:
    # Attempt to build from in-memory data if available
    if "docs" in globals() and "metadata" in globals():
        store = load_index_or_build_from_memory(docs, metadata, globals().get("vectorizer"))
        docs = store["docs"]
        metadata = store["metadata"]
        vectorizer = store["vectorizer"]
        X = store["X"]
        print(f"Index file was missing; created it at {INDEX_PATH}.")
    else:
        raise FileNotFoundError(
            "data/indexed_dataset.pkl not found and cannot be built yet. "
            "Run the data loading + indexing cells first (the ones that build docs/metadata), then rerun this cell."
        )
```

```python

def rewrite_query(query: str) -> str:
    q = query.strip()
    q = re.sub(r"\s+", " ", q)
    return q

```

```python
GREET_KEYWORDS = ["سلام", "درود", "صبح بخیر", "وقت بخیر" , "خوش آمدید", "خوش اومدید", "سلامتی"]
ABUSE_KEYWORDS = ["مسخره", "بی‌ادب", "احمق", "نادان", "بی‌فکر", "بی‌عقل", "بی‌شعور" , "بی‌خرد"]
```

```python

def classify_intent(query: str) -> dict:
    q = query.lower()
    for w in GREET_KEYWORDS:
        if w in q:
            return {"intent": "greeting", "response": "سلام! چطور می‌توانم کمک کنم؟"}
    for w in ABUSE_KEYWORDS:
        if w in q:
            return {
                "intent": "abuse",
                "response": "من اینجا هستم تا کمک کنم؛ لطفاً مودبانه سوالاتتان را مطرح کنید.",
            }
    return {"intent": "law_question"}

```

```python
def extract_metadata(rewritten_query: str) -> dict:
    m = {}
    years = re.findall(r"13\d{2}|14\d{2}|\d{4}", rewritten_query)
    if years:
        try:
            m["year"] = int(years[0])
        except:
            pass

    keywords = []
    if "کار" in rewritten_query:
        keywords.append("کار")
    if "مدنی" in rewritten_query:
        keywords.append("مدنی")
    if "جرایم" in rewritten_query or "رایانه" in rewritten_query:
        keywords.append("جرایم")
    if keywords:
        m["keywords"] = keywords
    return m
```

```python
def context_retrieve(rewritten_query: str, metadata_filter: dict = None, K: int = 10):
    # ۱. تبدیل کوئری به بردار
    q_vec = vectorizer.transform([rewritten_query])
    sims = cosine_similarity(q_vec, X).flatten()

    # ۲. اعمال فیلتر قبل از انتخاب نهایی
    eligible_indices = []

    for i in range(len(docs)):
        meta = metadata[i]
        is_ok = True

        if metadata_filter:
            # چک کردن سال
            if "year" in metadata_filter and str(meta.get("year")) != str(metadata_filter["year"]):
                is_ok = False

            # چک کردن کلمات کلیدی (فقط در عنوان برای سرعت بیشتر، یا در متن با احتیاط)
            if is_ok and "keywords" in metadata_filter:
                found_k = any(k in meta.get("title", "") for k in metadata_filter["keywords"])
                if not found_k:
                    is_ok = False

        if is_ok:
            eligible_indices.append(i)

    # ۳. رتبه‌بندی فقط روی اسنادی که فیلتر را رد کرده‌اند
    if not eligible_indices:
        return []

    # استخراج نمرات شباهت برای اسناد تایید شده
    filtered_sims = [(i, sims[i]) for i in eligible_indices]
    # مرتب‌سازی بر اساس نمره
    filtered_sims.sort(key=lambda x: x[1], reverse=True)

    results = filtered_sims[:K]

    # ۴. ساخت خروجی نهایی
    contexts = []
    for i, score in results:
        contexts.append({
            "doc_id": i,
            "text": docs[i],
            "metadata": metadata[i],
            "score": float(score)
        })
    return contexts
```

```python
from hazm import WordTokenizer, Normalizer
import re

# تعریف یک‌باره برای سرعت بیشتر
_tokenizer = WordTokenizer()
_normalizer = Normalizer()

def _token_set(text: str):
    if not text:
        return set()
    # ۱. نرمال‌سازی (یکسان کردن ی/ک و فواصل)
    normalized = _normalizer.normalize(text.lower())
    # ۲. استفاده از توکنایزر فارسی به جای Regex ساده
    tokens = _tokenizer.tokenize(normalized)
    # ۳. حذف علائم نگارشی احتمالی از داخل توکن‌ها
    return {re.sub(r'[^\w\s]', '', t) for t in tokens if t.strip()}

def rerank(contexts, query: str = None, top_n=3, relevance_threshold=0.01):
    if not contexts:
        return []

    q_tokens = _token_set(query) if query else set()

    for c in contexts:
        text_tokens = _token_set(c.get("text", ""))
        overlap = 0.0

        if q_tokens and text_tokens:
            # محاسبه Jaccard Similarity (میزان اشتراک کلمات)
            intersection = q_tokens.intersection(text_tokens)
            union = q_tokens.union(text_tokens)
            overlap = len(intersection) / len(union)

        # ترکیب امتیاز برداری (TF-IDF) و امتیاز واژگانی (Overlap)
        # ضریب ۰.۷ و ۰.۳ ضریب‌های استاندارد و خوبی هستند
        c["final_score"] = 0.7 * float(c.get("score", 0.0)) + 0.3 * overlap

    # مرتب‌سازی بر اساس امتیاز نهایی
    contexts.sort(key=lambda x: x.get("final_score", 0.0), reverse=True)
    top = contexts[:top_n]

    # چک کردن آستانه اعتبار
    if not top or all(c.get("final_score", 0.0) < relevance_threshold for c in top):
        print(f"⚠️ هشدار: هیچ محتوایی با آستانه {relevance_threshold} مطابقت نداشت.")
        return []

    return top
```

```python


def generate_answer(rewritten_query: str, contexts):
    if not contexts:
        return "متاسفانه نتوانستم پاسخی مرتبط در اسناد پیدا کنم. لطفاً سوال را با کلمات کلیدی دقیق‌تر مطرح کنید."

    parts = []
    for c in contexts:
        snippet = c["text"][:400] + "..."
        source = c["metadata"]["title"]
        parts.append(f"منبع: {source} (امتیاز: {c['score']:.2f})\nمتن: {snippet}")

    answer_body = "\n\n---\n\n".join(parts)
    final = f"پاسخ پیشنهادی (یافته شده در منابع):\n\n{answer_body}"
    return final
```

```python
def run_pipeline(query: str, K=10, top_n=3):
    timings = {}
    t0 = time.time()

    # 1. بازنویسی (با حفظ سوال اصلی برای رنکنیگ)
    t = time.time()
    try:
        rewritten = rewrite_query(query)
    except Exception:
        rewritten = query
    timings["rewrite"] = time.time() - t

    # 2. تشخیص قصد
    t = time.time()
    intent = classify_intent(rewritten)
    timings["classify_intent"] = time.time() - t

    if intent.get("intent") != "law_question":
        timings["total"] = time.time() - t0
        return {"answer": intent.get("response"), "timings": timings, "contexts": []}

    # 3. بازیابی با متادیتا
    t = time.time()
    meta_filter = extract_metadata(rewritten)
    contexts = context_retrieve(rewritten, meta_filter, K=K)

    # 4. رنکینگ مجدد (استفاده از هر دو کوئری برای دقت بیشتر)
    top = rerank(contexts, query=f"{query} {rewritten}", top_n=top_n)

    # Fallback Mechanism
    if not top or len(top) == 0:
        print("🔍 No relevant context with metadata. Widening search...")
        contexts = context_retrieve(rewritten, None, K=min(len(docs), K * 2))
        top = rerank(contexts, query=rewritten, top_n=top_n)

    timings["retrieval_and_rerank"] = time.time() - t

    # 5. تولید پاسخ
    t = time.time()
    if top and len(top) > 0:
        answer = generate_answer(rewritten, top)
    else:
        answer = "با عرض پوزش، سندی برای پاسخ به سوال شما پیدا نشد."

    timings["generate_answer"] = time.time() - t
    timings["total"] = time.time() - t0

    return {"answer": answer, "timings": timings, "contexts": top}
```

```python
import numpy as np
from collections import defaultdict

THRESHOLD = 0.01
K = 10
TOP_N = 3

diag = {
    'num_queries': len(queries),
    'no_top_after_rerank': 0,
    'raw_top_below_thresh': 0,
    'final_top_below_thresh': 0,
    'raw_scores': [],
    'final_scores': [],
}

failing_examples = []

for i, q in enumerate(queries):
    qtext = q['query']
    qvec = vectorizer.transform([qtext])
    sims = cosine_similarity(qvec, X).flatten()
    if np.all(np.isnan(sims)):
        raw_top = 0.0
    else:
        raw_top = float(np.max(sims))
    diag['raw_scores'].append(raw_top)

    # retrieval with metadata first
    meta = extract_metadata(qtext)
    contexts = context_retrieve(qtext, meta, K=K)
    top = rerank(contexts, query=qtext, top_n=TOP_N)

    if not top:
        # retry without metadata
        contexts = context_retrieve(qtext, None, K=min(len(docs), K * 2))
        top = rerank(contexts, query=qtext, top_n=TOP_N)

    if not top:
        diag['no_top_after_rerank'] += 1
        final_top = 0.0
    else:
        final_top = float(top[0].get('final_score', top[0].get('score', 0.0)))
    diag['final_scores'].append(final_top)

    if raw_top < THRESHOLD:
        diag['raw_top_below_thresh'] += 1
    if final_top < THRESHOLD:
        diag['final_top_below_thresh'] += 1
        if len(failing_examples) < 8:
            example = {
                'index': i,
                'query': qtext,
                'raw_top': raw_top,
                'final_top': final_top,
                'top_context': top[0] if top else (contexts[0] if contexts else None),
            }
            failing_examples.append(example)

    if i % 50 == 0:
        print(f"In processing query {i+1}/{len(queries)}...")

# Summary
print('\n--- Diagnostic summary ---')
print(f"Number of queries: {diag['num_queries']}")
print(f"Queries with no top after rerank: {diag['no_top_after_rerank']} ({diag['no_top_after_rerank']/diag['num_queries']:.2%})")
print(f"Queries where raw top < {THRESHOLD}: {diag['raw_top_below_thresh']} ({diag['raw_top_below_thresh']/diag['num_queries']:.2%})")
print(f"Queries where final top < {THRESHOLD}: {diag['final_top_below_thresh']} ({diag['final_top_below_thresh']/diag['num_queries']:.2%})")

# Show distribution stats
import numpy as _n
print('\nRaw top similarity: mean={:.4f}, median={:.4f}, max={:.4f}'.format(_n.mean(diag['raw_scores']), _n.median(diag['raw_scores']), _n.max(diag['raw_scores'])))
print('Final top score: mean={:.4f}, median={:.4f}, max={:.4f}'.format(_n.mean(diag['final_scores']), _n.median(diag['final_scores']), _n.max(diag['final_scores'])))

# Show failing examples to inspect content
print('\n--- Sample failing queries (final score < threshold) ---')
for ex in failing_examples:
    print('\nQuery (index {}): {}'.format(ex['index'], ex['query']))
    print(f"  raw_top={ex['raw_top']:.6f}, final_top={ex['final_top']:.6f}")
    ctx = ex['top_context']
    if ctx:
        title = ctx['metadata']['title'] if 'metadata' in ctx and ctx['metadata'] else ctx.get('metadata')
        snippet = ctx['text'][:300].replace('\n', ' ')
        print(f"  Top context title: {title}")
        print(f"  Snippet: {snippet[:300]}")
    else:
        print('  No context retrieved at all.')

# Suggest quick mitigations
print('\n--- Quick mitigation suggestions ---')
print('1) Lower the rerank relevance threshold (e.g., 0.0 or 0.001) — this will reduce false negatives.')
print('2) Increase K (retrieval window) so reranker has more candidates (e.g., K=20 or 50).')
print('3) Improve chunking — try slightly larger sentence chunks (e.g., 1200-2000 chars) or different overlap.')
print('4) Clean out noisy artifacts (URLs, IDS=, font params) — several top contexts contain such fragments.')
print('5) Replace TF-IDF with embedding-based retriever (Faiss/LanceDB + multilingual E5) for semantic matches.')
print('\nIf you want I can: 1) run these diagnostics now (done) 2) try lowering threshold and re-evaluate hit@3, 3) re-run with larger K, or 4) implement an embedding-based retriever. Tell me which to try next.')
```

```python
# Quick grid: see impact of thresholds and K on final selection hit rate (fast diagnostic)
from itertools import product

thresholds = [0.01, 0.001, 0.0]
Ks = [10, 20, 50]
TOP_N = 3
results_grid = []

for K_val, thr in product(Ks, thresholds):
    success = 0
    for q in queries:
        qtext = q['query']
        contexts = context_retrieve(qtext, extract_metadata(qtext), K=K_val)
        top = rerank(contexts, query=qtext, top_n=TOP_N, relevance_threshold=thr)
        if not top:
            contexts = context_retrieve(qtext, None, K=min(len(docs), K_val * 2))
            top = rerank(contexts, query=qtext, top_n=TOP_N, relevance_threshold=thr)
        if top and len(top) > 0 and top[0].get('final_score', 0.0) >= thr:
            # consider this a success (found relevant at top-N)
            success += 1
    hit_rate = success / len(queries)
    results_grid.append((K_val, thr, success, hit_rate))
    print(f"K={K_val}, thr={thr}: success={success}/{len(queries)} => hit_rate={hit_rate:.2%}")

print('\nGrid search complete.')

# show combinations where performance improves significantly
sorted_grid = sorted(results_grid, key=lambda x: x[3], reverse=True)
print('\nTop 5 settings:')
for r in sorted_grid[:5]:
    print(r)
```

```python
# --- RAGAS evaluation models (API-based) ---
# This section uses OpenAI API via langchain_openai.
# Make sure you ran the earlier “API bootstrap” cell and have OPENAI_API_KEY set.

OPENAI_API_KEY = os.environ.get("OPENAI_API_KEY")
if not OPENAI_API_KEY:
    print("OPENAI_API_KEY is not set. Skipping RAGAS evaluation that requires an LLM. Set the env var or keys.json to enable full evaluation.")
    eval_enabled = False
    eval_llm = None
    eval_embeddings = None
else:
    eval_enabled = True
    EVAL_OPENAI_MODEL = os.environ.get("EVAL_OPENAI_MODEL", "gpt-4o-mini")
    EVAL_OPENAI_EMBEDDING_MODEL = os.environ.get("EVAL_OPENAI_EMBEDDING_MODEL", "text-embedding-3-small")

    eval_llm = ChatOpenAI(model=EVAL_OPENAI_MODEL, temperature=0)
    eval_embeddings = OpenAIEmbeddings(model=EVAL_OPENAI_EMBEDDING_MODEL)

    print(f"✅ Using OpenAI for RAGAS eval: llm={EVAL_OPENAI_MODEL}, emb={EVAL_OPENAI_EMBEDDING_MODEL}")
```

```python
eval_records = []

for meta, doc in zip(metadata, docs):
    # انتخاب ۵۰۰ کاراکتر اول به عنوان مرجع پاسخ
    gold_content = doc[:500]

    # تعریف سوالات به زبان فارسی متناسب با عنوان هر سند
    questions = [
        (f"موضوع اصلی بخش {meta['title']} چیست؟", gold_content),
        (f"چه جزئیاتی در مورد {meta['title']} در متن آمده است؟", gold_content),
        (f"یک خلاصه کوتاه از {meta['title']} ارائه دهید.", gold_content),
    ]

    for q, gold in questions:
        eval_records.append({
            "doc_id": meta["doc_id"],
            "question": q,
            "ground_truth": gold
        })
```

```python
data_for_ragas = {
    "question": [],
    "answer": [],
    "contexts": [],
    "ground_truth": [],
    "doc_id": [],
}

print(f"Running pipeline for {len(eval_records)} queries...")

for record in eval_records:
    out = run_pipeline(record["question"]) or {}

    contexts = out.get("contexts", []) or []
    retrieved_contexts = [c.get("text", "") for c in contexts if isinstance(c, dict)]
    retrieved_contexts = [c for c in retrieved_contexts if c]

    data_for_ragas["question"].append(record.get("question", ""))
    data_for_ragas["answer"].append(out.get("answer", ""))
    data_for_ragas["contexts"].append(retrieved_contexts)
    data_for_ragas["ground_truth"].append(record.get("ground_truth", ""))
    data_for_ragas["doc_id"].append(record.get("doc_id"))
```

```python
hf_dataset = Dataset.from_dict(data_for_ragas)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def test_retrieval(query, vectorizer, X, docs, top_k=1):
    query_vec = vectorizer.transform([query])

    similarities = cosine_similarity(query_vec, X).flatten()

    best_idx = similarities.argsort()[-top_k:][::-1]

    print(f"Query: {query}")
    for idx in best_idx:
        print(f"Score: {similarities[idx]:.4f}")
        print(f"Content: {docs[idx][:200]}...") # نمایش بخشی از متن پیدا شده
        print("-" * 30)

# تست: یک سوال که کلماتش در یکی از سندها هست بپرسید
test_retrieval("موضوع اصلی سند اول چیست؟", vectorizer, X, docs)
```
