```python
from __future__ import annotations

import json
import os
from pathlib import Path

INDEX_PATH = Path("data/indexed_dataset.pkl")
INDEX_PATH.parent.mkdir(parents=True, exist_ok=True)

KEY_NAMES = ["OPENAI_API_KEY", "OPENAI_API_BASE"]

def _mask(val: str | None, keep: int = 4) -> str:
    if not val:
        return "not set"
    if len(val) <= keep:
        return "***"
    return val[:keep] + "***" + val[-keep:]

def load_keys_from_files() -> None:
    """Load secrets from keys.json/.keys.json if present.

    Expected format: {"OPENAI_API_KEY": "...", "OPENAI_API_BASE": "..."}
    """
    for name in ("keys.json", ".keys.json"):
        path = Path(name)
        if not path.exists():
            continue
        try:
            data = json.loads(path.read_text(encoding="utf-8"))
        except Exception:
            continue
        if not isinstance(data, dict):
            continue
        for k in KEY_NAMES:
            if os.environ.get(k):
                continue
            v = data.get(k)
            if v:
                os.environ[k] = str(v)

load_keys_from_files()

OPENAI_API_KEY = os.environ.get("OPENAI_API_KEY")
OPENAI_API_BASE = os.environ.get("OPENAI_API_BASE", "https://api.openai.com/v1")

print("Key status (masked):")
print(f"  OPENAI_API_BASE: {OPENAI_API_BASE}")
print(f"  OPENAI_API_KEY: {_mask(OPENAI_API_KEY)}")
```

```python
from dataclasses import dataclass
from pathlib import Path
import pickle
import re
from hazm import Normalizer as HazmNormalizer

# نرمال‌ساز واحد برای فارسی (برای ایندکس و کوئری)
# توجه: نسخه‌های مختلف hazm پارامترهای سازنده متفاوتی دارند؛ پس ساده و سازگار نگه می‌داریم.
_FA_INDEX_NORMALIZER = HazmNormalizer()

def fa_normalize(text: str) -> str:
    if not text:
        return ""
    text = str(text)
    # یکسان‌سازی چند نویسه‌ی رایج
    text = (
        text.replace("ي", "ی")
            .replace("ك", "ک")
            .replace("ﯿ", "ی")
            .replace("ﮑ", "ک")
    )
    text = _FA_INDEX_NORMALIZER.normalize(text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

@dataclass
class IndexManager:
    index_path: Path = Path("data/indexed_dataset.pkl")

    def build(self, docs: list, metadata: list, vectorizer=None):
        """ساخت ایندکس از اسناد و متادیتا (سازگار با فارسی)"""
        from sklearn.feature_extraction.text import TfidfVectorizer
        if vectorizer is None:
            # برای فارسی، char_wb n-gram معمولاً از word-token بهتر جواب می‌دهد (نیم‌فاصله/چسبندگی/املاء)
            vectorizer = TfidfVectorizer(
                preprocessor=fa_normalize,
                analyzer="char_wb",
                ngram_range=(3, 5),
                min_df=1,
                sublinear_tf=True,
                lowercase=False,
            )

        X = vectorizer.fit_transform(docs)
        self.store = {
            "docs": docs,
            "metadata": metadata,
            "vectorizer": vectorizer,
            "X": X,
        }
        return self.store

    def save(self):
        """ذخیره دیتای ایندکس شده در فایل pkl"""
        if not hasattr(self, "store"):
            raise RuntimeError("ابتدا باید متد build را اجرا کنید تا دیتایی برای ذخیره وجود داشته باشد.")

        self.index_path.parent.mkdir(parents=True, exist_ok=True)

        with open(self.index_path, "wb") as f:
            pickle.dump(self.store, f)
        print(f"✅ ایندکس با موفقیت در مسیر {self.index_path} ذخیره شد.")

    def load(self):
        """بارگذاری ایندکس از فایل"""
        if not self.index_path.exists():
            raise FileNotFoundError(f"فایلی در مسیر {self.index_path} یافت نشد.")
        with open(self.index_path, "rb") as f:
            self.store = pickle.load(f)
        return self.store

    def ensure(self, docs=None, metadata=None, vectorizer=None):
        """اگر ایندکس هست لود کن، اگر نیست و داده تو حافظه هست بساز."""
        try:
            return self.load()
        except FileNotFoundError:
            if docs is None or metadata is None:
                raise
            return self.build(docs, metadata, vectorizer=vectorizer)

# مقداردهی مجدد شیء مدیریت ایندکس
_index_mgr = IndexManager()
```

```python
import os
import fitz  # PyMuPDF
import pytesseract
from PIL import Image
import io
from hazm import Normalizer, sent_tokenize

normalizer = Normalizer()

def extract_text_with_ocr_fixed(pdf_path):
    """استخراج متن با استفاده از تبدیل صفحه به عکس (بدون نیاز به Poppler)"""
    print(f"🧐 در حال پردازش تصویری: {os.path.basename(pdf_path)}")
    try:
        doc = fitz.open(pdf_path)
        full_text = ""
        
        for page_index in range(len(doc)):
            page = doc.load_page(page_index)
            # تبدیل صفحه به عکس با کیفیت بالا (DPI 300)
            pix = page.get_pixmap(matrix=fitz.Matrix(2, 2)) 
            img_data = pix.tobytes("png")
            img = Image.open(io.BytesIO(img_data))
            
            # استخراج متن فارسی از تصویر
            page_text = pytesseract.image_to_string(img, lang='fas')
            full_text += page_text + "\n"
        
        doc.close()
        return full_text
    except Exception as e:
        print(f"❌ خطا در پردازش OCR فایل {pdf_path}: {e}")
        return ""

def chunk_by_sentences(text: str, max_chunk: int = 1000) -> list[str]:
    """تکه‌بندی هوشمند بر اساس جملات فارسی"""
    if not text: return []
    text = normalizer.normalize(text)
    sents = sent_tokenize(text)
    
    chunks = []
    cur = ""
    for s in sents:
        if len(cur) + len(s) + 1 <= max_chunk:
            cur = (cur + " " + s).strip() if cur else s
        else:
            if cur: chunks.append(cur)
            cur = s
    if cur: chunks.append(cur)
    return chunks

# --- فرآیند اصلی اجرا ---
docs = []
metadata = []
doc_id_counter = 0

for filename in pdf_files:
    filepath = os.path.join(PDF_DIR, filename)
    raw_content = extract_text_with_ocr_fixed(filepath)
    
    if raw_content.strip():
        text_chunks = chunk_by_sentences(raw_content, max_chunk=1000)
        title = file_to_title.get(filename, filename.replace('.pdf', ''))

        for i, chunk in enumerate(text_chunks):
            docs.append(chunk)
            metadata.append({
                "doc_id": doc_id_counter,
                "title": f"{title} - Part {i+1}",
                "filename": filename,
                "chunk_index": i
            })
            doc_id_counter += 1
        print(f"✅ {filename} با موفقیت به {len(text_chunks)} تکه تقسیم شد.")
    else:
        print(f"⚠️ هشدار: هیچ متنی از {filename} استخراج نشد.")
```

```python
INDEX_PATH = Path("data/indexed_dataset.pkl")
INDEX_PATH.parent.mkdir(parents=True, exist_ok=True)
# مطمئن شوید که ابتدا شیء منیجر ساخته شده است
if '_index_mgr' not in globals():
    _index_mgr = IndexManager()

def load_index():
    """فراخوانی متد لود از کلاس منیجر"""
    return _index_mgr.load()

def save_index():
    """فراخوانی متد ذخیره از کلاس منیجر"""
    return _index_mgr.save()

def load_index_or_build_from_memory(docs=None, metadata=None, vectorizer=None):
    """مدیریت هوشمند بارگذاری یا ساخت ایندکس"""
    return _index_mgr.ensure(docs=docs, metadata=metadata, vectorizer=vectorizer)
try:
    store = load_index()
    docs = store["docs"]
    metadata = store["metadata"]
    vectorizer = store["vectorizer"]
    X = store["X"]
    print("Index loaded successfully.")
except FileNotFoundError:
    # Attempt to build from in-memory data if available
    if "docs" in globals() and "metadata" in globals():
        store = load_index_or_build_from_memory(docs, metadata, globals().get("vectorizer"))
        docs = store["docs"]
        metadata = store["metadata"]
        vectorizer = store["vectorizer"]
        X = store["X"]
        print(f"Index file was missing; created it at {INDEX_PATH}.")
    else:
        raise FileNotFoundError(
            "data/indexed_dataset.pkl not found and cannot be built yet. "
            "Run the data loading + indexing cells first (the ones that build docs/metadata), then rerun this cell."
        )
```

```python

def rewrite_query(query: str) -> str:
    q = query.strip()
    q = re.sub(r"\s+", " ", q)
    return q

```

```python
GREET_KEYWORDS = ["سلام", "درود", "صبح بخیر", "وقت بخیر" , "خوش آمدید", "خوش اومدید", "سلامتی"]
ABUSE_KEYWORDS = ["مسخره", "بی‌ادب", "احمق", "نادان", "بی‌فکر", "بی‌عقل", "بی‌شعور" , "بی‌خرد"]
```

```python

def classify_intent(query: str) -> dict:
    q = query.lower()
    for w in GREET_KEYWORDS:
        if w in q:
            return {"intent": "greeting", "response": "سلام! چطور می‌توانم کمک کنم؟"}
    for w in ABUSE_KEYWORDS:
        if w in q:
            return {
                "intent": "abuse",
                "response": "من اینجا هستم تا کمک کنم؛ لطفاً مودبانه سوالاتتان را مطرح کنید.",
            }
    return {"intent": "law_question"}

```

```python
def extract_metadata(rewritten_query: str) -> dict:
    m = {}
    years = re.findall(r"13\d{2}|14\d{2}|\d{4}", rewritten_query)
    if years:
        try:
            m["year"] = int(years[0])
        except:
            pass

    keywords = []
    if "کار" in rewritten_query:
        keywords.append("کار")
    if "مدنی" in rewritten_query:
        keywords.append("مدنی")
    if "جرایم" in rewritten_query or "رایانه" in rewritten_query:
        keywords.append("جرایم")
    if keywords:
        m["keywords"] = keywords
    return m
```

```python
def context_retrieve(rewritten_query: str, metadata_filter: dict = None, K: int = 10):
    # ۱. تبدیل کوئری به بردار
    q_vec = vectorizer.transform([rewritten_query])
    sims = cosine_similarity(q_vec, X).flatten()

    # ۲. اعمال فیلتر قبل از انتخاب نهایی
    eligible_indices = []

    for i in range(len(docs)):
        meta = metadata[i]
        is_ok = True

        if metadata_filter:
            # چک کردن سال
            if "year" in metadata_filter and str(meta.get("year")) != str(metadata_filter["year"]):
                is_ok = False

            # چک کردن کلمات کلیدی (فقط در عنوان برای سرعت بیشتر، یا در متن با احتیاط)
            if is_ok and "keywords" in metadata_filter:
                found_k = any(k in meta.get("title", "") for k in metadata_filter["keywords"])
                if not found_k:
                    is_ok = False

        if is_ok:
            eligible_indices.append(i)

    # ۳. رتبه‌بندی فقط روی اسنادی که فیلتر را رد کرده‌اند
    if not eligible_indices:
        return []

    # استخراج نمرات شباهت برای اسناد تایید شده
    filtered_sims = [(i, sims[i]) for i in eligible_indices]
    # مرتب‌سازی بر اساس نمره
    filtered_sims.sort(key=lambda x: x[1], reverse=True)

    results = filtered_sims[:K]

    # ۴. ساخت خروجی نهایی
    contexts = []
    for i, score in results:
        contexts.append({
            "doc_id": i,
            "text": docs[i],
            "metadata": metadata[i],
            "score": float(score)
        })
    return contexts
```

```python
from hazm import WordTokenizer, Normalizer
import re

# تعریف یک‌باره برای سرعت بیشتر
_tokenizer = WordTokenizer()
_normalizer = Normalizer()

def _token_set(text: str):
    if not text:
        return set()
    # نرمال‌سازی (یکسان کردن ی/ک و فواصل)
    normalized = _normalizer.normalize(str(text).lower())
    tokens = _tokenizer.tokenize(normalized)
    # حذف علائم نگارشی احتمالی از داخل توکن‌ها
    cleaned = []
    for t in tokens:
        t = re.sub(r"[^\w\s]", "", t)
        t = t.strip()
        if t:
            cleaned.append(t)
    return set(cleaned)

def rerank(contexts, query: str = None, top_n=3, relevance_threshold=0.01, verbose: bool = False):
    if not contexts:
        return []

    q_tokens = _token_set(query) if query else set()

    for c in contexts:
        text_tokens = _token_set(c.get("text", ""))
        overlap = 0.0
        if q_tokens and text_tokens:
            # Jaccard Similarity
            intersection = q_tokens.intersection(text_tokens)
            union = q_tokens.union(text_tokens)
            overlap = len(intersection) / max(len(union), 1)

        # ترکیب امتیاز برداری (TF-IDF) و امتیاز واژگانی (Overlap)
        c["final_score"] = 0.7 * float(c.get("score", 0.0)) + 0.3 * overlap

    contexts.sort(key=lambda x: x.get("final_score", 0.0), reverse=True)
    top = contexts[:top_n]

    if not top or all(c.get("final_score", 0.0) < relevance_threshold for c in top):
        if verbose:
            print(f"⚠️ هشدار: هیچ محتوایی با آستانه {relevance_threshold} مطابقت نداشت.")
        return []

    return top
```

```python


def generate_answer(rewritten_query: str, contexts):
    if not contexts:
        return "متاسفانه نتوانستم پاسخی مرتبط در اسناد پیدا کنم. لطفاً سوال را با کلمات کلیدی دقیق‌تر مطرح کنید."

    parts = []
    for c in contexts:
        snippet = c["text"][:400] + "..."
        source = c["metadata"]["title"]
        parts.append(f"منبع: {source} (امتیاز: {c['score']:.2f})\nمتن: {snippet}")

    answer_body = "\n\n---\n\n".join(parts)
    final = f"پاسخ پیشنهادی (یافته شده در منابع):\n\n{answer_body}"
    return final
```

```python
def run_pipeline(query: str, K=10, top_n=3, verbose: bool = False):
    timings = {}
    t0 = time.time()

    # 1. بازنویسی (با حفظ سوال اصلی برای رنکنیگ)
    t = time.time()
    try:
        rewritten = rewrite_query(query)
    except Exception:
        rewritten = query
    timings["rewrite"] = time.time() - t

    # 2. تشخیص قصد
    t = time.time()
    intent = classify_intent(rewritten)
    timings["classify_intent"] = time.time() - t

    if intent.get("intent") != "law_question":
        timings["total"] = time.time() - t0
        return {"answer": intent.get("response"), "timings": timings, "contexts": []}

    # 3. بازیابی با متادیتا
    t = time.time()
    meta_filter = extract_metadata(rewritten)
    contexts = context_retrieve(rewritten, meta_filter, K=K)

    # 4. رنکینگ مجدد
    top = rerank(contexts, query=f"{query} {rewritten}", top_n=top_n, verbose=verbose)

    # Fallback Mechanism (only one retry)
    if not top:
        if verbose:
            print("🔍 No relevant context with metadata. Widening search...")
        contexts = context_retrieve(rewritten, None, K=min(len(docs), K * 2))
        top = rerank(contexts, query=rewritten, top_n=top_n, verbose=verbose)

    timings["retrieval_and_rerank"] = time.time() - t

    # 5. تولید پاسخ
    t = time.time()
    if top:
        answer = generate_answer(rewritten, top)
    else:
        answer = "با عرض پوزش، سندی برای پاسخ به سوال شما پیدا نشد."

    timings["generate_answer"] = time.time() - t
    timings["total"] = time.time() - t0

    return {"answer": answer, "timings": timings, "contexts": top}
```

```python
import re

def final_polish_text(text):
    if not text:
        return ""
    
    # ۱. اصلاح جداافتادگی "می" در افعال
    text = re.sub(r'\bیم\b', 'می', text) # تبدیل "یم" به "می"
    
    # ۲. اصلاح کلمات کلیدی حقوقی که معمولاً برعکس می‌شوند
    corrections = {
        'دشاب': 'باشد',
        'دﯾامن': 'نماید',
        'دقعنم': 'منعقد',
        'قباس': 'سابق',
        'جدﯾد': 'جدید',
        'میاق': 'قایم',
        'تیعطق': 'قطعیت',
        'یقوقح': 'حقوقی',
        'هب': 'به',
        'رثوم': 'موثر',
        'دنب': 'بند',
        'لیکشت': 'تشکیل'
    }
    
    for wrong, right in corrections.items():
        text = text.replace(wrong, right)
        
    # ۳. منطق هوشمند برای برگرداندن کلمات معکوس (اگر کلمه در دیکشنری فارسی نبود)
    words = text.split()
    fixed_words = []
    for w in words:
        # اگر کلمه با حروف فارسی است و طول آن بیش از ۲ است
        if any(u'\u0600' <= c <= u'\u06FF' for c in w) and len(w) > 2:
            # یک تست ساده: اگر معکوس کلمه رایج‌تر است (مثلاً 'دشاب' -> 'باشد')
            # اینجا می‌توان از یک کتابخانه لغت‌نامه استفاده کرد، اما دستی هم جواب می‌دهد
            if w in ['یم', 'دنشاب', 'دﯾامن', 'دقعنم']: # مثال‌های پرتکرار شما
                fixed_words.append(w[::-1])
            else:
                fixed_words.append(w)
        else:
            fixed_words.append(w)
            
    return " ".join(fixed_words)
```

```python
# Quick grid: see impact of thresholds and K on final selection hit rate (fast diagnostic)
from itertools import product

thresholds = [0.01, 0.001, 0.0]
Ks = [10, 20, 50]
TOP_N = 3
results_grid = []

for K_val, thr in product(Ks, thresholds):
    success = 0
    for q in queries:
        qtext = q['query']
        contexts = context_retrieve(qtext, extract_metadata(qtext), K=K_val)
        top = rerank(contexts, query=qtext, top_n=TOP_N, relevance_threshold=thr)
        if not top:
            contexts = context_retrieve(qtext, None, K=min(len(docs), K_val * 2))
            top = rerank(contexts, query=qtext, top_n=TOP_N, relevance_threshold=thr)
        if top and len(top) > 0 and top[0].get('final_score', 0.0) >= thr:
            # consider this a success (found relevant at top-N)
            success += 1
    hit_rate = success / len(queries)
    results_grid.append((K_val, thr, success, hit_rate))
    print(f"K={K_val}, thr={thr}: success={success}/{len(queries)} => hit_rate={hit_rate:.2%}")

print('\nGrid search complete.')

# show combinations where performance improves significantly
sorted_grid = sorted(results_grid, key=lambda x: x[3], reverse=True)
print('\nTop 5 settings:')
for r in sorted_grid[:5]:
    print(r)
```

```python
import os
import pandas as pd
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from ragas.metrics import faithfulness, answer_relevancy
from ragas import evaluate
from ragas.llms import LangchainLLMWrapper
from ragas.embeddings import LangchainEmbeddingsWrapper

# بررسی وجود کلید API
OPENAI_API_KEY = os.environ.get("OPENAI_API_KEY")

if not OPENAI_API_KEY:
    print("❌ OPENAI_API_KEY is not set. Evaluation disabled.")
    eval_enabled = False
else:
    eval_enabled = True
    EVAL_OPENAI_MODEL = os.environ.get("EVAL_OPENAI_MODEL", "gpt-4o-mini")
    EVAL_OPENAI_EMBEDDING_MODEL = os.environ.get("EVAL_OPENAI_EMBEDDING_MODEL", "text-embedding-3-small")

    # 1. تعریف مدل پایه با پارامترهای استاندارد
    base_llm = ChatOpenAI(
        model=EVAL_OPENAI_MODEL, 
        temperature=0,
    )

    # 2. تعریف مدل Embeddings
    base_embeddings = OpenAIEmbeddings(model=EVAL_OPENAI_EMBEDDING_MODEL)

    # 3. استفاده از Wrapperهای رسمی Ragas برای جلوگیری از خطای AttributeError
    ragas_llm = LangchainLLMWrapper(base_llm)
    ragas_emb = LangchainEmbeddingsWrapper(base_embeddings)

    # 4. آماده‌سازی متریک‌ها با مدل‌های جدید
    # در نسخه‌های جدید، این بهترین روش برای اطمینان از تزریق LLM است
    faithfulness.llm = ragas_llm
    answer_relevancy.llm = ragas_llm
    answer_relevancy.embeddings = ragas_emb

    print(f"✅ RAGAS setup completed: llm={EVAL_OPENAI_MODEL}")

# --- توابع کمکی برای اجرای ارزیابی ---
def run_evaluation(dataset):
    """
    dataset: باید شامل ستون‌های ['question', 'answer', 'contexts', 'ground_truth'] باشد.
    """
    if not eval_enabled:
        print("Evaluation is disabled due to missing API Key.")
        return None
        
    print("🚀 Running RAGAS evaluation on dataset...")
    
    try:
        # اجرای ارزیابی
        # پاس دادن llm و embeddings در اینجا لایه نهایی اطمینان است
        result = evaluate(
            dataset=dataset,
            metrics=[faithfulness, answer_relevancy],
            llm=ragas_llm,
            embeddings=ragas_emb
        )
        
        # تبدیل به DataFrame برای نمایش بهتر
        df_res = result.to_pandas()
        
        print("\n--- میانگین نمرات سیستم ---")
        print(f"Faithfulness: {df_res['faithfulness'].mean():.4f}")
        print(f"Answer Relevancy: {df_res['answer_relevancy'].mean():.4f}")
        
        return result
        
    except Exception as e:
        print(f"❌ An error occurred during evaluation: {e}")
        return None

# اجرای تست (در صورت آماده بودن hf_dataset)
# results = run_evaluation(hf_dataset)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

def test_retrieval(query, vectorizer, X, docs, top_k=1):
    query_vec = vectorizer.transform([query])

    similarities = cosine_similarity(query_vec, X).flatten()

    best_idx = similarities.argsort()[-top_k:][::-1]

    print(f"Query: {query}")
    for idx in best_idx:
        print(f"Score: {similarities[idx]:.4f}")
        print(f"Content: {docs[idx][:200]}...") # نمایش بخشی از متن پیدا شده
        print("-" * 30)

# تست: یک سوال که کلماتش در یکی از سندها هست بپرسید
test_retrieval("موضوع اصلی سند اول چیست؟", vectorizer, X, docs)
```

```python
import pandas as pd
from dotenv import load_dotenv
import os
from ragas.metrics.collections import faithfulness, answer_relevancy
from openai import OpenAI
from ragas.llms import llm_factory
from ragas.embeddings import embedding_factory

load_dotenv()

# Use modern ragas API
openai_client = OpenAI(api_key=os.getenv('OPENAI_API_KEY'), base_url=os.getenv('OPENAI_API_BASE'))
# eval_llm = llm_factory(EVAL_OPENAI_MODEL, client=openai_client)  # Remove this to avoid InstructorLLM
eval_embeddings = embedding_factory('openai', model='text-embedding-3-small', client=openai_client)

if eval_enabled and len(eval_records) > 0:
    print(f"Evaluating {len(eval_records)} samples...")

    # Prepare data for ragas (robust to missing keys)
    data_for_ragas = {
        'question': [rec.get('question', '') for rec in eval_records],
        'answer': [rec.get('answer', '') for rec in eval_records],
        'contexts': [rec.get('contexts', []) for rec in eval_records],
        'ground_truth': [rec.get('ground_truth', '') for rec in eval_records]  # Add if available
    }

    # Sanitize contexts: ensure list[str]
    for i, c in enumerate(data_for_ragas['contexts']):
        if c is None:
            data_for_ragas['contexts'][i] = []
        elif not isinstance(c, list):
            data_for_ragas['contexts'][i] = [str(c)]

    hf_dataset = Dataset.from_dict(data_for_ragas)

    # Sanity check
    print("Sanity-check sample (first 3 entries):")
    for i, row in enumerate(hf_dataset.select(range(min(3, len(hf_dataset))))):
        print(i, 'question_type=', type(row['question']), 'answer_type=', type(row['answer']), 'contexts_type=', type(row['contexts']))

    try:
        # Wrap LLM and embeddings to be RAGAS-compatible.
        # Use the base LangChain LLM directly (avoid ragas.llms.llm_factory which returns InstructorLLM without agenerate_prompt)
        from ragas.llms import LangchainLLMWrapper
        from ragas.embeddings import LangchainEmbeddingsWrapper
        from langchain_openai import ChatOpenAI, OpenAIEmbeddings

        # Create LangChain LLM directly
        base_llm = ChatOpenAI(
            model=EVAL_OPENAI_MODEL,
            openai_api_key=os.getenv('OPENAI_API_KEY'),
            openai_api_base=os.getenv('OPENAI_API_BASE'),
            temperature=0.0  # deterministic for evaluation
        )
        ragas_llm = LangchainLLMWrapper(base_llm)
        
        # Create LangChain embeddings directly
        base_embeddings = OpenAIEmbeddings(
            model='text-embedding-3-small',
            openai_api_key=os.getenv('OPENAI_API_KEY'),
            openai_api_base=os.getenv('OPENAI_API_BASE')
        )
        ragas_emb = LangchainEmbeddingsWrapper(base_embeddings)

        # Create metric instances with LLM set
        faithfulness_metric = faithfulness(llm=ragas_llm)
        answer_relevancy_metric = answer_relevancy(llm=ragas_llm, embeddings=ragas_emb)

        results = evaluate(
            dataset=hf_dataset,
            metrics=[faithfulness_metric, answer_relevancy_metric],
            raise_exceptions=False  # Don't fail on individual errors
        )

        df_results = results.to_pandas()
        print(f"\nEvaluation completed. Results shape: {df_results.shape}")
        print(f"Average Answer Relevancy: {df_results['answer_relevancy'].mean():.4f}")
        print(f"Average Faithfulness: {df_results['faithfulness'].mean():.4f}")

        # Save results
        output_path = "data/rag_evaluation_report.xlsx"
        df_results.to_excel(output_path, index=False)
        print(f"Results saved to {output_path}")

    except Exception as e:
        print(f"Evaluation failed: {e}")
        # Create placeholder results
        df_results = pd.DataFrame({
            'question': data_for_ragas['question'],
            'answer': data_for_ragas['answer'],
            'contexts': data_for_ragas['contexts'],
            'answer_relevancy': [0.0] * len(eval_records),
            'faithfulness': [0.0] * len(eval_records),
            'error': str(e)
        })
        print("Created placeholder results due to evaluation failure")
else:
    print("Evaluation disabled or no records to evaluate")
    df_results = pd.DataFrame()  # Empty dataframe
```

<b>۱. خطاهای TimeoutError:</b>
<ul>
<li>خطاهای <b>TimeoutError</b> نشان می‌دهد که مدل LLM یا سرور Ollama در زمان پاسخ‌دهی به درخواست‌های Ragas دچار تاخیر زیاد شده و نتوانسته در بازه زمانی مجاز پاسخ دهد. این مشکل معمولاً به دلیل کند بودن مدل، محدودیت منابع سخت‌افزاری (CPU/RAM)، یا حجم زیاد درخواست‌های موازی رخ می‌دهد.</li>
<li>راه‌حل: کاهش تعداد درخواست‌های همزمان (batch size)، افزایش منابع سرور، یا استفاده از مدل‌های سبک‌تر می‌تواند این مشکل را کاهش دهد.</li>
</ul>

<b>۲. خطاهای OutputParserException:</b>
<ul>
<li>این خطاها معمولاً زمانی رخ می‌دهد که خروجی مدل LLM با فرمت مورد انتظار Ragas (مثلاً JSON یا دیکشنری خاص) مطابقت ندارد و مدل به جای خروجی ساختاریافته، متن آزاد تولید می‌کند.</li>
<li>راه‌حل: باید prompt مدل را به گونه‌ای تنظیم کنید که خروجی دقیقاً مطابق با فرمت مورد انتظار باشد (مثلاً با دستور "Please answer in valid JSON format"). همچنین می‌توانید از مدل‌های سازگارتر با LangChain/Ragas استفاده کنید.</li>
</ul>

<b>۳. هشدار No statements were generated from the answer:</b>
<ul>
<li>این هشدار نشان می‌دهد که Ragas نتوانسته از پاسخ مدل، جمله یا statement معتبری استخراج کند تا معیار Faithfulness را محاسبه کند. در نتیجه مقدار <b>Faithfulness: nan</b> شده است.</li>
<li>علت: معمولاً به دلیل خروجی غیرساختاریافته یا نامفهوم مدل رخ می‌دهد.</li>
