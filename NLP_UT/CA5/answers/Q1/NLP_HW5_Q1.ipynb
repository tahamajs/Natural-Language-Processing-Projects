توجه داشته باشید که metadataهای انتخاب‌شده باید در سناریوی بازیابی اطلاعات، قابل‌توجیه و منطقی باشند. در پایان این مرحله، یک schema مناسب برای پایگاه داده طراحی و تعریف کنید.

```python
# Dataset preparation (self-contained example)
# This cell builds a small toy dataset, computes TF-IDF embeddings
import os
import pickle
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd

# Small example documents (replace with real dataset paths / OCR outputs)
docs = [
    "قانون جرایم رایانه‌ای: تعریف جرم و مجازات‌ها و نحوه‌ی رسیدگی به شکایات",
    "قانون کار: شرایط قرارداد کاری، حقوق کارگران و مزایای بیمه‌ای",
    "قانون مدنی: قواعد مربوط به مالکیت، قراردادها و مسئولیت مدنی",
]

# Example metadata schema for each document
metadata = [
    {
        "doc_id": 0,
        "title": "قانون جرایم رایانه‌ای",
        "year": 1399,
        "source": "مجله رسمی",
        "jurisdiction": "ملی",
    },
    {
        "doc_id": 1,
        "title": "قانون کار",
        "year": 1385,
        "source": "مجله رسمی",
        "jurisdiction": "ملی",
    },
    {
        "doc_id": 2,
        "title": "قانون مدنی",
        "year": 1370,
        "source": "مجله رسمی",
        "jurisdiction": "ملی",
    },
]

# Build TF-IDF index as a lightweight vector DB substitute
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(docs)

# Persist index so later cells can load it
os.makedirs("data", exist_ok=True)
with open("data/indexed_dataset.pkl", "wb") as f:
    pickle.dump(
        {"docs": docs, "metadata": metadata, "vectorizer": vectorizer, "X": X}, f
    )

print("Indexed %d documents. Sample metadata schema:" % len(docs))
print(metadata[0])
```

## <div style="text-align: center; direction: rtl; font-family: Vazir;">طراحی عامل(۲۵ امتیاز)</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
تمامی کد های شما می بایست با استفاده از <a href="https://www.langchain.com/" target="_blank" rel="noopener noreferrer">langchain</a>
 , <a href="https://www.langchain.com/langgraph" target="_blank" rel="noopener noreferrer">langgraph</a>
 پیاده سازی شده باشد

قابلیت های کلی Agent باید به صورت زیر باشد :

یک نود با نام rewrite_query طراحی کنید که ورودی آن، query اصلی کاربر و خروجی آن، یک rewritten_query به‌عنوان نسخه‌ای بازنویسی‌شده و شفاف‌تر از سؤال باشد. وظیفه این ماژول آن است که ابهام‌های زبانی موجود در پرسش را برطرف کند، لحن را استاندارد و یکنواخت سازد و در نهایت پرسش کاربر را به فرمی تبدیل کند که برای مرحله‌ی بازیابی اطلاعات (retrieval) مناسب‌تر باشد.

نودی با نام classify_intent طراحی کنید که وظیفه آن تشخیص نوع پرسش کاربر بر اساس query دریافتی است. این نود باید تعیین کند که پرسش کاربر در کدام یک از دسته‌های «Greeting / Small Talk» (مانند سلام، احوال‌پرسی، تبریک و عبارات مشابه)، «Abusive / Insult / بی‌احترامی» یا «Law Question» (سؤال مرتبط با قانون و نیازمند مراجعه به پایگاه دانش) قرار می‌گیرد. در صورت تشخیص دسته‌ی Greeting، نود باید یک پاسخ کوتاه، محترمانه و دوستانه تولید کند. در صورت تشخیص دسته‌ی Abusive، نود موظف است پاسخی حرفه‌ای، محترمانه و بدون هرگونه توهین ارائه دهد و در آن توضیح دهد که مدل برای کمک طراحی شده است و وارد جدل یا بحث تنش‌زا نمی‌شود. اگر خروجی نود در دسته‌ی Law Question قرار گرفت، جریان پردازش باید به نود بعدی، یعنی نود «Metadata Extraction»، منتقل شود تا مراحل بعدی بازیابی اطلاعات انجام گیرد.

یک نود با نام extract_metadata طراحی کنید که ورودی آن rewritten_query و خروجی آن یک ساختار metadata_filter باشد؛ این ساختار در مرحله‌ی بازیابی اطلاعات از LanceDB برای اعمال فیلترهای مرتبط استفاده خواهد شد. در این نود، با استفاده از یک LLM، از روی rewritten_query اطلاعات ساختارمند و قابل‌استفاده به‌عنوان فیلتر استخراج می‌شود تا در قالب metadata_filter به LanceDB ارسال شده و بازیابی اسناد با دقت و ارتباط بیشتری انجام پذیرد.

نودی با نام context_retrieve طراحی کنید که با استفاده از rewritten_query در LanceDB عملیات جستجو را انجام دهد. این نود باید از ساختار metadata_filter برای محدود کردن فضای جستجو و اعمال فیلترهای موردنظر بهره بگیرد و در نهایت، تعداد مشخصی سند (chunk) مرتبط با پرسش کاربر، مثلاً به تعداد K سند (برای نمونه K = 10)، را به‌عنوان خروجی بازگرداند.

یک نود با نام rerank طراحی کنید که روی خروجی مرحله‌ی context_retrieve یک لایه‌ی بازمرتب‌سازی (Reranking) اعمال کند؛ به‌عنوان مثال می‌توان از یک LLM-based reranker یا یک معیار شباهت ثانویه برای امتیازدهی مجدد به اسناد استفاده کرد. این نود باید بر اساس مقادیر similarity یا score، تعداد محدودی از بهترین سندها (top-N، مثلاً N = 3) را برای مرحله‌ی تولید پاسخ (generation) انتخاب کند. علاوه بر این، در همین نود باید یک مرحله‌ی بررسی ارتباط (Relevance Check) انجام شود؛ به این صورت که اگر هیچ‌یک از contextهای انتخاب‌شده در top-N، بر اساس یک آستانه‌ی تعریف‌شده یا قضاوت LLM، ارتباط کافی با query نداشتند، فرآیند retrieval یک‌بار دیگر تکرار شده و تلاش مجدد برای بازیابی contextهای مرتبط‌تر انجام گیرد. دقت شود که نهایتا یک بار می بایست این مرحله انجام شود و Agent در loop بی نهایت قرار نگیرد پس این موضوع باید توسط شما کنترل شود.

نودی با نام generate_answer طراحی کنید که ورودی آن شامل rewritten_query به‌همراه contextهای نهایی انتخاب‌شده از مراحل قبل باشد و بر اساس این ورودی، وظیفه دارد یک پاسخ متنی نهایی، منسجم، دقیق و قابل‌فهم برای کاربر تولید و به‌عنوان خروجی برگرداند.

به طور کلی معماری پیشنهادی یک حالت از نحوه پیاده سازی این Agent هست و می توان نوع نگاه متفاوتی در طراحی معماری Agent داشت پس شما مجاز هستید که با پیاده سازی همه مواردی که بیان شد با معماری متفاوت عمل کنید معماری پیشنهادی شما می بایست تمامی این موارد را شامل شده باشد:

query rewriting, extract metadata , context retrieve , reranking , generate answer

</p>
</div>


```python
# Agent components: rewrite, classify_intent, extract_metadata, retrieve, rerank, generate
import time
import re
import pickle
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Load index
with open("data/indexed_dataset.pkl", "rb") as f:
    store = pickle.load(f)
docs = store["docs"]
metadata = store["metadata"]
vectorizer = store["vectorizer"]
X = store["X"]


# 1) Query rewriting (simple normalization + expansion)
def rewrite_query(query: str) -> str:
    q = query.strip()
    # lowercase, remove extra spaces
    q = re.sub("\s+", " ", q)
    return q


# 2) Intent classification (keyword based)
GREET_KEYWORDS = ["سلام", "درود", "صبح بخیر", "وقت بخیر"]
ABUSE_KEYWORDS = ["مسخره", "بی‌ادب"]


def classify_intent(query: str) -> dict:
    q = query.lower()
    for w in GREET_KEYWORDS:
        if w in q:
            return {"intent": "greeting", "response": "سلام! چطور می‌توانم کمک کنم؟"}
    for w in ABUSE_KEYWORDS:
        if w in q:
            return {
                "intent": "abuse",
                "response": "من اینجا هستم تا کمک کنم؛ لطفاً مودبانه سوالاتتان را مطرح کنید.",
            }
    return {"intent": "law_question"}


# 3) Extract metadata (very simple heuristics)
def extract_metadata(rewritten_query: str) -> dict:
    # try to extract year or jurisdiction words
    m = {}
    years = re.findall(r"13\d{2}|14\d{2}|\d{4}", rewritten_query)
    if years:
        try:
            m["year"] = int(years[0])
        except:
            pass
    # example: look for words like "کار", "مدنی" to filter by title keywords
    keywords = []
    if "کار" in rewritten_query:
        keywords.append("کار")
    if "مدنی" in rewritten_query:
        keywords.append("مدنی")
    if "جرایم" in rewritten_query or "رایانه" in rewritten_query:
        keywords.append("جرایم")
    if keywords:
        m["keywords"] = keywords
    return m


# 4) Context retrieve using TF-IDF + optional metadata filtering
def context_retrieve(rewritten_query: str, metadata_filter: dict = None, K: int = 10):
    q_vec = vectorizer.transform([rewritten_query])
    sims = cosine_similarity(q_vec, X).flatten()
    # apply metadata filter if present
    idxs = np.argsort(-sims)
    if metadata_filter:
        # simple filter: keep docs whose metadata contain any keyword
        filtered = []
        for i in idxs:
            meta = metadata[i]
            ok = True
            if "year" in metadata_filter:
                ok = ok and (meta.get("year") == metadata_filter["year"])
            if "keywords" in metadata_filter:
                ok = ok and any(
                    k in meta.get("title", "") or k in docs[i]
                    for k in metadata_filter["keywords"]
                )
            if ok:
                filtered.append((i, sims[i]))
        results = filtered[:K]
    else:
        results = [(int(i), float(sims[int(i)])) for i in idxs[:K]]
    contexts = []
    for i, score in results:
        contexts.append(
            {"doc_id": i, "text": docs[i], "metadata": metadata[i], "score": score}
        )
    return contexts


# 5) Rerank + relevance check
def rerank(contexts, top_n=3, relevance_threshold=0.05):
    if not contexts:
        return []
    contexts = sorted(contexts, key=lambda x: -x["score"])
    top = contexts[:top_n]
    # relevance check: if top scores are too low, return empty to signal retry
    if all(c["score"] < relevance_threshold for c in top):
        return []
    return top


# 6) Generate answer (simple extractive aggregator)
def generate_answer(rewritten_query: str, contexts):
    if not contexts:
        return "متاسفانه نتوانستم پاسخی مرتبط پیدا کنم. لطفاً سوال را واضح‌تر مطرح کنید."
    # join top contexts and return small synthesized answer
    parts = [c["text"] for c in contexts]
    answer = "\n\n---\n\n".join(parts)
    final = f"پاسخ پیشنهادی (خلاصه از منابع):\n{answer}"
    return final


# Pipeline runner with timing logs


def run_pipeline(query: str, K=10, top_n=3):
    timings = {}
    t0 = time.time()
    timings["start"] = t0

    t = time.time()
    rewritten = rewrite_query(query)
    timings["rewrite"] = time.time() - t

    t = time.time()
    intent = classify_intent(rewritten)
    timings["classify_intent"] = time.time() - t
    if intent.get("intent") != "law_question":
        return {"answer": intent.get("response"), "timings": timings, "contexts": []}

    t = time.time()
    meta_filter = extract_metadata(rewritten)
    timings["extract_metadata"] = time.time() - t

    t = time.time()
    contexts = context_retrieve(rewritten, meta_filter, K=K)
    timings["context_retrieve"] = time.time() - t

    t = time.time()
    top = rerank(contexts, top_n=top_n)
    # if rerank returned empty (low relevance), try once more with larger K
    if not top:
        contexts = context_retrieve(rewritten, None, K=min(len(docs), K * 2))
        top = rerank(contexts, top_n=top_n)
    timings["rerank"] = time.time() - t

    t = time.time()
    answer = generate_answer(rewritten, top)
    timings["generate_answer"] = time.time() - t

    timings["total"] = time.time() - t0
    return {"answer": answer, "timings": timings, "contexts": top}


# Example quick test
if __name__ == "__main__":
    q = "حقوق کارگر در قرارداد کاری چیست؟"
    out = run_pipeline(q)
    print(out["answer"])
    print(out["timings"])
```

## <div style="text-align: center; direction: ltr;   font-family: Vazir;">تحلیل زمانی عملکرد عامل(۵ امتیاز)</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
   در اطراف هر نود اصلی سیستم یعنی rewrite, classify_intent, extract_metadata, context_retrieve, rerank و generate_answer، زمان شروع و پایان اجرا را ثبت کنید. برای مجموعه‌ای شامل ۳ سوال از هر داکیومنت، کل pipeline سیستم RAG را اجرا کرده و زمان اجرای هر نود را برای هر پرسش لاگ کنید.

مشخص کنید گلوگاه زمانی (bottleneck) سیستم کجاست.

پیشنهاد دهید برای کاهش latency چه بهبودهایی می‌توان انجام داد. . </p>

</div>


```python
# Timing analysis: run pipeline for 3 queries per document and log node times
import pandas as pd

queries = []
for meta, doc in zip(metadata, docs):
    for i in range(3):
        queries.append(
            {
                "query": f"سوال درباره {meta['title']} شماره {i+1}",
                "doc_id": meta["doc_id"],
            }
        )

records = []
for q in queries:
    out = run_pipeline(q["query"])
    rec = {"query": q["query"], "doc_id": q["doc_id"]}
    # timings keys: rewrite, classify_intent, extract_metadata, context_retrieve, rerank, generate_answer, total
    for k, v in out["timings"].items():
        rec[k] = v
    records.append(rec)

df = pd.DataFrame(records)
print(df.head())
# aggregate
agg = df.mean(numeric_only=True).to_dict()
print("Average timings (s):", agg)
# find bottleneck node
nodes = [
    "rewrite",
    "classify_intent",
    "extract_metadata",
    "context_retrieve",
    "rerank",
    "generate_answer",
]
avg_nodes = {n: agg.get(n, 0) for n in nodes}
bottleneck = max(avg_nodes, key=avg_nodes.get)
print(f"Bottleneck node: {bottleneck} with avg {avg_nodes[bottleneck]:.4f}s")

# save timings
df.to_csv("data/pipeline_timings.csv", index=False)

# Suggestions (print to notebook)
print("\nبهبودهای پیشنهادی برای کاهش latency:")
print("- استفاده از یک vector DB در حافظه یا LanceDB با قابلیت kNN سریع")
print("- استفاده از مدل‌های Embedding سریع‌تر و نگهداری و کش کردن بردارها")
print("- موازی‌سازی درخواست‌ها و افزایش batch size در محاسبه‌ی بردارها")
```

## <div style="text-align: center; direction: ltr; font-family: Vazir;">(۱۰ امتیاز) ارزیابی کیفیت بازیابی</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
  ابتدا یک مجموعه سوال از دیتاست ها طراحی کنید ( از هر داکیومنت سه سوال و جواب ) با مدل های زبانی میتوانید برا تسریع این بخش کارتان را پیش ببرید از کتابخانه‌ی <a href="https://docs.ragas.io/en/stable/" target="_blank" rel="noopener noreferrer">ragas</a>
 برای ارزیابی سیستم RAG خود استفاده کنید و معیارهای زیر را محاسبه کنید: answer_relevancy، faithfulness  در این بخش لازم است یک pipeline ارزیابی طراحی کنید که طی آن، برای هر سؤال موجود در مجموعه‌ی ارزیابی، سیستم RAG شما اجرا شده، پاسخ تولیدی مدل به‌همراه contextهای استفاده‌شده در فرایند بازیابی ذخیره شوند. سپس این خروجی‌ها را در قالب مورد نیاز کتابخانه‌ی RAGAS آماده کرده و معیارهای انتخاب‌شده را بر روی آن‌ها محاسبه و تحلیل کنید. </p>
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
یک اکسل شامل سوال و محتوای بازیابی شده به همراه جواب عامل هوشمند و جواب صحیح مورد انتظار که در نهایت در این اکسل مقادیر هر کدام از معیار های بالا گزارش شده است
در نهایت در قسمت پاسخ تشریحی مقادیر را گزارش کنید و مقادیرتان را تحلیل کنبد خطاها را علت یابی کنید و توضیح دهید
</p>


```python
# Evaluation pipeline: build Q/A pairs, run the agent, compute simple metrics and export to Excel
import pandas as pd
import re

# Build evaluation set: 3 questions per document with gold answers (extractive)
eval_records = []
for i, (meta, doc) in enumerate(zip(metadata, docs)):
    # make three simple extractive QAs
    gold1 = doc.split("：")[-1] if "：" in doc else doc
    questions = [
        (f"موضوع اصلیِ {meta['title']} چیست؟", gold1[:200]),
        (f"چه مواردی در {meta['title']} ذکر شده‌اند؟", gold1[:200]),
        (f"خلاصه‌ای از {meta['title']} را بگویید.", gold1[:200]),
    ]
    for q, a in questions:
        eval_records.append({"doc_id": meta["doc_id"], "question": q, "gold_answer": a})

# Run agent on each question and compute simple metrics
results = []
for r in eval_records:
    out = run_pipeline(r["question"])
    pred = out["answer"]
    contexts = " ".join([c["text"] for c in out["contexts"]]) if out["contexts"] else ""

    # simple token overlap based metrics
    def tokenize(x):
        return set(re.findall(r"[ء-ي\w]+", x))

    gold_tokens = tokenize(r["gold_answer"])
    pred_tokens = tokenize(pred)
    ctx_tokens = tokenize(contexts)
    # answer_relevancy: jaccard between pred and gold
    if gold_tokens or pred_tokens:
        jaccard = len(gold_tokens & pred_tokens) / max(
            1, len(gold_tokens | pred_tokens)
        )
    else:
        jaccard = 0.0
    # faithfulness: proportion of pred tokens that appear in contexts
    if pred_tokens:
        faithfulness = len(pred_tokens & ctx_tokens) / len(pred_tokens)
    else:
        faithfulness = 0.0
    results.append(
        {
            "doc_id": r["doc_id"],
            "question": r["question"],
            "gold_answer": r["gold_answer"],
            "predicted_answer": pred,
            "answer_relevancy": jaccard,
            "faithfulness": faithfulness,
        }
    )

df_eval = pd.DataFrame(results)
print(df_eval.head())
# save to excel
df_eval.to_excel("data/eval_results.xlsx", index=False)

# summary
print("\nAverage answer_relevancy:", df_eval["answer_relevancy"].mean())
print("Average faithfulness:", df_eval["faithfulness"].mean())
```
