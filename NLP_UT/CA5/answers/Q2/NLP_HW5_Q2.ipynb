# CA5 Q2 TravBot Agent Pipeline

This notebook runs the full Q2 workflow and generates report artifacts.


## Steps

1. Validate strict live prerequisites.
2. Execute the Q2 runner script.
3. Verify generated artifacts.


```python
from pathlib import Path
import sys

ROOT = Path.cwd()
SCRIPTS_DIR = ROOT / "answers" / "scripts"
if str(SCRIPTS_DIR) not in sys.path:
    sys.path.append(str(SCRIPTS_DIR))

from preflight import check_all

status = check_all(strict_live=True)
if status != 0:
    raise RuntimeError("Preflight failed")
print("Preflight passed")
```

```python
import subprocess

cmd = ["python", "answers/scripts/run_q2.py"]
completed = subprocess.run(cmd, capture_output=True, text=True)
print(completed.stdout)
if completed.returncode != 0:
    print(completed.stderr)
    raise RuntimeError("Q2 runner failed")
```
