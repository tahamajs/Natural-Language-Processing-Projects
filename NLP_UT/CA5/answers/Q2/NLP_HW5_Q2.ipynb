````python
# Load API keys from env/keys.json/.keys.json/.env and initialize clients
import json
import os
from pathlib import Path

from amadeus import Client

KEY_NAMES = [
    "OPENAI_API_KEY",
    "OPENAI_API_BASE",
    "TAVILY_API_KEY",
    "AMADEUS_CLIENT_ID",
    "AMADEUS_CLIENT_SECRET",
    "LLAMA_CLOUD_API_KEY",
    "KAGGLE_USERNAME",
    "KAGGLE_KEY",
]

def _mask(val: str | None, keep: int = 4) -> str:
    if not val:
        return "not set"
    if len(val) <= keep:
        return "***"
    return val[:keep] + "***" + val[-keep:]

def load_keys_from_files() -> None:
    for name in ("keys.json", ".keys.json"):
        path = Path(name)
        if not path.exists():
            continue
        try:
            data = json.loads(path.read_text(encoding="utf-8"))
        except Exception:
            continue
        if not isinstance(data, dict):
            continue
        for k in KEY_NAMES:
            if os.environ.get(k):
                continue
            v = data.get(k)
            if v:
                os.environ[k] = str(v)

    # minimal .env parser (KEY=VALUE, optional 'export')
    for name in (".env", "env"):
        path = Path(name)
        if not path.exists():
            continue
        try:
            for line in path.read_text(encoding="utf-8").splitlines():
                line = line.strip()
                if not line or line.startswith("#"):
                    continue
                if line.startswith("export "):
                    line = line[len("export ") :]
                if "=" not in line:
                    continue
                k, v = line.split("=", 1)
                k = k.strip()
                v = v.strip().strip('"').strip("'")
                if k in KEY_NAMES and not os.environ.get(k) and v:
                    os.environ[k] = v
        except Exception:
            pass

load_keys_from_files()

OPENAI_API_KEY = os.environ.get("OPENAI_API_KEY")
OPENAI_API_BASE = os.environ.get("OPENAI_API_BASE", "https://api.openai.com/v1")

print("Key status (masked):")
for k in KEY_NAMES:
    if k == "OPENAI_API_BASE":
        print(f"  {k}: {os.environ.get(k) or 'default'}")
    else:
        print(f"  {k}: {_mask(os.environ.get(k))}")

# Initialize Amadeus client if credentials exist
if os.environ.get("AMADEUS_CLIENT_ID") and os.environ.get("AMADEUS_CLIENT_SECRET"):
    amadeus = Client(
        client_id=os.environ.get("AMADEUS_CLIENT_ID"),
        client_secret=os.environ.get("AMADEUS_CLIENT_SECRET"),
    )
    print("✅ Amadeus client initialized.")
else:
    amadeus = None
    print("⚠️ Amadeus credentials not set.")
print(f"OpenAI API configured with base URL: {OPENAI_API_BASE}")
````

````python
path = kagglehub.dataset_download("phanee16/currency-and-country-code-mapping")
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
print(f"Found CSV files: {csv_files}")
````

````python
country_to_currency = {}
if csv_files:
    # Try each CSV until we find suitable columns
    found = False
    for csv_name in csv_files:
        csv_path = os.path.join(path, csv_name)
        try:
            df = pd.read_csv(csv_path)
        except Exception:
            # Fallback: try with engine/python for odd delimiters
            try:
                df = pd.read_csv(csv_path, engine="python")
            except Exception:
                continue
        df.columns = [str(c).lower().strip() for c in df.columns]
        # Possible column name synonyms
        country_candidates = [
            "country", "country_name", "name", "entity", "country or territory", "country/ territory"
        ]
        code_candidates = [
            "currency_code", "currency code", "code", "iso_code", "currency",
            "alphabetic_code", "alphabetic code", "iso 4217 code", "iso4217"
        ]
        country_col = next((c for c in df.columns if c in country_candidates), None)
        code_col = next((c for c in df.columns if c in code_candidates), None)
        if country_col and code_col:
            for _, row in df.iterrows():
                country = str(row[country_col]).lower().strip()
                code = str(row[code_col]).strip()
                if country and code and code.lower() != "nan":
                    country_to_currency[country] = code
            print(f"Using {csv_name} with columns: country='{country_col}', code='{code_col}'")
            found = True
            break
    if found:
        print(len(country_to_currency))
        print(f"Sample: france -> {country_to_currency.get('france')}, iran -> {country_to_currency.get('iran')}")
    else:
        print("Could not automatically find columns.")
else:
    print("No CSV file found.")
````

````python
import pikepdf
import os

def repair_pdf(file_path):
    fixed_path = file_path.replace(".pdf", "_fixed.pdf")
    try:
        print(f"Attempting to repair corrupted PDF: {file_path}")
        with pikepdf.open(file_path) as pdf:
            pdf.save(fixed_path)
        print("Repair successful!")
        return fixed_path
    except Exception as e:
        print(f"Could not repair PDF: {e}")
        return file_path

# Use it before parsing
file_to_parse = repair_pdf("./tourism_book.pdf")
# Now pass file_to_parse to your LlamaParse or PyMuPDF functions
````

````python
import os
import pandas as pd

# PyMuPDF import compatibility
try:
    import pymupdf  # PyMuPDF >= 1.24
except ImportError:
    import fitz as pymupdf  # type: ignore


def get_tourism_data(file_path: str, api_key: str | None):
    """Extract tourism book text into a DataFrame.

    Tries LlamaParse if an API key is provided; otherwise falls back to local PDF parsing.
    """

    # 1) Attempt LlamaParse (optional)
    if api_key:
        try:
            from llama_parse import LlamaParse

            print("Attempting LlamaParse...")
            parser = LlamaParse(api_key=api_key, result_type="markdown")
            documents = parser.load_data(file_path)
            if documents:
                return pd.DataFrame(
                    [
                        {"text": doc.text, "metadata": getattr(doc, "metadata", {})}
                        for doc in documents
                    ]
                )
        except Exception as e:
            print(f"LlamaParse failed (falling back to local parsing): {e}")
    else:
        print("LLAMA_CLOUD_API_KEY not set. Skipping LlamaParse.")

    # 2) Local fallback (PyMuPDF)
    print("Falling back to local PyMuPDF parsing...")
    if os.path.exists(file_path):
        try:
            doc = pymupdf.open(file_path)
            fallback_data = []
            for page in doc:
                fallback_data.append(
                    {
                        "text": page.get_text(),
                        "metadata": {"page": page.number + 1, "source": "local_pymupdf"},
                    }
                )
            return pd.DataFrame(fallback_data)
        except Exception as e:
            print(f"Local fallback failed: {e}")
    else:
        print(f"File not found: {file_path}")

    return pd.DataFrame(columns=["text", "metadata"])


# --- RUNNING THE CODE ---
# Provide your LlamaParse key via env var instead of hard-coding it in the notebook.
llama_api_key = "llx-0p7M463eE51Afc6IwUUDUgPUL7QWghvtJZjNx1RbNtOV2JxC"

df_tourism = get_tourism_data("./tourism_book_fixed.pdf", llama_api_key)
if not df_tourism.empty:
    print(f"Success! Extracted {len(df_tourism)} sections.")
    display(df_tourism.head())
else:
    print("All parsing methods failed.")
````

<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
یکی از ابزارهای مهمی که تراول‌بات با استفاده از آن به کاربران خدمات می‌دهد، ابزار جستجوی پرواز است. شما باید ابزاری تعریف کنید که پارامترهایی مانند مبدا، مقصد و تاریخ پرواز را از کاربر دریافت کرده و با استفاده از Amadeus API به جستجوی پروازهای مورد نظر بپردازد. این ابزار باید حداقل ۳ پرواز را با اطلاعات کاملی شامل شماره پرواز، کد ایرلاین، قیمت دقیق (به همراه واحد پول)، تاریخ و ساعت حرکت، مدت پرواز و تعداد توقف‌ها به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **تبدیل نام شهر به کد IATA:** ابزار Amadeus بر اساس کدهای یکتای فرودگاه (IATA Code) کار می‌کند. از تابع مپینگی که در بخش قبل نوشته‌اید استفاده کنید تا نام شهرها را به کد فرودگاه تبدیل کنید (مثلاً Tehran → IKA، Dubai → DXB).

- **مدیریت فرمت تاریخ:** Amadeus تاریخ را به فرمت YYYY-MM-DD می‌پذیرد. از آنجایی که ماهیت چت‌بات صحبت به زبان طبیعی است، کاربران فرمت تاریخ را به گونه‌های متفاوتی وارد می‌کنند(برای مثال به صورت متن و نه عدد). شما باید مدیریت فرمت تاریخ در ابزار و System Prompt را مد نظر داشته باشید.

- **مدیریت خطا:** در صورت بروز خطا در جستجو (مثل عدم یافتن پرواز یا خطای API)، پیام خطای مناسب و قابل فهم به کاربر بازگردانید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید تا LangGraph بتواند آن را شناسایی کند.

````python
from langchain.tools import tool
from amadeus import Client, ResponseError
import os
import re
# اصلاح ایمپورت‌ها برای جلوگیری از ارور strptime
from datetime import date, datetime, timedelta

# ==========================================
# 1. Amadeus Client Setup
# ==========================================
def _get_amadeus_client():
    os.environ["AMADEUS_CLIENT_ID"] = "a6IDhfIeaa5hWcY6HMbGwfDoUCjyubc3"
    os.environ["AMADEUS_CLIENT_SECRET"] = "s7wGCKGTlQpYIhw4"


    client_id = "a6IDhfIeaa5hWcY6HMbGwfDoUCjyubc3"
    client_secret = "s7wGCKGTlQpYIhw4"
    if not client_id or not client_secret:
        return None
    try:
        return Client(client_id=client_id, client_secret=client_secret)
    except Exception as e:
        print(f"Amadeus init error: {e}")
        return None

# Initialize client globally
try:
    if 'amadeus' not in globals():
        amadeus = _get_amadeus_client()
except NameError:
    amadeus = _get_amadeus_client()


# ==========================================
# 2. Helper Functions
# ==========================================

def get_iata_code(city_name: str) -> str | None:
    """Finds the IATA code for a city (Offline dict + API fallback)."""

    # لیست آفلاین برای شهرهای پرکاربرد (جهت سرعت و جلوگیری از خطای API)
    COMMON_CITIES = {
        "tehran": "IKA", "mashhad": "MHD", "shiraz": "SYZ", "isfahan": "IFN", "tabriz": "TBZ",
        "dubai": "DXB", "istanbul": "IST", "london": "LHR", "new york": "JFK", "nyc": "JFK",
        "paris": "CDG", "rome": "FCO", "milan": "MXP", "frankfurt": "FRA", "berlin": "BER",
        "tokyo": "HND", "beijing": "PEK", "shanghai": "PVG", "sydney": "SYD", "toronto": "YYZ",
        "barcelona": "BCN", "madrid": "MAD", "amsterdam": "AMS", "doha": "DOH"
    }

    clean_city = city_name.lower().strip()

    # 1. Check offline dictionary first
    if clean_city in COMMON_CITIES:
        return COMMON_CITIES[clean_city]

    # 2. Fallback to Amadeus API if available
    if amadeus:
        try:
            response = amadeus.reference_data.locations.get(
                keyword=city_name,
                subType="CITY"
            )
            if response.data:
                return response.data[0]['iataCode']
        except Exception:
            pass # If API fails, return None

    return None

def _normalize_yyyy_mm_dd(date_text: str, base=None) -> str | None:
    """Normalize common natural-language dates to YYYY-MM-DD."""
    if not date_text or not isinstance(date_text, str):
        return None

    base = base or date.today()
    s = date_text.strip()
    s_low = s.lower()

    # Regex for YYYY-MM-DD
    if re.fullmatch(r"\d{4}-\d{2}-\d{2}", s):
        return s

    _WEEKDAYS = {
        "monday": 0, "tuesday": 1, "wednesday": 2,
        "thursday": 3, "friday": 4, "saturday": 5, "sunday": 6
    }

    if s_low == "today":
        return base.strftime("%Y-%m-%d")
    if s_low == "tomorrow":
        return (base + timedelta(days=1)).strftime("%Y-%m-%d")
    if s_low == "next week":
        return (base + timedelta(days=7)).strftime("%Y-%m-%d")

    # Handle "in X days"
    m = re.fullmatch(r"in\s+(\d+)\s+days?", s_low)
    if m:
        return (base + timedelta(days=int(m.group(1)))).strftime("%Y-%m-%d")

    # Handle "next Friday"
    if s_low.startswith("next "):
        weekday = s_low.replace("next ", "", 1).strip()
        if weekday in _WEEKDAYS:
            today_wd = base.weekday()
            target_wd = _WEEKDAYS[weekday]
            delta = (target_wd - today_wd) % 7
            delta = 7 if delta == 0 else delta
            return (base + timedelta(days=delta)).strftime("%Y-%m-%d")

    # Handle common formats using datetime.strptime
    # نکته: اینجا از datetime.strptime استفاده می‌کنیم چون from datetime import datetime داریم
    formats = [
        "%B %d, %Y", "%B %d %Y", "%b %d, %Y", "%b %d %Y",
        "%d %B %Y", "%d %b %Y", "%m/%d/%Y", "%d/%m/%Y", "%Y/%m/%d"
    ]

    for fmt in formats:
        try:
            return datetime.strptime(s, fmt).date().strftime("%Y-%m-%d")
        except ValueError:
            continue

    return None

# ==========================================
# 3. The Main Tool
# ==========================================

@tool
def flight_search_tool(origin_city: str, destination_city: str, date_str: str):
    """
    Search for flight offers between two cities on a specific date.
    Args:
        origin_city: Name of the departure city (e.g., 'Tehran', 'London')
        destination_city: Name of the arrival city
        date_str: Date of flight (e.g., '2024-05-20', 'tomorrow', 'next Friday')
    """

    # 1. Get IATA Codes
    origin_code = get_iata_code(origin_city)
    dest_code = get_iata_code(destination_city)

    if not origin_code:
        return f"Error: Could not find IATA code for origin city '{origin_city}'. Please check the spelling."
    if not dest_code:
        return f"Error: Could not find IATA code for destination city '{destination_city}'. Please check the spelling."

    # 2. Normalize Date
    normalized_date = _normalize_yyyy_mm_dd(date_str)
    if not normalized_date:
        return f"Error: Date format '{date_str}' not understood. Please use YYYY-MM-DD or say 'tomorrow'."

    print(f"✈️ Searching flights: {origin_city} ({origin_code}) -> {destination_city} ({dest_code}) on {normalized_date}")

    # 3. Execute Search
    try:
        # Mock mode if Amadeus is not connected
        if not amadeus:
            print("⚠️ Amadeus client not available — returning mock flight results.")
            return "\n".join([
                f"Flight results for {origin_city} -> {destination_city} ({normalized_date}) [MOCK DATA]:",
                f"- Option 1: {origin_code} to {dest_code} | 08:00 AM | $250 | Airline: MockAir",
                f"- Option 2: {origin_code} to {dest_code} | 02:30 PM | $320 | Airline: TestFly",
            ])

        response = amadeus.shopping.flight_offers_search.get(
            originLocationCode=origin_code,
            destinationLocationCode=dest_code,
            departureDate=normalized_date,
            adults=1,
            max=5
        )

        if not response.data:
            return f"No flights found from {origin_city} to {destination_city} on {normalized_date}. Try a different date."

        lines = [f"Found flights for {origin_city} ({origin_code}) -> {destination_city} ({dest_code}) on {normalized_date}:"]

        for idx, offer in enumerate(response.data, start=1):
            price = offer.get("price", {}).get("total")
            currency = offer.get("price", {}).get("currency")
            itinerary = offer.get("itineraries")[0]
            segments = itinerary.get("segments")

            # Extract details
            carrier = segments[0].get("carrierCode")
            flight_num = segments[0].get("number")
            departure = segments[0].get("departure", {}).get("at").split("T")[1][:5]
            arrival = segments[-1].get("arrival", {}).get("at").split("T")[1][:5]
            duration = itinerary.get("duration")[2:] # Remove 'PT'
            stops = len(segments) - 1
            stop_text = "Nonstop" if stops == 0 else f"{stops} stop(s)"

            lines.append(
                f"{idx}. {carrier}{flight_num} | Dep: {departure} - Arr: {arrival} | "
                f"Dur: {duration} | {stop_text} | Price: {price} {currency}"
            )

        return "\n\n".join(lines)

    except ResponseError as error:
        return f"Amadeus API Error: {error.response.body if hasattr(error, 'response') else error}"
    except Exception as e:
        return f"Unknown system error during flight search: {str(e)}"
````

````python
def get_iata_code(city_name: str) -> str:
    try:
        response = amadeus.reference_data.locations.get(keyword=city_name, subType='CITY')
        if response.data:
            return response.data[0]['iataCode']
        return None
    except:
        return None
````

````python


@tool
def hotel_search_tool(city: str, check_in: str, check_out: str, budget: int = None):
    """
    Searches for hotel recommendations in a specific city for a given date range.
    Can filter by budget if provided. Uses Amadeus API or falls back to Web Search.

    Args:
        city: The name of the city (e.g., 'London').
        check_in: Check-in date in YYYY-MM-DD format.
        check_out: Check-out date in YYYY-MM-DD format.
        budget: Optional maximum total price (integer).
    """
    try:
        city_code = get_iata_code(city)
        if not city_code:
            raise ValueError(f"Code not found for {city}")

        print(f"Searching Amadeus for {city} ({city_code})...")

        search_params = {
            'cityCode': city_code,
            'checkInDate': check_in,
            'checkOutDate': check_out,
            'adults': 1,
            'radius': 5,
            'radiusUnit': 'KM',
            'paymentPolicy': 'NONE',
            'includeClosed': 'false',
            'bestRateOnly': 'true',
            'view': 'LIGHT',
            'max': 5
        }

        if budget:
            search_params['priceRange'] = f"1-{budget}"

        response = amadeus.shopping.hotel_offers_search.get(**search_params)

        if not response.data:
            raise ValueError("No hotels found in Amadeus.")

        results = [f"Recommended hotels in {city} (source: Amadeus):"]

        for offer in response.data:
            hotel = offer.get('hotel', {})
            offer_details = offer.get('offers', [])[0] if offer.get('offers') else {}
            name = hotel.get('name', 'Unknown')
            rating = hotel.get('rating', 'N/A')
            price_obj = offer_details.get('price', {})
            price = price_obj.get('total', 'N/A')
            currency = price_obj.get('currency', '')
            address = city_code
            results.append(
                f"{name}\n   - Rating: {rating} stars\n   - Total price: {price} {currency}\n   - Location: {address}"
            )

        return "\n\n".join(results)

    except (ResponseError, ValueError, Exception) as e:
        try:
            budget_str = f"under {budget} currency" if budget else ""
            query = f"Best hotels in {city} from {check_in} to {check_out} {budget_str} with price and rating"

            tavily_response = tavily_client.search(
                query=query,
                search_depth="advanced",
                max_results=5
            )

            results = [f"Recommended hotels in {city} (source: Web Search):\n(These results are gathered from the web due to booking limitations)\n"]

            for result in tavily_response.get('results', []):
                title = result.get('title')
                content = result.get('content')[:200]
                url = result.get('url')

                results.append(
                    f"{title}\n   - Description: {content}...\n   - Link: {url}"
                )

            return "\n\n".join(results)

        except Exception as tavily_error:
            return f"Unfortunately an error occurred in both search systems: {str(tavily_error)}"
````

````python

def get_currency_code(country_name: str) -> str:
    clean_name = country_name.strip().lower()
    manual_map = {
        "usa": "USD", "united states": "USD", "america": "USD",
        "uk": "GBP", "united kingdom": "GBP", "england": "GBP",
        "uae": "AED", "dubai": "AED",
        "iran": "IRR", "eu": "EUR", "europe": "EUR"
    }
    if clean_name in manual_map:
        return manual_map[clean_name]
    try:
        if 'df' in globals():
            result = df[df['Country'].str.lower() == clean_name]
            if not result.empty:
                return result.iloc[0]['Code']
    except Exception:
        pass
    return None
````

````python


@tool
def currency_info_tool(source_country: str, target_country: str):
    """
    Retrieves the current exchange rate and conversion info between two countries.
    Use this when a user asks about money, prices in different currencies, or exchange rates.
    """
    source_code = get_currency_code(source_country)
    target_code = get_currency_code(target_country)
    if not source_code:
        return f"Error: currency code for '{source_country}' not found. Please provide a more precise country name."
    if not target_code:
        return f"Error: currency code for '{target_country}' not found."
    try:
        query = f"{source_code} to {target_code} exchange rate today conversion"
        print(f"Converting {source_country} ({source_code}) to {target_country} ({target_code})...")
        response = tavily_client.search(
            query=query,
            search_depth="basic",
            max_results=3
        )
        results = response.get('results', [])
        if not results:
            return f"Unfortunately, the exchange rate for {source_code} to {target_code} was not found."
        top_result = results[0]
        content = top_result.get('content')
        url = top_result.get('url')
        return (
            f"Exchange rate info:\nFrom: {source_country} ({source_code})\nTo: {target_country} ({target_code})\n\n"
            f"Found info:\n{content}\n\nSource: {url}"
        )
    except Exception as e:
        return f"Error while fetching exchange rate: {str(e)}"
````

ابزار جستجوی FAQ یکی از ابزارهای کلیدی برای پاسخ به سوالات متداول مسافران است. این ابزار به کاربران اجازه می‌دهد بدون نیاز به جستجوی وب، پاسخ سوالات رایج خود را درباره سفر دریافت کنند. شما باید ابزاری تعریف کنید که از سیستم RAG و LanceDB برای جستجوی معنایی در پایگاه دانش FAQ استفاده کند و مرتبط‌ترین پاسخ‌ها را به کاربر بازگرداند.

- **محدود کردن نتایج:** از `limit(2)` استفاده کنید تا فقط 2 مرتبط‌ترین پرسش و پاسخ برگردانده شود و خروجی خلاصه باشد.

- **فرمت خروجی:** برای هر نتیجه:

  - فرمت واضح و خوانا

````python
@tool
def faq_search_tool(query: str):
    """
    Search the internal company FAQ and knowledge base to answer questions
    about policies, travel rules, and common procedures.
    """
    try:
        db = lancedb.connect("./lancedb_data")
        tables = set(db.list_tables()) if hasattr(db, "list_tables") else set(db.table_names())
        if "faq_table" not in tables:
            return "System error: FAQ table not found."

        tbl_faq = db.open_table("faq_table")
        query_vector = embedding_model.encode([query])[0]
        results = tbl_faq.search(query_vector).limit(2).to_pandas()
        if results.empty:
            return "No relevant FAQ found."

        formatted_response = "Based on our FAQ:\n\n"
        for _, row in results.iterrows():
            formatted_response += f"Q: {row['question']}\n"
            formatted_response += f"A: {row['answer']}\n"
            formatted_response += "-" * 30 + "\n"
        return formatted_response.strip()
    except Exception as e:
        return f"Error performing FAQ search: {str(e)}"
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Trip Planning</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
ابزار برنامه‌ریزی سفر یکی از قدرتمندترین ابزارهای تراول‌بات است که با استفاده از سیستم RAG و پایگاه دانش راهنمای سفر، یک برنامه کامل و جامع برای سفر کاربر ایجاد می‌کند. شما باید ابزاری تعریف کنید که اطلاعات سفر را دریافت کرده و با جستجو در پایگاه دانش، یک itinerary روزانه همراه با پیشنهادات جاذبه‌ها، رستوران‌ها و فعالیت‌ها تولید کند.

**نکات مهم پیاده‌سازی:**

- **استفاده از RAG System:** از LanceDB و اطلاعاتی که از کتاب جهانگردی امبد کرده‌اید برای جستجوی اطلاعات مرتبط با مقصد استفاده کنید. چندین query مختلف بزنید تا اطلاعات جامعی جمع‌آوری شود.

- **پارامترهای ورودی:**

  - `destination`: نام شهر یا کشور مقصد (الزامی)
  - `arrival_date`: تاریخ ورود
  - `departure_date`: تاریخ خروج
  - `interests`: علایق کاربر مانند culture، food، nature، adventure، shopping، nightlife

- **محاسبه تعداد روزها:** با استفاده از `datetime` تعداد روزهای سفر را محاسبه کنید تا itinerary مناسب بسازید.

- **جستجوهای متنوع:** برای جمع‌آوری اطلاعات کامل، چندین query به پایگاه دانش بزنید:

  - جاذبه‌های اصلی و دیدنی‌ها
  - بهترین رستوران‌ها و غذاهای محلی
  - فعالیت‌ها و تفریحات
  - نواحی مناسب اقامت
  - (در صورت وجود interests) موضوعات خاص کاربر

- **مدیریت اطلاعات ناکافی:** اگر اطلاعات کافی در پایگاه دانش یافت نشد، از Tavily Search استفاده کنید.

- **مدیریت خطا:** خطاهای احتمالی خطای جستجو در پایگاه داده را مدیریت کنید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید.

**مثال درخواست کاربر:**

```
"Plan my trip to Paris from December 18th to 22th"
"Create a 4-day itinerary for Dubai with focus on food and culture"
```

````python
from langchain.tools import tool
# اصلاح مهم: ایمپورت مستقیم کلاس datetime
from datetime import datetime
import lancedb
# (فرض بر این است که embedding_model و tavily_client در محیط تعریف شده‌اند)
# اگر تعریف نشده‌اند، کدهای ماک قبلی را قبل از این سلول اجرا کنید.

@tool
def trip_planning_tool(destination: str, arrival_date: str, departure_date: str, interests: str = ""):
    """
    Creates a comprehensive day-by-day trip itinerary based on destination and dates.
    """
    try:
        # اصلاح شده: استفاده مستقیم از datetime که ایمپورت شده
        start = datetime.strptime(arrival_date, "%Y-%m-%d")
        end = datetime.strptime(departure_date, "%Y-%m-%d")

        duration = (end - start).days + 1
        if duration < 1:
            return "Error: Departure date must be after arrival date."

        season = start.strftime("%B") # مثلا نام ماه (May)
    except ValueError:
        return "Error: Date format must be YYYY-MM-DD."

    print(f"🗺️ Planning a {duration}-day trip to {destination} ({season})...")

    # لیست جستجوها
    search_queries = [
        f"top tourist attractions in {destination}",
        f"best local food in {destination}",
        f"history of {destination}",
    ]
    if interests:
        search_queries.append(f"best {interests} activities in {destination}")

    rag_results = []

    # --- 1. LanceDB Search (RAG) ---
    try:
        db = lancedb.connect("./lancedb_data")

        # اصلاحیه: به جای چک کردن لیست جدول‌ها، مستقیم تلاش می‌کنیم بازش کنیم
        # این روش امن‌تر است و ارور unhashable نمی‌دهد
        try:
            tbl = db.open_table("book_table")

            for q in search_queries:
                if 'embedding_model' in globals():
                    q_vec = embedding_model.encode([q])[0]
                    results = tbl.search(q_vec).limit(2).to_pandas()
                    for _, row in results.iterrows():
                        if row['_distance'] < 0.6:
                            rag_results.append(f"- {row['text']}")
        except:
            # اگر جدول وجود نداشت یا باز نشد، مهم نیست، برو بعدی
            pass

    except Exception as e:
        print(f"⚠️ LanceDB skipped: {e}")
    # --- 2. Web Search (Fallback) ---
    sources = []
    # اگر اطلاعات دیتابیس کم بود، برو سراغ وب
    if len(rag_results) < 3:
        try:
            # استفاده از کلاینت Tavily که گلوبال تعریف شده
            if 'tavily_client' in globals():
                for q in search_queries:
                    # مکس ریزالت کم برای سرعت
                    search_resp = tavily_client.search(query=q, max_results=1)
                    for r in search_resp.get('results', []):
                        sources.append(r.get('content')[:200] + "...") # خلاصه متن
        except Exception as e:
            print(f"⚠️ Web search skipped: {e}")

    # --- 3. Build Itinerary ---
    itinerary = f"Trip Itinerary for {destination} ({arrival_date} to {departure_date}):\n\n"

    if rag_results:
        itinerary += "📚 **Guide Book Insights**:\n" + "\n".join(list(set(rag_results))[:5]) + "\n\n"

    if sources:
        itinerary += "🌐 **Web Insights**:\n" + "\n".join(sources[:3]) + "\n\n"

    if not rag_results and not sources:
        itinerary += "No specific data found, but based on general knowledge: Visit the city center and try local food!"

    return itinerary
````

````python
result = trip_planning_tool.invoke({
    "destination": "Paris",
    "arrival_date": "2025-06-01",
    "departure_date": "2025-06-05",
    "interests": "museums"
})

print("\nRESULT OUTPUT:\n")
print(result)
````

# ReAct Paper Explanation

The paper "ReAct: SYNERGIZING REASONING AND ACTING IN LANGUAGE MODELS" by Shunyu Yao, Jeffrey Zhao, Dian Yu, Nan Du, Izhak Shafran, Karthik Narasimhan, and Yuan Cao, published at ICLR 2023, introduces a novel paradigm called ReAct (Reasoning + Acting) for enhancing the capabilities of large language models (LLMs) in task-solving scenarios. Below, I provide a comprehensive, point-by-point explanation of the paper's key elements, structured around its main sections and contributions. This covers the motivation, methodology, experiments, results, and implications, drawing directly from the paper's content.

## 1. Abstract and Core Motivation
- **Human-Inspired Synergy**: The paper draws inspiration from human cognition, where verbal reasoning (e.g., inner speech or "thinking aloud") seamlessly integrates with physical actions (e.g., cooking a meal by reasoning about steps while interacting with tools). Humans reason to plan, handle exceptions, and gather information, while actions provide feedback to refine reasoning. LLMs, however, have been studied separately for reasoning (e.g., chain-of-thought prompting) or acting (e.g., generating action plans), leading to limitations like hallucination in reasoning or poor planning in acting.
- **ReAct's Innovation**: ReAct prompts LLMs to generate both reasoning traces (thoughts) and task-specific actions in an interleaved manner. This creates synergy: reasoning helps induce, track, and update action plans, while actions allow interaction with external sources (e.g., knowledge bases) to ground reasoning in reality.
- **Applications and Benefits**: Applied to diverse tasks (question answering, fact verification, text-based games, web navigation), ReAct outperforms baselines, improves interpretability/trustworthiness, and enables human-aligned, controllable behavior. It uses few-shot prompting (1-6 examples) and shows robustness across tasks.
- **Key Contributions**:
  - Introduces ReAct as a general paradigm for synergizing reasoning and acting.
  - Extensive experiments demonstrating advantages over isolated reasoning/acting methods.
  - Ablations showing the value of combining internal knowledge (model's pretraining) with external interactions.
  - Analysis of limitations (e.g., reliance on prompting) and potential for finetuning.

## 2. Introduction
- **Problem Setup**: LLMs excel at language understanding and decision-making but struggle with integrated reasoning and acting. Chain-of-thought (CoT) reasoning is static and black-box, prone to hallucination (fabricating facts) and error propagation. Acting methods (e.g., planning in environments) lack abstract reasoning for goals or working memory.
- **ReAct Overview**: ReAct augments the LLM's action space with a language space for "thoughts" (reasoning traces). Thoughts are interleaved with actions, allowing dynamic reasoning-to-act (e.g., planning based on thoughts) and act-to-reason (e.g., using external feedback to update thoughts). This mirrors human cognition for robust, adaptive task-solving.
- **Evaluation Domains**: Tested on four benchmarks:
  - **HotpotQA**: Multi-hop question answering requiring reasoning over Wikipedia passages.
  - **Fever**: Fact verification (SUPPORTS/REFUTES/NOT ENOUGH INFO) based on evidence.
  - **AlfWorld**: Text-based household tasks (e.g., "put a clean lettuce in diningtable") in a simulated environment.
  - **WebShop**: Real-world web navigation for shopping (e.g., buying products matching user instructions).
- **Results Summary**: ReAct achieves strong performance with few-shot prompting, e.g., 34% success rate improvement on AlfWorld over imitation learning, and 10% on WebShop. It also enhances interpretability, allowing humans to inspect reasoning and correct errors.
- **Broader Impact**: ReAct could unlock LLMs for more applications, combining with reinforcement learning for better agents.

## 3. ReAct: SYNERGIZING REASONING + ACTING
- **Formal Setup**: In a standard agent-environment loop, the agent observes (ot), takes actions (at), and updates context (ct). ReAct augments the action space A to A ∪ L, where L is the language space for thoughts (aˆt ∈ L). Thoughts reason over ct to update it (ct+1 = ct ∪ aˆt) without external effects, enabling planning, exception handling, and knowledge integration.
- **Prompting Approach**: Uses frozen LLMs (e.g., PaLM-540B) prompted with few-shot examples. Each example is a trajectory of interleaved thoughts, actions, and observations. For knowledge tasks (e.g., HotpotQA), thoughts are dense (every few steps); for decision-making (e.g., AlfWorld), thoughts are sparse (at key points) to let the model decide their placement.
- **Design Principles**:
  - **Intuitive and Easy**: Annotators simply write thoughts alongside actions, no complex formats.
  - **General and Flexible**: Works across tasks with varying action spaces and reasoning needs.
  - **Performant and Robust**: Few-shot learning with generalization; robust to prompt variations.
  - **Human-Aligned and Controllable**: Trajectories are interpretable; humans can edit thoughts for corrections (e.g., fixing hallucinations).
- **Comparison to Baselines**:
  - **Standard**: No thoughts/actions (just input-output).
  - **CoT (Reason-Only)**: Thoughts only, no actions.
  - **Act (Action-Only)**: Actions only, no thoughts.
  - ReAct interleaves both, outperforming each.

## 4. Knowledge-Intensive Reasoning Tasks (HotpotQA and Fever)
- **Setup**: Tasks require external knowledge; models access a simple Wikipedia API with actions: search[entity] (returns top sentences), lookup[string] (next sentence containing string), finish[answer].
- **ReAct Prompting**: Uses 6 (HotpotQA) or 3 (Fever) annotated trajectories with dense thought-action-observation interleaving. Thoughts handle decomposition, commonsense, search reformulation, and synthesis.
- **Baselines**:
  - **CoT**: Thoughts only.
  - **CoT-SC**: Self-consistency with 21 CoT samples and majority voting.
  - **Act**: Actions only.
- **Hybrid Methods**: Combines ReAct and CoT-SC for robustness:
  - **ReAct → CoT-SC**: If ReAct fails to answer within steps (7 for HotpotQA, 5 for Fever), fall back to CoT-SC.
  - **CoT-SC → ReAct**: If CoT-SC confidence is low (majority <50%), fall back to ReAct.
- **Finetuning**: Bootstraps 3,000 ReAct-generated trajectories to finetune smaller models (PaLM-8/62B), showing ReAct's superiority over baselines.
- **Results and Analysis**:
  - **Performance**: ReAct > Act; competitive with CoT; hybrids (e.g., ReAct → CoT-SC) best overall (e.g., 35.1% EM on HotpotQA). Finetuning boosts ReAct most.
  - **Success/Failure Modes**: ReAct reduces hallucination (6% false positives vs. 14% in CoT) but has more reasoning errors (e.g., loops). CoT has flexible reasoning but high hallucination (56% failures). Hybrids mitigate trade-offs.
  - **Ablations**: Acting grounds reasoning; reasoning guides better searches. External knowledge (via API) improves factuality over internal-only CoT.

## 5. Decision Making Tasks (AlfWorld and WebShop)
- **AlfWorld**: 134 tasks across 6 types (e.g., pick/clean/heat/cool/look). Actions: navigate, interact (e.g., take, clean). ReAct uses sparse thoughts for goal decomposition, progress tracking, commonsense (e.g., where items are likely found).
- **WebShop**: 500 instructions for product purchase. Actions: search, click, buy. ReAct reasons to explore options, match attributes, decide when to buy.
- **Baselines**:
  - **BUTLER**: Imitation learning on 105 expert trajectories per task.
  - **IL/IL+RL**: Imitation learning ± reinforcement learning on WebShop.
  - **Act**: Actions only.
- **ReAct Prompting**: 3 trajectories per AlfWorld task type; sparse thoughts. For WebShop, reasoning bridges noisy text to actions.
- **Ablation (ReAct-IM)**: Uses dense feedback thoughts (e.g., "current subgoal") instead of flexible reasoning. Shows ReAct > ReAct-IM (71% vs. 53% success on AlfWorld), highlighting internal reasoning's value over simple reactions.
- **Results**:
  - **AlfWorld**: ReAct (best prompt: 71% success) > Act (45%) > BUTLER (37%). Consistent gains across tasks.
  - **WebShop**: ReAct (40% success) > Act (30%) > IL/IL+RL (29-30%). Improves attribute matching.
  - **Qualitative**: ReAct avoids loops/errors in Act; enables subgoal tracking and exploration.

## 6. Related Work
- **Reasoning in LLMs**: CoT and variants (e.g., self-consistency, least-to-most) focus on isolated reasoning. ReAct integrates reasoning with actions for grounded, dynamic processes.
- **Acting in LLMs**: Methods like WebGPT, SayCan, Inner Monologue generate actions but lack abstract reasoning. ReAct's flexible thoughts surpass simple feedback (e.g., vs. Inner Monologue).
- **Broader**: Builds on embodied AI (e.g., ALFRED) and generalist agents, emphasizing language as a cognitive tool.

## 7. Conclusion and Limitations
- **Contributions Recap**: ReAct paradigm, empirical advantages, ablations, and analysis.
- **Limitations**: Prompting limits scalability (long contexts); needs more human data for finetuning. Not tested on non-language actions.
- **Future Work**: Scale with multi-task training, combine with RL, explore human-in-the-loop editing for alignment.

## Overall Implications
ReAct demonstrates that interleaving reasoning and acting makes LLMs more reliable, interpretable, and capable of complex tasks. It reduces hallucinations via external grounding and enables adaptive planning. While prompting-based, it hints at finetuning potential. This work advances LLM applications in AI agents, emphasizing human-like cognition for safer, more controllable systems.

    <li><b>نوآوری ReAct:</b> مدل به‌جای فقط فکر کردن یا فقط عمل کردن، هر دو را با هم انجام می‌دهد. مثلاً ابتدا فکر می‌کند چه باید بکند، سپس یک عمل (مثل جستجو) انجام می‌دهد، نتیجه را می‌بیند و دوباره فکر می‌کند.</li>
    <li><b>کاربردها:</b> این روش در کارهایی مثل پاسخ به سوالات چندمرحله‌ای، راستی‌آزمایی حقایق، بازی‌های متنی و خرید آنلاین تست شده و نسبت به روش‌های قبلی عملکرد بهتری داشته است.</li>

````python
from langchain_openai import ChatOpenAI  # <--- THIS IS THE MISSING PIECE
OPENAI_API_KEY = "[REDACTED_OPENAI_KEY]"
OPENAI_API_BASE = "https://api.openai.com/v1"
os.environ["OPENAI_API_KEY"] = OPENAI_API_KEY


llm = ChatOpenAI(
    model="gpt-5",
    temperature=0,
    openai_api_key=OPENAI_API_KEY,
    openai_api_base=OPENAI_API_BASE
)
````

````python

current_date = datetime.now().strftime("%Y-%m-%d") # کلاس.متد
system_prompt_text = f"""You are an intelligent and resourceful **Travel Assistant**. Your primary goal is to help users plan their trips, book flights/hotels, check weather, and answer travel-related questions accurately.

**Current Date:** {current_date}

### AVAILABLE TOOLS:
1.  **flight_search_tool**: Use for finding flight options. Requires origin, destination, and date (YYYY-MM-DD).
2.  **hotel_search_tool**: Use for finding accommodation. Requires city, check-in, and check-out dates (YYYY-MM-DD).
3.  **restaurant_search_tool**: Search for food and dining recommendations in a specific city.
4.  **weather_info_tool**: Get weather forecast for a destination and date.
5.  **currency_info_tool**: Get current exchange rates between currencies/countries.
6.  **faq_search_tool**: Answer general travel questions (visa, baggage, etc.) using the internal knowledge base.
7.  **trip_planning_tool**: Generate full daily itineraries based on user interests and dates.

### IMPORTANT RULES:
-   **Tool Use**: Always choose the single most relevant tool. Do not call multiple tools unless the user explicitly requests multiple distinct tasks.
-   **Date Handling**: The user might use relative dates like "next Friday", "tomorrow", "in 3 days", or vague ranges like "next week".
    - You MUST calculate the exact date(s) based on the **Current Date** ({current_date}) and convert to **YYYY-MM-DD** before calling any tool.
    - If the user gives a vague range (e.g., "next week") and they are *not* trying to book a ticket, infer a reasonable default and proceed with helpful suggestions.
      Example: for "a 2-day layover in Dubai next week", assume a 2-day window starting {current_date} + 7 days, and provide an initial itinerary. You may ask a follow-up for exact dates only to refine details.
-   **City Codes**: When using flight/hotel tools, ensure city names map to IATA codes (e.g., Tehran → IKA, Dubai → DXB).
-   **Safety/Accuracy**: If information is uncertain or depends on nationality (e.g., visas), ask clarifying questions but still provide general guidance.

You should respond clearly and concisely.
"""
````

````python

tools = [
    flight_search_tool,
    hotel_search_tool,
    restaurant_search_tool,
    weather_info_tool,
    currency_info_tool,
    faq_search_tool,
    trip_planning_tool
]
````

````python


print(f"Tools list created with {len(tools)} tools.")
llm_with_tools = llm.bind_tools(tools)
print(f"Successfully bound {len(tools)} tools to the LLM model.")
print(f"System date set to {current_date}.")
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Creating Graph Nodes</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
Reasoner node را پیاده‌سازی کنید که:
- System message و تاریخچه مکالمه را ترکیب کند
- LLM را invoke کند
- پاسخ را برگرداند
</p>
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
تعریف موفق reasoner node
</p>


````python
from typing import TypedDict, Annotated
from langgraph.graph import add_messages

class AgentState(TypedDict):
    messages: Annotated[list, add_messages]

def reasoner_node(state: AgentState):
    messages = state["messages"]
    full_history = [SystemMessage(content=system_prompt_text)] + messages
    response = llm_with_tools.invoke(full_history)
    return {"messages": [response]}

````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Creating and Compiling Graph</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
StateGraph را بسازید و compile کنید و تصویر گراف را رسم کنید.

</p>
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
• Graph compiled موفق<br>
• تصویر visualization گراف<br>
</p>


````python
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode, tools_condition
from IPython.display import Image, display

workflow = StateGraph(AgentState)
workflow.add_node("reasoner", reasoner_node)
workflow.add_node("tools", ToolNode(tools))
workflow.set_entry_point("reasoner")
workflow.add_conditional_edges("reasoner", tools_condition)
workflow.add_edge("tools", "reasoner")
app = workflow.compile()

try:
    display(Image(app.get_graph().draw_mermaid_png()))
except Exception as e:
    print("Could not render graph (pygraphviz may need to be installed).")
````

````python
import uuid
from langchain_core.messages import BaseMessage, FunctionMessage, HumanMessage, SystemMessage, AIMessage

def chat_cli():
    print("Travel Bot CLI")
    print("Type 'exit' to quit, 'clear' to reset chat.")
    config = {"configurable": {"thread_id": str(uuid.uuid4())}}
    current_chat_history = []
    while True:
        try:
            user_input = input("\nYou: ").strip()
            if user_input.lower() in ["exit", "quit"]:
                print("Goodbye!")
                break
            if user_input.lower() == "clear":
                current_chat_history = []
                print("Memory cleared.")
                continue
            if not user_input:
                continue
            print("Bot is thinking...", end="\r")
            inputs = {"messages": [HumanMessage(content=user_input)]}
            final_state = app.invoke(inputs, config=config)
            all_messages = final_state["messages"]
            last_message = all_messages[-1]
            tool_calls = [m.name for m in all_messages if hasattr(m, 'tool_call_id') and m.type == 'tool']
            if tool_calls:
                print(f"Tools used: {', '.join(tool_calls)}")
            print(f"Bot: {last_message.content}")
        except Exception as e:
            print(f"Error: {e}")
````

# <h1 style="text-align: right;">**نکات مهم و قوانین تحویل**</h1>


<h4 dir="rtl" style="font-family: Vazir; width: 85%;">فایل ارسالی شما باید با فرمت زیر نامگذاری شود: <code>NLP_CA{n}_{LASTNAME}_{STUDENTID}.ipynb</code></h4>
<h4 dir="rtl" style="font-family: Vazir; width: 85%;">نحوه انجام تمرین:</h4>
<ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;">
  <li>سلول‌های کد با برچسب <code>WRITE YOUR CODE HERE</code> را تکمیل کنید.</li>
  <li>برای پاسخ‌های متنی، متن <code>{{پاسخ_خود_را_اینجا_بنویسید}}</code> را با پاسخ خود جایگزین کنید.</li>
</ul>
<h4 dir="rtl" style="font-family: Vazir; width: 85%;">صداقت علمی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;"> <li>ما نوت‌بوک‌های تعداد مشخصی از دانشجویان که به صورت تصادفی انتخاب می‌شوند، بررسی خواهیم کرد. این بررسی‌ها اطمینان حاصل می‌کنند که کدی که نوشتید واقعاً پاسخ‌های موجود در نوت‌بوک شما را تولید می‌کند. اگر پاسخ‌های صحیح را در نوت‌بوک خود بدون کدی که واقعاً آن پاسخ‌ها را تولید کند تحویل دهید، این یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> <li>ما همچنین بررسی‌های خودکاری را برای تشخیص سرقت علمی در نوت‌بوک‌های کولب انجام خواهیم داد. کپی کردن کد از دیگران نیز یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> </ul>
<h4 dir="rtl" style="font-family: Vazir; width: 85%;">توضیحات تکمیلی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;">
<li>
خوانایی و دقت بررسی‌ها در گزارش نهایی از اهمیت ویژه‌ای برخوردار است. به تمرین‌هایی که به صورت کاغذی تحویل داده شوند یا به صورت عکس در سایت بارگذاری شوند، ترتیب اثری داده نخواهد شد.</li>
<li>
 همه‌ی کدهای پیوست گزارش بایستی قابلیت اجرای مجدد داشته باشند. در صورتی که برای اجرا مجدد آن‌ها نیاز به تنظیمات خاصی می‌باشد، بایستی تنظیمات مورد نیاز را نیز در گزارش خود ذکر کنید.  دقت کنید که  تمامی کدها باید توسط شما اجرا شده باشند و نتایج اجرا در فایل کدهای ارسالی مشخص باشد. به کدهایی که نتایج اجرای آن‌ها در فایل ارسالی مشخص نباشد نمره‌ای تعلق نمی‌گیرد.
</li>
<li>توجه کنید این تمرین باید به صورت تک‌نفره انجام شود و پاسخ‌های ارائه شده باید نتیجه فعالیت فرد نویسنده باشد (همفکری و به اتفاق هم نوشتن تمرین نیز ممنوع است). در صورت مشاهده
