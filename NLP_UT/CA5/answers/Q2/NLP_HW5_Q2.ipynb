<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">

````python
path = kagglehub.dataset_download("phanee16/currency-and-country-code-mapping")
print(f"Dataset downloaded to: {path}")

csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
````

````python

if csv_files:
    target_csv = next((f for f in csv_files if 'country_code_to_currency_code' in f), csv_files[0])
    csv_file_path = os.path.join(path, target_csv)
    currency_df = pd.read_csv(csv_file_path)
    currency_df.columns = [c.lower().strip() for c in currency_df.columns]
    print(f"Detected columns in {target_csv}: {list(currency_df.columns)}")
    country_col = next((c for c in currency_df.columns if 'country' in c and 'code' not in c), None)
    currency_code_col = next((c for c in currency_df.columns if 'currency_code' in c or 'currency' == c), None)
    if country_col and currency_code_col:
        print(f"Using columns: Country='{country_col}', CurrencyCode='{currency_code_col}'")
        for index, row in currency_df.iterrows():
            country_name = str(row[country_col]).lower().strip()
            currency_code = str(row[currency_code_col]).strip()
            if country_name and currency_code and currency_code.lower() != 'nan':
                country_to_currency[country_name] = currency_code
        print(f"Number of currencies: {len(currency_df[currency_code_col].dropna().unique())}")
        print(f"Number of entries in mapping dictionary: {len(country_to_currency)}")
        print(
            f"Sample search: IRAN -> {country_to_currency.get('iran')}, UAE -> {country_to_currency.get('united arab emirates')}"
        )
    else:
        print("Could not automatically find 'Country' or 'Currency Code' columns.")
else:
    print("Error: No CSV file found in the downloaded folder.")
````

````python
faq_data = []
import json
import re
import os

possible_paths = ["./FAQ.js", os.path.join(os.getcwd(), "FAQ.js")]
faq_file_path = next((p for p in possible_paths if os.path.exists(p)), None)

if faq_file_path:
    try:
        with open(faq_file_path, 'r', encoding='utf-8') as f:
            content = f.read()
            match = re.search(r'\[\s*{.*}\s*\]', content, re.DOTALL)
            if match:
                json_string = match.group(0)
                faq_data = json.loads(json_string)
                print(f"Successfully loaded {len(faq_data)} FAQs from {faq_file_path}")
                if faq_data:
                    print("Sample FAQ entry:")
                    for key, value in list(faq_data[0].items())[:5]:
                        if isinstance(value, str):
                            print(f"  {key}: {value[:70]}...")
                        else:
                            print(f"  {key}: {value}")
            else:
                print(f"Could not find a JSON array in {faq_file_path}.")
    except json.JSONDecodeError as e:
        print(f"Error decoding JSON from {faq_file_path}: {e}")
    except Exception as e:
        print(f"Unexpected error while loading {faq_file_path}: {e}")
else:
    print("FAQ.js not found in the current directory. Creating a small fallback FAQ dataset for testing.")
    faq_data = [
        {"question": "Can I bring food on the plane?", "answer": "In general, you can bring solid food. Liquids must respect security limits."},
        {"question": "What documents do I need for international travel?", "answer": "Passport, visa (if required), travel insurance, and any required vaccination certificates."},
        {"question": "How early should I arrive at the airport?", "answer": "For international flights, arrive at least 3 hours before departure; for domestic, 1.5-2 hours."}
    ]
    print(f"Fallback FAQ dataset created with {len(faq_data)} entries.")
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Flight Search</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
یکی از ابزارهای مهمی که تراول‌بات با استفاده از آن به کاربران خدمات می‌دهد، ابزار جستجوی پرواز است. شما باید ابزاری تعریف کنید که پارامترهایی مانند مبدا، مقصد و تاریخ پرواز را از کاربر دریافت کرده و با استفاده از Amadeus API به جستجوی پروازهای مورد نظر بپردازد. این ابزار باید حداقل ۳ پرواز را با اطلاعات کاملی شامل شماره پرواز، کد ایرلاین، قیمت دقیق (به همراه واحد پول)، تاریخ و ساعت حرکت، مدت پرواز و تعداد توقف‌ها به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **تبدیل نام شهر به کد IATA:** ابزار Amadeus بر اساس کدهای یکتای فرودگاه (IATA Code) کار می‌کند. از تابع مپینگی که در بخش قبل نوشته‌اید استفاده کنید تا نام شهرها را به کد فرودگاه تبدیل کنید (مثلاً Tehran → IKA، Dubai → DXB).

- **مدیریت فرمت تاریخ:** Amadeus تاریخ را به فرمت YYYY-MM-DD می‌پذیرد. از آنجایی که ماهیت چت‌بات صحبت به زبان طبیعی است، کاربران فرمت تاریخ را به گونه‌های متفاوتی وارد می‌کنند(برای مثال به صورت متن و نه عدد). شما باید مدیریت فرمت تاریخ در ابزار و System Prompt را مد نظر داشته باشید.

- **مدیریت خطا:** در صورت بروز خطا در جستجو (مثل عدم یافتن پرواز یا خطای API)، پیام خطای مناسب و قابل فهم به کاربر بازگردانید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید تا LangGraph بتواند آن را شناسایی کند.

</div>


````python
def get_iata_code(city_name: str) -> str:
    city_name_clean = city_name.strip().lower()
    manual_map = {
        'tehran': 'IKA', 'mashhad': 'MHD', 'dubai': 'DXB', 'london': 'LON', 'paris': 'PAR', 'istanbul': 'IST', 'rome': 'ROM'
    }
    if city_name_clean in manual_map:
        return manual_map[city_name_clean]
    if amadeus:
        try:
            response = amadeus.reference_data.locations.get(keyword=city_name, subType='CITY')
            if response.data:
                return response.data[0]['iataCode']
        except Exception as e:
            print(f"Amadeus lookup failed for {city_name}: {e}")
    return None
````

````python

@tool
def flight_search_tool(origin_city: str, destination_city: str, date: str):
    origin_code = get_iata_code(origin_city)
    dest_code = get_iata_code(destination_city)

    if not origin_code:
        return f"Error: IATA code for origin city '{origin_city}' not found."
    if not dest_code:
        return f"Error: IATA code for destination city '{destination_city}' not found."

    print(f"Searching flights: {origin_city} ({origin_code}) -> {destination_city} ({dest_code}) on {date}")

    try:
        if not amadeus:
            print("Amadeus client not available — returning mock flight results.")
            mock_results = [f"Sample flight options (mock) {origin_city} -> {destination_city} ({date}):"]
            mock_results.extend([
                "Option 1: Airline: AB 123, Price: 250 USD, Time: 09:00, Duration: 3h, Nonstop",
                "Option 2: Airline: CD 456, Price: 320 USD, Time: 13:30, Duration: 4h, 1 stop",
                "Option 3: Airline: EF 789, Price: 400 USD, Time: 21:10, Duration: 6h, 2 stops"
            ])
            return "\n\n".join(mock_results)

        response = amadeus.shopping.flight_offers_search.get(
            originLocationCode=origin_code,
            destinationLocationCode=dest_code,
            departureDate=date,
            adults=1,
            max=3
        )

        if not response.data:
            return f"No flights found from {origin_city} to {destination_city} on {date}."

        results = []
        results.append(f"Flight results {origin_city} to {destination_city} ({date}):")

        for idx, offer in enumerate(response.data):
            price = offer['price']['total']
            currency = offer['price']['currency']
            itinerary = offer['itineraries'][0]
            duration = itinerary['duration'].replace("PT", "").lower()
            segments = itinerary['segments']
            stops = len(segments) - 1
            stop_text = "Nonstop" if stops == 0 else f"{stops} stop(s)"
            first_segment = segments[0]
            airline_code = first_segment['carrierCode']
            flight_number = f"{airline_code} {first_segment['number']}"
            departure_time = first_segment['departure']['at'].split("T")[1][:5]
            results.append(
                f"Option {idx + 1}:\n   - Airline: {airline_code} ({flight_number})\n   - Price: {price} {currency}\n   - Time: {departure_time} | Duration: {duration} | {stop_text}"
            )

        return "\n\n".join(results)

    except ResponseError as error:
        return f"Amadeus API error: {error}"
    except Exception as e:
        return f"Unknown error: {str(e)}"

````

````python

tomorrow = (datetime.date.today() + datetime.timedelta(days=1)).strftime("%Y-%m-%d")

test_result = flight_search_tool.invoke({
    "origin_city": "London",
    "destination_city": "Paris",
    "date": tomorrow
})

print(test_result)
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Hotel Search</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
یکی دیگر از ابزارهای کاربردی تراول‌بات، جستجوی هتل است. شما باید ابزاری تعریف کنید که مقصد، تاریخ ورود (check-in) و تاریخ خروج (check-out) را از کاربر دریافت کرده و با استفاده از Amadeus API به یافتن هتل‌های مناسب در مقصد مورد نظر بپردازد. این ابزار باید لیستی از هتل‌های پیشنهادی را همراه با نام هتل، رتبه (rating)، قیمت شب، موقعیت مکانی به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **استفاده از Amadeus Hotel Search API:** از endpoint مناسب Amadeus برای جستجوی هتل استفاده کنید. این API نیاز به کد شهر (IATA City Code) دارد که باید از mapping قبلی استفاده کنید.

- **تبدیل نام شهر به کد IATA:** مشابه جستجوی پرواز، از تابع از تابع مپینگ هوشمند برای تبدیل نام شهر به کد فرودگاه استفاده کنید (مثلاً Dubai → DXB).

- **فرمت تاریخ:** Amadeus تاریخ را به فرمت YYYY-MM-DD می‌پذیرد. از آنجایی که ماهیت چت‌بات صحبت به زبان طبیعی است، کاربران فرمت تاریخ را به گونه‌های متفاوتی وارد می‌کنند(برای مثال به صورت متن و نه عدد). شما باید مدیریت فرمت تاریخ در ابزار و System Prompt را مد نظر داشته باشید.

- **پارامتر Budget (اختیاری):** این ابزار می‌تواند یک پارامتر اختیاری budget داشته باشد که هتل‌ها را بر اساس محدوده قیمتی فیلتر کند.

- **خروجی کامل:** برای هر هتل، اطلاعاتی مانند نام، آدرس، رتبه و قیمت هر شب (به همراه واحد پول) را نمایش دهید.

- **مدیریت خطا:** در صورت عدم یافتن هتل یا خطای API، پیام مناسب به کاربر بازگردانید.

- **Fallback به Tavily:** در صورتی که Amadeus Hotel API در دسترس نبود یا محدودیت داشت، می‌توانید به عنوان fallback از Tavily Search استفاده کنید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید.
<div>


````python
def get_iata_code(city_name: str) -> str:
    try:
        response = amadeus.reference_data.locations.get(keyword=city_name, subType='CITY')
        if response.data:
            return response.data[0]['iataCode']
        return None
    except:
        return None
````

````python


@tool
def hotel_search_tool(city: str, check_in: str, check_out: str, budget: int = None):
    try:
        city_code = get_iata_code(city)
        if not city_code:
            raise ValueError(f"Code not found for {city}")

        print(f"Searching Amadeus for {city} ({city_code})...")

        search_params = {
            'cityCode': city_code,
            'checkInDate': check_in,
            'checkOutDate': check_out,
            'adults': 1,
            'radius': 5,
            'radiusUnit': 'KM',
            'paymentPolicy': 'NONE',
            'includeClosed': 'false',
            'bestRateOnly': 'true',
            'view': 'LIGHT',
            'max': 5
        }

        if budget:
            search_params['priceRange'] = f"1-{budget}"

        response = amadeus.shopping.hotel_offers_search.get(**search_params)

        if not response.data:
            raise ValueError("No hotels found in Amadeus.")

        results = [f"Recommended hotels in {city} (source: Amadeus):"]

        for offer in response.data:
            hotel = offer.get('hotel', {})
            offer_details = offer.get('offers', [])[0] if offer.get('offers') else {}
            name = hotel.get('name', 'Unknown')
            rating = hotel.get('rating', 'N/A')
            price_obj = offer_details.get('price', {})
            price = price_obj.get('total', 'N/A')
            currency = price_obj.get('currency', '')
            address = city_code
            results.append(
                f"{name}\n   - Rating: {rating} stars\n   - Total price: {price} {currency}\n   - Location: {address}"
            )

        return "\n\n".join(results)

    except (ResponseError, ValueError, Exception) as e:
        try:
            budget_str = f"under {budget} currency" if budget else ""
            query = f"Best hotels in {city} from {check_in} to {check_out} {budget_str} with price and rating"

            tavily_response = tavily_client.search(
                query=query,
                search_depth="advanced",
                max_results=5
            )

            results = [f"Recommended hotels in {city} (source: Web Search):\n(These results are gathered from the web due to booking limitations)\n"]

            for result in tavily_response.get('results', []):
                title = result.get('title')
                content = result.get('content')[:200]
                url = result.get('url')

                results.append(
                    f"{title}\n   - Description: {content}...\n   - Link: {url}"
                )

            return "\n\n".join(results)

        except Exception as tavily_error:
            return f"Unfortunately an error occurred in both search systems: {str(tavily_error)}"
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Resturant Search</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
یکی دیگر از ابزارهای مفید تراول‌بات، جستجوی رستوران است. شما باید ابزاری تعریف کنید که نام مقصد را از کاربر دریافت کرده و با استفاده از Tavily Search API به جستجوی بهترین رستوران‌ها و غذاهای محلی در آن مقصد بپردازد. این ابزار باید لیستی از رستوران‌های پیشنهادی را همراه با نام، نوع غذا، توضیحات مختصر و لینک اطلاعات بیشتر به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **استفاده از Tavily Search:** از آنجایی که اطلاعات رستوران‌ها و غذاهای محلی در وب به‌روز و متنوع هستند، از Tavily برای جستجوی این اطلاعات استفاده می‌کنیم.

- **ساخت Query مناسب:** query باید شامل کلمات کلیدی مناسب باشد. مثال: `"best restaurants local food in {destination}"` یا `"top rated restaurants cuisine in {destination}"`.

- **فقط پارامتر Destination:** برخلاف ابزار جستجوی پرواز و هتل، این ابزار فقط نام مقصد را به عنوان ورودی می‌گیرد.

````python

@tool
def weather_info_tool(destination: str, date: str):
    query = f"weather forecast in {destination} on {date} temperature humidity rain chance and clothing recommendation"
    try:
        print(f"Retrieving weather forecast for {destination} on {date}...")
        response = tavily_client.search(
            query=query,
            search_depth="advanced",
            max_results=5
        )
        results = response.get('results', [])
        if not results:
            return f"Unfortunately, weather information for {destination} on {date} was not found."
        weather_summary = [f"Weather forecast for {destination} ({date}):\n"]
        top_result = results[0]
        weather_summary.append(f"Forecast summary: {top_result.get('content')}")
        other_details = []
        for res in results[1:3]:
            content = res.get('content', '')
            if len(content) > 50:
                other_details.append(f"- {content}")
        if other_details:
            weather_summary.append("\nAdditional details:")
            weather_summary.extend(other_details)
        weather_summary.append(f"\nSource: {top_result.get('url')}")
        return "\n".join(weather_summary)
    except Exception as e:
        return f"Error retrieving weather information: {str(e)}"
````

  - شما باید از تابع mapping که نوشته اید برای map کردن نام هر کشور به کد ارزی مربوطه آن کشور استفاده کنید

````python

def get_currency_code(country_name: str) -> str:
    clean_name = country_name.strip().lower()
    manual_map = {
        "usa": "USD", "united states": "USD", "america": "USD",
        "uk": "GBP", "united kingdom": "GBP", "england": "GBP",
        "uae": "AED", "dubai": "AED",
        "iran": "IRR", "eu": "EUR", "europe": "EUR"
    }
    if clean_name in manual_map:
        return manual_map[clean_name]
    try:
        if 'df' in globals():
            result = df[df['Country'].str.lower() == clean_name]
            if not result.empty:
                return result.iloc[0]['Code']
    except Exception:
        pass
    return None
````

````python


@tool
def currency_info_tool(source_country: str, target_country: str):
    source_code = get_currency_code(source_country)
    target_code = get_currency_code(target_country)
    if not source_code:
        return f"Error: currency code for '{source_country}' not found. Please provide a more precise country name."
    if not target_code:
        return f"Error: currency code for '{target_country}' not found."
    try:
        query = f"{source_code} to {target_code} exchange rate today conversion"
        print(f"Converting {source_country} ({source_code}) to {target_country} ({target_code})...")
        response = tavily_client.search(
            query=query,
            search_depth="basic",
            max_results=3
        )
        results = response.get('results', [])
        if not results:
            return f"Unfortunately, the exchange rate for {source_code} to {target_code} was not found."
        top_result = results[0]
        content = top_result.get('content')
        url = top_result.get('url')
        return (
            f"Exchange rate info:\nFrom: {source_country} ({source_code})\nTo: {target_country} ({target_code})\n\n"
            f"Found info:\n{content}\n\nSource: {url}"
        )
    except Exception as e:
        return f"Error while fetching exchange rate: {str(e)}"
````

````python
@tool
def trip_planning_tool(destination: str, arrival_date: str, departure_date: str, interests: str = ""):
    try:
        start = datetime.strptime(arrival_date, "%Y-%m-%d")
        end = datetime.strptime(departure_date, "%Y-%m-%d")
        duration = (end - start).days + 1
        if duration < 1:
            return "Error: departure date must be after arrival date."
        season = start.strftime("%B")
    except ValueError:
        return "Error: date format must be YYYY-MM-DD."

    print(f"Planning a {duration}-day trip to {destination}...")

    search_queries = [
        f"top tourist attractions and landmarks in {destination}",
        f"famous local food and best restaurants in {destination}",
        f"cultural activities and history of {destination}",
    ]

    if interests:
        search_queries.append(f"best {interests} activities in {destination}")

    rag_results = []

    try:
        db = lancedb.connect("./lancedb_data")
        if "book_table" in db.table_names():
            tbl = db.open_table("book_table")
            for q in search_queries:
                q_vec = embedding_model.encode([q])[0]
                results = tbl.search(q_vec).limit(2).to_pandas()
                for _, row in results.iterrows():
                    if row['_distance'] < 0.6:
                        rag_results.append(f"- {row['text']}")
    except Exception as e:
        print(f"Error accessing LanceDB: {e}")

    sources = []
    if len(rag_results) < 3:
        try:
            for q in search_queries:
                search_resp = tavily_client.search(query=q, max_results=2)
                for r in search_resp.get('results', []):
                    sources.append(r.get('snippet') or r.get('content'))
        except Exception as e:
            print(f"Error using Tavily Search: {e}")

    itinerary = f"Trip itinerary for {destination} ({arrival_date} to {departure_date}):\n\n"
    if rag_results:
        itinerary += "Notes from Tourism Book:\n" + "\n".join(rag_results[:10]) + "\n\n"
    if sources:
        itinerary += "Supplemental Web Sources:\n" + "\n".join(sources[:10])

    return itinerary
````

# ReAct Paper Explanation

The paper "ReAct: SYNERGIZING REASONING AND ACTING IN LANGUAGE MODELS" by Shunyu Yao, Jeffrey Zhao, Dian Yu, Nan Du, Izhak Shafran, Karthik Narasimhan, and Yuan Cao, published at ICLR 2023, introduces a novel paradigm called ReAct (Reasoning + Acting) for enhancing the capabilities of large language models (LLMs) in task-solving scenarios. Below, I provide a comprehensive, point-by-point explanation of the paper's key elements, structured around its main sections and contributions. This covers the motivation, methodology, experiments, results, and implications, drawing directly from the paper's content.

## 1. Abstract and Core Motivation
- **Human-Inspired Synergy**: The paper draws inspiration from human cognition, where verbal reasoning (e.g., inner speech or "thinking aloud") seamlessly integrates with physical actions (e.g., cooking a meal by reasoning about steps while interacting with tools). Humans reason to plan, handle exceptions, and gather information, while actions provide feedback to refine reasoning. LLMs, however, have been studied separately for reasoning (e.g., chain-of-thought prompting) or acting (e.g., generating action plans), leading to limitations like hallucination in reasoning or poor planning in acting.
- **ReAct's Innovation**: ReAct prompts LLMs to generate both reasoning traces (thoughts) and task-specific actions in an interleaved manner. This creates synergy: reasoning helps induce, track, and update action plans, while actions allow interaction with external sources (e.g., knowledge bases) to ground reasoning in reality.
- **Applications and Benefits**: Applied to diverse tasks (question answering, fact verification, text-based games, web navigation), ReAct outperforms baselines, improves interpretability/trustworthiness, and enables human-aligned, controllable behavior. It uses few-shot prompting (1-6 examples) and shows robustness across tasks.
- **Key Contributions**:
  - Introduces ReAct as a general paradigm for synergizing reasoning and acting.
  - Extensive experiments demonstrating advantages over isolated reasoning/acting methods.
  - Ablations showing the value of combining internal knowledge (model's pretraining) with external interactions.
  - Analysis of limitations (e.g., reliance on prompting) and potential for finetuning.

## 2. Introduction
- **Problem Setup**: LLMs excel at language understanding and decision-making but struggle with integrated reasoning and acting. Chain-of-thought (CoT) reasoning is static and black-box, prone to hallucination (fabricating facts) and error propagation. Acting methods (e.g., planning in environments) lack abstract reasoning for goals or working memory.
- **ReAct Overview**: ReAct augments the LLM's action space with a language space for "thoughts" (reasoning traces). Thoughts are interleaved with actions, allowing dynamic reasoning-to-act (e.g., planning based on thoughts) and act-to-reason (e.g., using external feedback to update thoughts). This mirrors human cognition for robust, adaptive task-solving.
- **Evaluation Domains**: Tested on four benchmarks:
  - **HotpotQA**: Multi-hop question answering requiring reasoning over Wikipedia passages.
  - **Fever**: Fact verification (SUPPORTS/REFUTES/NOT ENOUGH INFO) based on evidence.
  - **AlfWorld**: Text-based household tasks (e.g., "put a clean lettuce in diningtable") in a simulated environment.
  - **WebShop**: Real-world web navigation for shopping (e.g., buying products matching user instructions).
- **Results Summary**: ReAct achieves strong performance with few-shot prompting, e.g., 34% success rate improvement on AlfWorld over imitation learning, and 10% on WebShop. It also enhances interpretability, allowing humans to inspect reasoning and correct errors.
- **Broader Impact**: ReAct could unlock LLMs for more applications, combining with reinforcement learning for better agents.

## 3. ReAct: SYNERGIZING REASONING + ACTING
- **Formal Setup**: In a standard agent-environment loop, the agent observes (ot), takes actions (at), and updates context (ct). ReAct augments the action space A to A ∪ L, where L is the language space for thoughts (aˆt ∈ L). Thoughts reason over ct to update it (ct+1 = ct ∪ aˆt) without external effects, enabling planning, exception handling, and knowledge integration.
- **Prompting Approach**: Uses frozen LLMs (e.g., PaLM-540B) prompted with few-shot examples. Each example is a trajectory of interleaved thoughts, actions, and observations. For knowledge tasks (e.g., HotpotQA), thoughts are dense (every few steps); for decision-making (e.g., AlfWorld), thoughts are sparse (at key points) to let the model decide their placement.
- **Design Principles**:
  - **Intuitive and Easy**: Annotators simply write thoughts alongside actions, no complex formats.
  - **General and Flexible**: Works across tasks with varying action spaces and reasoning needs.
  - **Performant and Robust**: Few-shot learning with generalization; robust to prompt variations.
  - **Human-Aligned and Controllable**: Trajectories are interpretable; humans can edit thoughts for corrections (e.g., fixing hallucinations).
- **Comparison to Baselines**:
  - **Standard**: No thoughts/actions (just input-output).
  - **CoT (Reason-Only)**: Thoughts only, no actions.
  - **Act (Action-Only)**: Actions only, no thoughts.
  - ReAct interleaves both, outperforming each.

## 4. Knowledge-Intensive Reasoning Tasks (HotpotQA and Fever)
- **Setup**: Tasks require external knowledge; models access a simple Wikipedia API with actions: search[entity] (returns top sentences), lookup[string] (next sentence containing string), finish[answer].
- **ReAct Prompting**: Uses 6 (HotpotQA) or 3 (Fever) annotated trajectories with dense thought-action-observation interleaving. Thoughts handle decomposition, commonsense, search reformulation, and synthesis.
- **Baselines**:
  - **CoT**: Thoughts only.
  - **CoT-SC**: Self-consistency with 21 CoT samples and majority voting.
  - **Act**: Actions only.
- **Hybrid Methods**: Combines ReAct and CoT-SC for robustness:
  - **ReAct → CoT-SC**: If ReAct fails to answer within steps (7 for HotpotQA, 5 for Fever), fall back to CoT-SC.
  - **CoT-SC → ReAct**: If CoT-SC confidence is low (majority <50%), fall back to ReAct.
- **Finetuning**: Bootstraps 3,000 ReAct-generated trajectories to finetune smaller models (PaLM-8/62B), showing ReAct's superiority over baselines.
- **Results and Analysis**:
  - **Performance**: ReAct > Act; competitive with CoT; hybrids (e.g., ReAct → CoT-SC) best overall (e.g., 35.1% EM on HotpotQA). Finetuning boosts ReAct most.
  - **Success/Failure Modes**: ReAct reduces hallucination (6% false positives vs. 14% in CoT) but has more reasoning errors (e.g., loops). CoT has flexible reasoning but high hallucination (56% failures). Hybrids mitigate trade-offs.
  - **Ablations**: Acting grounds reasoning; reasoning guides better searches. External knowledge (via API) improves factuality over internal-only CoT.

## 5. Decision Making Tasks (AlfWorld and WebShop)
- **AlfWorld**: 134 tasks across 6 types (e.g., pick/clean/heat/cool/look). Actions: navigate, interact (e.g., take, clean). ReAct uses sparse thoughts for goal decomposition, progress tracking, commonsense (e.g., where items are likely found).
- **WebShop**: 500 instructions for product purchase. Actions: search, click, buy. ReAct reasons to explore options, match attributes, decide when to buy.
- **Baselines**:
  - **BUTLER**: Imitation learning on 105 expert trajectories per task.
  - **IL/IL+RL**: Imitation learning ± reinforcement learning on WebShop.
  - **Act**: Actions only.
- **ReAct Prompting**: 3 trajectories per AlfWorld task type; sparse thoughts. For WebShop, reasoning bridges noisy text to actions.
- **Ablation (ReAct-IM)**: Uses dense feedback thoughts (e.g., "current subgoal") instead of flexible reasoning. Shows ReAct > ReAct-IM (71% vs. 53% success on AlfWorld), highlighting internal reasoning's value over simple reactions.
- **Results**:
  - **AlfWorld**: ReAct (best prompt: 71% success) > Act (45%) > BUTLER (37%). Consistent gains across tasks.
  - **WebShop**: ReAct (40% success) > Act (30%) > IL/IL+RL (29-30%). Improves attribute matching.
  - **Qualitative**: ReAct avoids loops/errors in Act; enables subgoal tracking and exploration.

## 6. Related Work
- **Reasoning in LLMs**: CoT and variants (e.g., self-consistency, least-to-most) focus on isolated reasoning. ReAct integrates reasoning with actions for grounded, dynamic processes.
- **Acting in LLMs**: Methods like WebGPT, SayCan, Inner Monologue generate actions but lack abstract reasoning. ReAct's flexible thoughts surpass simple feedback (e.g., vs. Inner Monologue).
- **Broader**: Builds on embodied AI (e.g., ALFRED) and generalist agents, emphasizing language as a cognitive tool.

## 7. Conclusion and Limitations
- **Contributions Recap**: ReAct paradigm, empirical advantages, ablations, and analysis.
- **Limitations**: Prompting limits scalability (long contexts); needs more human data for finetuning. Not tested on non-language actions.
- **Future Work**: Scale with multi-task training, combine with RL, explore human-in-the-loop editing for alignment.

## Overall Implications
ReAct demonstrates that interleaving reasoning and acting makes LLMs more reliable, interpretable, and capable of complex tasks. It reduces hallucinations via external grounding and enables adaptive planning. While prompting-based, it hints at finetuning potential. This work advances LLM applications in AI agents, emphasizing human-like cognition for safer, more controllable systems.

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Defining System Message and Bind Tools</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

یک System Message جامع بنویسید که:

- نقش agent را مشخص کند (Travel Assistant)
- لیست 8 tool موجود را معرفی کند
- دستورالعمل استفاده از هر tool را توضیح دهد
- سپس tools را به LLM bind کنید
</div>
<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
توجه کنید که مدیریت فرمت‌های تاریخ با توجه به ابزار مورد نظر را باید در  System Prompt در نظر بگیرید.
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
پیام تایید bind موفق 7 tool
</p>


````python

llm = ChatGoogleGenerativeAI(
    model="gemini-2.0-flash-lite",
    temperature=0
)
````

````python
current_date = datetime.now().strftime("%Y-%m-%d")
````

````python

system_prompt_text = f"""You are an intelligent and resourceful **Travel Assistant**. Your primary goal is to help users plan their trips, book flights/hotels, check weather, and answer travel-related questions accurately.

**Current Date:** {current_date}

### AVAILABLE TOOLS:
1.  **flight_search_tool**: Use for finding flight options. Requires origin, destination, and date (YYYY-MM-DD).
2.  **hotel_search_tool**: Use for finding accommodation. Requires city, check-in, and check-out dates (YYYY-MM-DD).
3.  **restaurant_search_tool**: Search for food and dining recommendations in a specific city.
4.  **weather_info_tool**: Get weather forecasts. Requires destination and date.
5.  **currency_info_tool**: Convert currency exchange rates (e.g., USD to EUR).
6.  **faq_search_tool**: Answer general travel questions (visa, baggage, etc.) using the internal knowledge base.
7.  **trip_planning_tool**: Generate full daily itineraries based on user interests and dates.

### INSTRUCTIONS:
-   **Date Handling**: The user might use relative dates like "next Friday", "tomorrow", or "in 3 days". You MUST calculate the exact date based on the **Current Date** ({current_date}) and convert it to **YYYY-MM-DD** format before calling any tool.
-   **Tool Usage**: Always choose the most specific tool for the task. If you need more information from the user (like a missing city or date) to use a tool, ask them politely before guessing.
-   **Response**: Provide friendly, helpful, and concise answers. You can answer in Persian or English based on the user's language, but tool arguments must be in English.
"""
````

````python

tools = [
    flight_search_tool,
    hotel_search_tool,
    restaurant_search_tool,
    weather_info_tool,
    currency_info_tool,
    faq_search_tool,
    trip_planning_tool
]
````

````python


print(f"Tools list created with {len(tools)} tools.")
llm_with_tools = llm.bind_tools(tools)
print(f"Successfully bound {len(tools)} tools to the LLM model.")
print(f"System date set to {current_date}.")
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Creating Graph Nodes</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
Reasoner node را پیاده‌سازی کنید که:
- System message و تاریخچه مکالمه را ترکیب کند
- LLM را invoke کند
- پاسخ را برگرداند
</p>
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
تعریف موفق reasoner node
</p>


````python
class AgentState(TypedDict):
    messages: Annotated[list, add_messages]

def reasoner_node(state: AgentState):
    messages = state["messages"]
    full_history = [SystemMessage(content=system_prompt_text)] + messages
    response = llm_with_tools.invoke(full_history)
    return {"messages": [response]}

````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Creating and Compiling Graph</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
StateGraph را بسازید و compile کنید و تصویر گراف را رسم کنید.

</p>
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
• Graph compiled موفق<br>
• تصویر visualization گراف<br>
</p>


````python
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode, tools_condition
from IPython.display import Image, display

workflow = StateGraph(AgentState)
workflow.add_node("reasoner", reasoner_node)
workflow.add_node("tools", ToolNode(tools))
workflow.set_entry_point("reasoner")
workflow.add_conditional_edges("reasoner", tools_condition)
workflow.add_edge("tools", "reasoner")
app = workflow.compile()

try:
    display(Image(app.get_graph().draw_mermaid_png()))
except Exception as e:
    print("Could not render graph (pygraphviz may need to be installed).")
````

````python
import uuid

def chat_cli():
    print("Travel Bot CLI")
    print("Type 'exit' to quit, 'clear' to reset chat.")
    config = {"configurable": {"thread_id": str(uuid.uuid4())}}
    current_chat_history = []
    while True:
        try:
            user_input = input("\nYou: ").strip()
            if user_input.lower() in ["exit", "quit"]:
                print("Goodbye!")
                break
            if user_input.lower() == "clear":
                current_chat_history = []
                print("Memory cleared.")
                continue
            if not user_input:
                continue
            print("Bot is thinking...", end="\r")
            inputs = {"messages": [HumanMessage(content=user_input)]}
            final_state = app.invoke(inputs, config=config)
            all_messages = final_state["messages"]
            last_message = all_messages[-1]
            tool_calls = [m.name for m in all_messages if hasattr(m, 'tool_call_id') and m.type == 'tool']
            if tool_calls:
                print(f"Tools used: {', '.join(tool_calls)}")
            print(f"Bot: {last_message.content}")
        except Exception as e:
            print(f"Error: {e}")
````

- فراخوانی پیشنهادی: اگر return_date مشخص نیست: ask_user("What return date would you prefer?") — بعد از دریافت، فراخوانی flight_search_tool برای رفت و برگشت
- حالات شکست: کاربر تاریخ بازگشت را نداده؛ مدل باید سؤال تکمیلی بپرسد
- معیارهای ارزیابی: آیا agent قبل از اجرای call از کاربر اطلاعات لازم را گرفت؟ آیا پیشنهادهای رفت/برگشت ساختارمند بودند؟

<h4>سناریو 3 — Flight Search (Mashhad → Istanbul, tomorrow)</h4>
- Query: "Check for available flights from Mashhad to Istanbul for tomorrow." 
- پردازش تاریخ: date = today + 1 day → validate (not in past)
- نمونه خروجی: اگر پرواز موجود بود، 1–3 پیشنهاد با زمان و توقف‌ها؛ در غیر این صورت پیام «No flights found» + پیشنهاد تاریخ‌های نزدیک
- پیشنهادات: اگر تخفیف یا محدودیت وجود دارد، به کاربر اطلاع دهید (مثلاً sold out). اگر direct match نبود، پیشنهاد پرواز در روز قبل/بعد را بدهید.

<h4 dir="rtl" style="font-family: Vazir; width: 85%;">صداقت علمی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;"> <li>ما نوت‌بوک‌های تعداد مشخصی از دانشجویان که به صورت تصادفی انتخاب می‌شوند، بررسی خواهیم کرد. این بررسی‌ها اطمینان حاصل می‌کنند که کدی که نوشتید واقعاً پاسخ‌های موجود در نوت‌بوک شما را تولید می‌کند. اگر پاسخ‌های صحیح را در نوت‌بوک خود بدون کدی که واقعاً آن پاسخ‌ها را تولید کند تحویل دهید، این یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> <li>ما همچنین بررسی‌های خودکاری را برای تشخیص سرقت علمی در نوت‌بوک‌های کولب انجام خواهیم داد. کپی کردن کد از دیگران نیز یک مورد جدی از عدم صداقت علمی محسوب می‌شود.</li> </ul>
<h4 dir="rtl" style="font-family: Vazir; width: 85%;">توضیحات تکمیلی:</h4> <ul dir="rtl" style="font-family: Vazir; width: 85%; font-size: 16px;">
<li>
