در این بخش لازم است با استفاده از ابزار مناسب مجموعه داده (zinovadr/iata-airport-code) را از بستر Kaggle دانلود نموده و یک Dictionary mapping هوشمند از نام شهر به کد فرودگاه مورد نظر بسازید.

````python
import kagglehub
import pandas as pd
import os

print("Downloading Currency dataset...")
# 1. Download the dataset
currency_path = kagglehub.dataset_download("phanee16/currency-and-country-code-mapping")
print(f"Currency Data downloaded to: {currency_path}")

# 2. Find the correct CSV file
csv_files = [f for f in os.listdir(currency_path) if f.endswith('.csv')]

if csv_files:
    currency_csv_path = os.path.join(currency_path, csv_files[0])
    currency_df = pd.read_csv(currency_csv_path)

    # --- DEBUGGING STEP ---
    # Normalize columns to lower case and strip spaces to handle variations
    currency_df.columns = [c.lower().strip() for c in currency_df.columns]
    print(f"Detected columns: {list(currency_df.columns)}")

    # 3. Smart Column Detection
    # We look for the best matching column names
    country_col = next((c for c in currency_df.columns if c in ['country', 'country_name', 'name', 'entity']), None)
    code_col = next((c for c in currency_df.columns if c in ['currency_code', 'currency code', 'code', 'iso_code', 'currency']), None)

    if country_col and code_col:
        print(f"Using columns: Country='{country_col}', Code='{code_col}'")

        country_to_currency = {}
        for index, row in currency_df.iterrows():
            country = str(row[country_col]).lower().strip()
            code = str(row[code_col]).strip()

            # Filter out bad data (NaN or empty)
            if country and code and code.lower() != 'nan':
                country_to_currency[country] = code

        print(f"Number of currency mappings loaded: {len(country_to_currency)}")
        print(f"Sample: France -> {country_to_currency.get('france')}, Iran -> {country_to_currency.get('iran')}")
    else:
        print("❌ Could not automatically find 'Country' or 'Currency Code' columns.")
        print("Please check the 'Detected columns' printed above.")

else:
    print("❌ No CSV file found.")
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Download Kaggle Currency Dataset</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

در این بخش لازم است با استفاده از ابزار مناسب مجموعه داده ("phanee16/currency-and-country-code-mapping") را از بستر Kaggle دانلود نموده و یک Dictionary mapping هوشمند از نام کشور به کد ارز مورد نظر بسازید.

</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
• تعداد ارزها<br>
• تعداد ورودی‌های mapping dictionary<br>
• نمونه جستجو: IRAN → IRR, UAE → AED
</p>


````python
import kagglehub
import pandas as pd
import os

print("Downloading Currency dataset...")

# 1. Download the dataset (Handles downloading & unzipping automatically)
# Note: kagglehub stores files in a cache folder, not locally in "./data" by default.
path = kagglehub.dataset_download("phanee16/currency-and-country-code-mapping")
print(f"Dataset downloaded to: {path}")

# 2. Find the correct CSV file path automatically
# We search the download folder for the csv file to avoid path errors
csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]

if csv_files:
    # Use the first CSV found (usually 'country_code_to_currency_code.csv')
    # It seems there are two CSVs, let's try the one that contains country_code_to_currency_code.csv as it seems more relevant
    target_csv = next((f for f in csv_files if 'country_code_to_currency_code' in f), csv_files[0])
    csv_file_path = os.path.join(path, target_csv)
    currency_df = pd.read_csv(csv_file_path)

    # 3. Create the dictionary
    country_to_currency = {}

    # Normalize column names to lowercase and strip spaces
    currency_df.columns = [c.lower().strip() for c in currency_df.columns]

    print(f"Detected columns in {target_csv}: {list(currency_df.columns)}")

    # Dynamically find the best matching columns for country and currency code
    country_col = next((c for c in currency_df.columns if 'country' in c and 'code' not in c), None)
    currency_code_col = next((c for c in currency_df.columns if 'currency_code' in c or 'currency' == c), None)

    if country_col and currency_code_col:
        print(f"Using columns: Country='{country_col}', CurrencyCode='{currency_code_col}'")
        for index, row in currency_df.iterrows():
            country_name = str(row[country_col]).lower().strip()
            currency_code = str(row[currency_code_col]).strip()

            # Filter out bad data (NaN or empty)
            if country_name and currency_code and currency_code.lower() != 'nan':
                country_to_currency[country_name] = currency_code

        # 4. Print results
        print(f"Number of currencies: {len(currency_df[currency_code_col].dropna().unique())}")
        print(f"Number of entries in mapping dictionary: {len(country_to_currency)}")
        print(
            f"Sample search: IRAN -> {country_to_currency.get('iran')}, UAE -> {country_to_currency.get('united arab emirates')}" # Note: 'uae' might be stored as 'united arab emirates' in the data
        )
    else:
        print("❌ Could not automatically find 'Country' or 'Currency Code' columns.")
        print("Please check the 'Detected columns' printed above.")

else:
    print("❌ Error: No CSV file found in the downloaded folder.")
````

<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
یکی از ابزارهای مهمی که تراول‌بات با استفاده از آن به کاربران خدمات می‌دهد، ابزار جستجوی پرواز است. شما باید ابزاری تعریف کنید که پارامترهایی مانند مبدا، مقصد و تاریخ پرواز را از کاربر دریافت کرده و با استفاده از Amadeus API به جستجوی پروازهای مورد نظر بپردازد. این ابزار باید حداقل ۳ پرواز را با اطلاعات کاملی شامل شماره پرواز، کد ایرلاین، قیمت دقیق (به همراه واحد پول)، تاریخ و ساعت حرکت، مدت پرواز و تعداد توقف‌ها به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **تبدیل نام شهر به کد IATA:** ابزار Amadeus بر اساس کدهای یکتای فرودگاه (IATA Code) کار می‌کند. از تابع مپینگی که در بخش قبل نوشته‌اید استفاده کنید تا نام شهرها را به کد فرودگاه تبدیل کنید (مثلاً Tehran → IKA، Dubai → DXB).

- **مدیریت فرمت تاریخ:** Amadeus تاریخ را به فرمت YYYY-MM-DD می‌پذیرد. از آنجایی که ماهیت چت‌بات صحبت به زبان طبیعی است، کاربران فرمت تاریخ را به گونه‌های متفاوتی وارد می‌کنند(برای مثال به صورت متن و نه عدد). شما باید مدیریت فرمت تاریخ در ابزار و System Prompt را مد نظر داشته باشید.

- **مدیریت خطا:** در صورت بروز خطا در جستجو (مثل عدم یافتن پرواز یا خطای API)، پیام خطای مناسب و قابل فهم به کاربر بازگردانید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید تا LangGraph بتواند آن را شناسایی کند.

</div>


````python
from langchain_core.tools import tool
from amadeus import ResponseError, Client
import datetime
import os

# اطمینان از تعریف کلاینت
# amadeus = Client(client_id=os.environ.get("AMADEUS_CLIENT_ID"), client_secret=os.environ.get("AMADEUS_CLIENT_SECRET"))

def get_iata_code(city_name: str) -> str:
    # Clean the input
    city_name_clean = city_name.strip().lower()
    # Manual fallback for common cities
    manual_map = {
        'tehran': 'IKA', 'mashhad': 'MHD', 'dubai': 'DXB', 'london': 'LON', 'paris': 'PAR', 'istanbul': 'IST', 'rome': 'ROM'
    }
    if city_name_clean in manual_map:
        return manual_map[city_name_clean]

    # Try Amadeus if available
    if amadeus:
        try:
            response = amadeus.reference_data.locations.get(keyword=city_name, subType='CITY')
            if response.data:
                return response.data[0]['iataCode']
        except Exception as e:
            # Log the error but don't crash
            print(f"Amadeus lookup failed for {city_name}: {e}")
    return None

@tool
def flight_search_tool(origin_city: str, destination_city: str, date: str):
    """
    Searches for flights between two cities on a specific date using Amadeus API.

    Args:
        origin_city (str): Name of the origin city (e.g., "Tehran").
        destination_city (str): Name of the destination city (e.g., "Paris").
        date (str): Date of the flight in 'YYYY-MM-DD' format.
    """

    # 1. تبدیل نام شهرها به کد IATA
    origin_code = get_iata_code(origin_city)
    dest_code = get_iata_code(destination_city)

    if not origin_code:
        return f"❌ Error: IATA code for origin city '{origin_city}' not found."
    if not dest_code:
        return f"❌ Error: IATA code for destination city '{destination_city}' not found."

    print(f"✈️ Searching flights: {origin_city} ({origin_code}) -> {destination_city} ({dest_code}) on {date}")

    try:
        # 2. جستجوی پرواز
        if not amadeus:
            # Fallback mock response when Amadeus client is not available
            print("⚠️ Amadeus client not available — returning mock flight results.")
            mock_results = [f"✈️ **Sample flight options (mock) {origin_city} -> {destination_city} ({date}):**\n"]
            mock_results.extend([
                "🔹 **Option 1:**\n   - Airline: AB 123\n   - Price: 250 USD\n   - Time: 09:00 | Duration: 3h | Nonstop",
                "🔹 **Option 2:**\n   - Airline: CD 456\n   - Price: 320 USD\n   - Time: 13:30 | Duration: 4h | 1 stop",
                "🔹 **Option 3:**\n   - Airline: EF 789\n   - Price: 400 USD\n   - Time: 21:10 | Duration: 6h | 2 stops"
            ])
            return "\n\n".join(mock_results)

        response = amadeus.shopping.flight_offers_search.get(
            originLocationCode=origin_code,
            destinationLocationCode=dest_code,
            departureDate=date,
            adults=1,
            max=3
        )

        if not response.data:
            return f"No flights found from {origin_city} to {destination_city} on {date}."

        results = []
        results.append(f"✈️ **Flight results {origin_city} to {destination_city} ({date}):**\n")

        for idx, offer in enumerate(response.data):
            price = offer['price']['total']
            currency = offer['price']['currency']
            itinerary = offer['itineraries'][0]
            duration = itinerary['duration'].replace("PT", "").lower()
            segments = itinerary['segments']
            stops = len(segments) - 1
            stop_text = "Nonstop" if stops == 0 else f"{stops} stop(s)"

            first_segment = segments[0]
            airline_code = first_segment['carrierCode']
            flight_number = f"{airline_code} {first_segment['number']}"
            departure_time = first_segment['departure']['at'].split("T")[1][:5]

            results.append(
                f"🔹 **Option {idx + 1}:**\n"
                f"   - Airline: {airline_code} ({flight_number})\n"
                f"   - Price: {price} {currency}\n"
                f"   - Time: {departure_time} | Duration: {duration} | {stop_text}"
            )

        return "\n\n".join(results)

    except ResponseError as error:
        return f"❌ Amadeus API error: {error}"
    except Exception as e:
        return f"❌ Unknown error: {str(e)}"

# Re-run test
print("✅ Tool updated with fallback. Please test again.")
# ==========================================
# Test tool (sample successful search)
# ==========================================
print("⏳ Testing flight search tool...")

# فرض می‌کنیم تاریخ پرواز فردا است (برای اینکه تاریخ گذشته نباشد)
tomorrow = (datetime.date.today() + datetime.timedelta(days=1)).strftime("%Y-%m-%d")

# فراخوانی ابزار با نام شهرها (کد به صورت خودکار تبدیل می‌شود)
test_result = flight_search_tool.invoke({
    "origin_city": "London",
    "destination_city": "Paris",
    "date": tomorrow
})

print(test_result)
````

- **استفاده از Amadeus Hotel Search API:** از endpoint مناسب Amadeus برای جستجوی هتل استفاده کنید. این API نیاز به کد شهر (IATA City Code) دارد که باید از mapping قبلی استفاده کنید.

- **تبدیل نام شهر به کد IATA:** مشابه جستجوی پرواز، از تابع از تابع مپینگ هوشمند برای تبدیل نام شهر به کد فرودگاه استفاده کنید (مثلاً Dubai → DXB).

- **فرمت تاریخ:** Amadeus تاریخ را به فرمت YYYY-MM-DD می‌پذیرد. از آنجایی که ماهیت چت‌بات صحبت به زبان طبیعی است، کاربران فرمت تاریخ را به گونه‌های متفاوتی وارد می‌کنند(برای مثال به صورت متن و نه عدد). شما باید مدیریت فرمت تاریخ در ابزار و System Prompt را مد نظر داشته باشید.

- **پارامتر Budget (اختیاری):** این ابزار می‌تواند یک پارامتر اختیاری budget داشته باشد که هتل‌ها را بر اساس محدوده قیمتی فیلتر کند.

- **خروجی کامل:** برای هر هتل، اطلاعاتی مانند نام، آدرس، رتبه و قیمت هر شب (به همراه واحد پول) را نمایش دهید.

- **مدیریت خطا:** در صورت عدم یافتن هتل یا خطای API، پیام مناسب به کاربر بازگردانید.

- **Fallback به Tavily:** در صورتی که Amadeus Hotel API در دسترس نبود یا محدودیت داشت، می‌توانید به عنوان fallback از Tavily Search استفاده کنید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید.

````python
from langchain_core.tools import tool
from amadeus import ResponseError
from tavily import TavilyClient
import os

# اطمینان از تنظیم بودن کلاینت‌ها
# amadeus = Client(...) # از قبل موجود است
tavily_client = TavilyClient(api_key=os.environ.get("TAVILY_API_KEY"))

# استفاده مجدد از تابع کمکی (اگر در سلول قبلی اجرا شده نیاز به تعریف مجدد نیست، اما برای کامل بودن اینجا می‌آوریم)
def get_iata_code(city_name: str) -> str:
    try:
        response = amadeus.reference_data.locations.get(keyword=city_name, subType='CITY')
        if response.data:
            return response.data[0]['iataCode']
        return None
    except:
        return None

@tool
def hotel_search_tool(city: str, check_in: str, check_out: str, budget: int = None):
    """
    Searches for hotels in a specific city for a given date range.
    Uses Amadeus API first, and falls back to Tavily Search if Amadeus fails or finds no results.

    Args:
        city (str): Name of the destination city (e.g., "Paris").
        check_in (str): Check-in date in 'YYYY-MM-DD' format.
        check_out (str): Check-out date in 'YYYY-MM-DD' format.
        budget (int, optional): Maximum price per night in currency units (e.g., 200). Defaults to None.

    Returns:
        str: A text summary of available hotels with details (Name, Rating, Price, Address).
    """

    # 1. تلاش با Amadeus API
    try:
        city_code = get_iata_code(city)
        if not city_code:
            raise ValueError(f"Code not found for {city}")

        print(f"🏨 Searching Amadeus for {city} ({city_code})...")

        # search parameters
        search_params = {
            'cityCode': city_code,
            'checkInDate': check_in,
            'checkOutDate': check_out,
            'adults': 1,
            'radius': 5, # 5 km radius from city center
            'radiusUnit': 'KM',
            'paymentPolicy': 'NONE', # do not filter by payment policy
            'includeClosed': 'false',
            'bestRateOnly': 'true',
            'view': 'LIGHT',
            'max': 5 # max 5 hotels
        }

        if budget:
            search_params['priceRange'] = f"1-{budget}"

        response = amadeus.shopping.hotel_offers_search.get(**search_params)

        if not response.data:
            raise ValueError("No hotels found in Amadeus.")

        results = [f"🏨 **Recommended hotels in {city} (source: Amadeus):**\n"]

        for offer in response.data:
            hotel = offer.get('hotel', {})
            offer_details = offer.get('offers', [])[0] if offer.get('offers') else {}

            name = hotel.get('name', 'Unknown')
            rating = hotel.get('rating', 'N/A')

            price_obj = offer_details.get('price', {})
            price = price_obj.get('total', 'N/A')
            currency = price_obj.get('currency', '')

            address = "City center" # default value

            results.append(
                f"🔹 **{name}**\n"
                f"   - Rating: {rating} stars\n"
                f"   - Total price: {price} {currency}\n"
                f"   - Location: {city_code}"
            )

        return "\n\n".join(results)

    except (ResponseError, ValueError, Exception) as e:
        print(f"⚠️ Amadeus error or no results ({str(e)}). Switching to Tavily Search...")

        try:
            budget_str = f"under {budget} currency" if budget else ""
            query = f"Best hotels in {city} from {check_in} to {check_out} {budget_str} with price and rating"

            tavily_response = tavily_client.search(
                query=query,
                search_depth="advanced",
                max_results=5
            )

            results = [f"🌐 **Recommended hotels in {city} (source: Web Search):**\n(These results are gathered from the web due to booking limitations)\n"]

            for result in tavily_response.get('results', []):
                title = result.get('title')
                content = result.get('content')[:200]
                url = result.get('url')

                results.append(
                    f"🔹 **{title}**\n"
                    f"   - Description: {content}...\n"
                    f"   - Link: [View details]({url})"
                )

            return "\n\n".join(results)

        except Exception as tavily_error:
            return f"❌ Unfortunately an error occurred in both search systems: {str(tavily_error)}"

# ==========================================
# تست ابزار (Hotel Search)
# ==========================================
print("\n⏳ Testing hotel search tool...")

# تست ۱: جستجو با تاریخ مشخص (احتمالاً Amadeus جواب می‌دهد یا ارور می‌دهد و به Tavily می‌رود)
# توجه: هتل‌های آمادئوس در محیط Test محدود هستند، پس احتمال رفتن به Fallback زیاد است.
import datetime
next_week = (datetime.date.today() + datetime.timedelta(days=7)).strftime("%Y-%m-%d")
next_week_checkout = (datetime.date.today() + datetime.timedelta(days=9)).strftime("%Y-%m-%d")

test_result = hotel_search_tool.invoke({
    "city": "Istanbul",
    "check_in": next_week,
    "check_out": next_week_checkout,
    "budget": 500 # بودجه اختیاری
})

print(test_result)
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Resturant Search</div>

یکی دیگر از ابزارهای مفید تراول‌بات، جستجوی رستوران است. شما باید ابزاری تعریف کنید که نام مقصد را از کاربر دریافت کرده و با استفاده از Tavily Search API به جستجوی بهترین رستوران‌ها و غذاهای محلی در آن مقصد بپردازد. این ابزار باید لیستی از رستوران‌های پیشنهادی را همراه با نام، نوع غذا، توضیحات مختصر و لینک اطلاعات بیشتر به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **استفاده از Tavily Search:** از آنجایی که اطلاعات رستوران‌ها و غذاهای محلی در وب به‌روز و متنوع هستند، از Tavily برای جستجوی این اطلاعات استفاده می‌کنیم.

- **ساخت Query مناسب:** query باید شامل کلمات کلیدی مناسب باشد. مثال: `"best restaurants local food in {destination}"` یا `"top rated restaurants cuisine in {destination}"`.

- **فقط پارامتر Destination:** برخلاف ابزار جستجوی پرواز و هتل، این ابزار فقط نام مقصد را به عنوان ورودی می‌گیرد.

- **خروجی متنوع:** نتایج باید شامل رستوران‌های مختلف با انواع غذاها باشد (غذای محلی، رستوران‌های معروف، کافه‌ها و ...).

- **فرمت خروجی:** هر رستوران باید شامل: نام، نوع غذا/آشپزی، توضیح کوتاه (100-150 کلمه)، و لینک URL برای اطلاعات بیشتر باشد.

- **محدود کردن نتایج:** تعداد نتایج را محدود کنید (مثلاً 5 رستوران برتر) تا خروجی خلاصه و کاربردی باشد.

- **مدیریت خطا:** در صورت عدم یافتن نتیجه یا خطای API، پیام واضح و مفیدی به کاربر بازگردانید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید.

````python
from langchain_core.tools import tool
from tavily import TavilyClient
import os

# اطمینان از تنظیم بودن کلاینت
# فرض بر این است که TAVILY_API_KEY در os.environ موجود است
tavily_client = TavilyClient(api_key=os.environ.get("TAVILY_API_KEY"))

@tool
def restaurant_search_tool(destination: str):
    """
    Searches for top-rated restaurants, local food, and dining experiences in a specific destination.
    Returns a list of 5 recommended places with descriptions and links.

    Args:
        destination (str): The name of the city or location (e.g., "Paris", "Tehran").

    Returns:
        str: A formatted text containing restaurant names, descriptions, and URLs.
    """

    # 1. ساخت کوئری هوشمند
    # ترکیب "local food" و "top rated" باعث می‌شود هم غذاهای سنتی و هم رستوران‌های باکیفیت پیدا شوند.
    query = f"best top rated restaurants and local food in {destination} with cuisine type and reviews"

    try:
        print(f"🍽️ Searching for restaurants in {destination}...")

        # 2. انجام جستجو با Tavily
        response = tavily_client.search(
            query=query,
            search_depth="advanced", # جستجوی عمیق برای نتایج باکیفیت‌تر
            max_results=5,           # محدود کردن به ۵ نتیجه
            include_domains=[],      # می‌توان دامنه‌های خاصی مثل TripAdvisor را فیلتر یا محدود کرد (اختیاری)
        )

        results_list = response.get('results', [])

        if not results_list:
            return f"Unfortunately, no restaurants or food information was found for '{destination}'."

        # 3. فرمت‌دهی خروجی
        formatted_output = [f"🍽️ **پیشنهادات رستوران و غذای محلی در {destination}:**\n"]

        for item in results_list:
            name = item.get('title', 'نامشخص')
            url = item.get('url', '#')
            # تاویلی معمولاً توضیحات را در فیلد content برمی‌گرداند که شامل نوع غذا و نقد کوتاه است
            description = item.get('content', 'توضیحات در دسترس نیست.')

            # تمیز کردن متن (حذف کاراکترهای اضافی اگر خیلی طولانی باشد)
            if len(description) > 300:
                description = description[:300] + "..."

            formatted_output.append(
                f"🥘 **{name}**\n"
                f"   - 📝 توضیحات/نوع غذا: {description}\n"
                f"   - 🔗 لینک اطلاعات بیشتر: [کلیک کنید]({url})"
            )

        return "\n\n".join(formatted_output)

    except Exception as e:
        # 4. مدیریت خطا
        return f"❌ خطا در جستجوی رستوران: {str(e)}"

# ==========================================
# تست ابزار (Restaurant Search)
# ==========================================
print("\n⏳ Testing restaurant search tool...")

# تست با شهر رم (به دلیل معروف بودن غذاها)
test_result = restaurant_search_tool.invoke({
    "destination": "Rome"
})

print(test_result)
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Get Weather Info</div>

ابزار دریافت اطلاعات آب و هوا یکی از ابزارهای حیاتی برای برنامه‌ریزی سفر است. شما باید ابزاری تعریف کنید که نام مقصد و تاریخ مورد نظر را از کاربر دریافت کرده و با استفاده از Tavily Search API اطلاعات پیش‌بینی آب و هوا را برای آن مقصد و تاریخ جستجو کند. این ابزار باید اطلاعاتی مانند دما، وضعیت هوا (آفتابی، بارانی، ابری)، رطوبت، احتمال بارندگی و توصیه‌های لباس‌پوشی را به کاربر بازگرداند.

**نکات مهم پیاده‌سازی:**

- **استفاده از Tavily Search:** برای دریافت پیش‌بینی آب و هوای به‌روز از جستجوی وب استفاده می‌کنیم، چون منابع هواشناسی آنلاین اطلاعات دقیق‌تری دارند.

- **ساخت Query مناسب:** query باید شامل نام مقصد و تاریخ باشد. مثال: `"weather forecast {destination} {date}"` یا `"weather in {destination} on {date}"`.

- **پارامترهای ورودی:**

  - `destination`: نام شهر یا مقصد (الزامی)
  - `date`: تاریخ مورد نظر به فرمت های مختلف

- **خروجی کامل:** اطلاعات آب و هوا باید شامل

- **پیش‌بینی چند روزه:** اگر ممکن است، اطلاعات پیش‌بینی چند روز آینده را نیز ارائه دهید.

- **مدیریت خطا:** در صورت عدم دسترسی به اطلاعات آب و هوا یا خطای API، پیام مناسب به کاربر بازگردانید.

- **استفاده از Decorator:** از `@tool` decorator برای تعریف این ابزار استفاده کنید.

**مثال Query:**

```
"weather forecast Dubai December 16 2025"
```

</div>


````python
from langchain_core.tools import tool
from tavily import TavilyClient
import os

# اطمینان از تنظیم بودن کلاینت
# tavily_client = TavilyClient(api_key=os.environ.get("TAVILY_API_KEY"))

@tool
def weather_info_tool(destination: str, date: str):
    """
    Retrieves weather forecast information for a specific destination and date.
    Returns details including temperature, weather condition (sunny, rainy, etc.), humidity, precipitation chance, and clothing recommendations.

    Args:
        destination (str): The name of the city or location (e.g., "Paris").
        date (str): The specific date or relative date (e.g., "2024-12-20", "next Friday", "today").

    Returns:
        str: A summary of the weather forecast and clothing advice.
    """

    # 1. ساخت کوئری هوشمند
    # ما در کوئری صراحتاً کلماتی مثل "clothing recommendation" و "humidity" را می‌آوریم
    # تا نتایج جستجو شامل این موارد باشند.
    query = f"weather forecast in {destination} on {date} temperature humidity rain chance and clothing recommendation"

    try:
        print(f"🌦️ Retrieving weather forecast for {destination} on {date}...")

        # 2. جستجو با Tavily
        response = tavily_client.search(
            query=query,
            search_depth="advanced", # برای اطمینان از یافتن سایت‌های معتبر هواشناسی
            max_results=5
        )

        results = response.get('results', [])

        if not results:
            return f"Unfortunately, weather information for {destination} on {date} was not found."

        # 3. پردازش و تجمیع نتایج
        # از آنجایی که خروجی متن است، ما چکیده‌های مرتبط را جمع‌آوری می‌کنیم
        # تا LLM در مرحله بعد بتواند پاسخ نهایی را به کاربر بدهد.
        weather_summary = [f"🌦️ **Weather forecast for {destination} ({date}):**\n"]

        top_result = results[0]
        weather_summary.append(f"🔹 **Forecast summary:** {top_result.get('content')}")

        other_details = []
        for res in results[1:3]:
            content = res.get('content', '')
            if len(content) > 50:
                other_details.append(f"- {content}")

        if other_details:
            weather_summary.append("\n📝 **Additional details from other sources:**")
            weather_summary.extend(other_details)

        weather_summary.append(f"\n🔗 Source: {top_result.get('url')}")

        return "\n".join(weather_summary)

    except Exception as e:
        return f"❌ Error retrieving weather information: {str(e)}"

# ==========================================
# تست ابزار (Weather Search)
# ==========================================
print("\n⏳ Testing weather tool...")

# تست برای دبی در هفته آینده
import datetime
next_monday = (datetime.date.today() + datetime.timedelta(days=5)).strftime("%Y-%m-%d")

test_result = weather_info_tool.invoke({
    "destination": "Dubai",
    "date": next_monday
})

print(test_result)
````

````python
from langchain_core.tools import tool
from tavily import TavilyClient
import os
import pandas as pd

# Ensure the Tavily client is configured (example placeholder)
# tavily_client = TavilyClient(api_key=os.environ.get("TAVILY_API_KEY"))

# ---------------------------------------------------------
# 1. Helper: convert country name to currency code
# ---------------------------------------------------------
def get_currency_code(country_name: str) -> str:
    """
    Given a country name, return the currency code (ISO 4217).
    Uses a preloaded DataFrame (df) or a fallback manual map.
    """
    # normalize input
    clean_name = country_name.strip().lower()

    # manual mapping for common cases
    manual_map = {
        "usa": "USD", "united states": "USD", "america": "USD",
        "uk": "GBP", "united kingdom": "GBP", "england": "GBP",
        "uae": "AED", "dubai": "AED", # Dubai commonly used
        "iran": "IRR", "eu": "EUR", "europe": "EUR"
    }

    if clean_name in manual_map:
        return manual_map[clean_name]

    # fallback: search in DataFrame if available (columns usually 'Country' and 'Code')
    try:
        if 'df' in globals():
            result = df[df['Country'].str.lower() == clean_name]
            if not result.empty:
                return result.iloc[0]['Code']
    except Exception:
        # ignore if df not available or if lookup fails
        pass

    return None

# ---------------------------------------------------------
# 2. Currency info tool
# ---------------------------------------------------------
@tool
def currency_info_tool(source_country: str, target_country: str):
    """
    Get current exchange rate between two countries.

    Args:
        source_country (str): Name of the source country (e.g., "USA").
        target_country (str): Name of the target country (e.g., "UAE").

    Returns:
        str: Exchange rate and source details or an error message.
    """

    # 1. Convert country names to currency codes
    source_code = get_currency_code(source_country)
    target_code = get_currency_code(target_country)

    if not source_code:
        return f"❌ Error: currency code for '{source_country}' not found. Please provide a more precise country name."
    if not target_code:
        return f"❌ Error: currency code for '{target_country}' not found."

    try:
        query = f"{source_code} to {target_code} exchange rate today conversion"

        print(f"💱 Converting {source_country} ({source_code}) to {target_country} ({target_code})...")

        # 3. Search using Tavily
        response = tavily_client.search(
            query=query,
            search_depth="basic",
            max_results=3
        )

        results = response.get('results', [])

        if not results:
            return f"Unfortunately, the exchange rate for {source_code} to {target_code} was not found."

        top_result = results[0]
        content = top_result.get('content')
        url = top_result.get('url')

        return (
            f"💰 **Exchange rate info:**\n"
            f"From: {source_country} ({source_code})\n"
            f"To: {target_country} ({target_code})\n\n"
            f"📊 **Found info:**\n{content}\n\n"
            f"🔗 Source: {url}"
        )

    except Exception as e:
        return f"❌ Error while fetching exchange rate: {str(e)}"


# ==========================================
# Test Currency Info tool
# ==========================================
print("\n⏳ Testing currency tool...")

# Test converting US Dollar to UAE Dirham
test_result = currency_info_tool.invoke({
    "source_country": "United States",
    "target_country": "UAE"
})

print(test_result)
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Trip Planning</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
ابزار برنامه‌ریزی سفر یکی از قدرتمندترین ابزارهای تراول‌بات است که با استفاده از سیستم RAG و پایگاه دانش راهنمای سفر، یک برنامه کامل و جامع برای سفر کاربر ایجاد می‌کند. شما باید ابزاری تعریف کنید که اطلاعات سفر را دریافت کرده و با جستجو در پایگاه دانش، یک itinerary روزانه همراه با پیشنهادات جاذبه‌ها، رستوران‌ها و فعالیت‌ها تولید کند.

**نکات مهم پیاده‌سازی:**

- **استفاده از RAG System:** از LanceDB و اطلاعاتی که از کتاب جهانگردی امبد کرده‌اید برای جستجوی اطلاعات مرتبط با مقصد استفاده کنید. چندین query مختلف بزنید تا اطلاعات جامعی جمع‌آوری شود.

- **پارامترهای ورودی:**

  - `destination`: نام شهر یا کشور مقصد (الزامی)
  - `arrival_date`: تاریخ ورود
  - `departure_date`: تاریخ خروج
  - `interests`: علایق کاربر مانند culture، food، nature، adventure، shopping، nightlife

- **محاسبه تعداد روزها:** با استفاده از `datetime` تعداد روزهای سفر را محاسبه کنید تا itinerary مناسب بسازید.

  - جاذبه‌های اصلی و دیدنی‌ها
  - بهترین رستوران‌ها و غذاهای محلی
  - فعالیت‌ها و تفریحات

````python
from langchain_core.tools import tool
from datetime import datetime
import lancedb
from tavily import TavilyClient
import os
import pandas as pd

# Ensure clients and models are available
# assume embedding_model and tavily_client exist in the environment
# tavily_client = TavilyClient(api_key=os.environ.get("TAVILY_API_KEY"))

@tool
def trip_planning_tool(destination: str, arrival_date: str, departure_date: str, interests: str = ""):
    """
    Generates a comprehensive trip itinerary based on destination, dates, and user interests.
    It uses a RAG system (Tourism Book) to find local insights and falls back to Web Search (Tavily) if data is missing.

    Args:
        destination (str): City or Country to visit (e.g., "Paris", "Japan").
        arrival_date (str): Start date in 'YYYY-MM-DD' format.
        departure_date (str): End date in 'YYYY-MM-DD' format.
        interests (str, optional): User interests like "history", "food", "shopping". Defaults to "".

    Returns:
        str: A structured set of information (attractions, food, activities) to build the daily itinerary.
    """

    # 1. Date calculations
    try:
        start = datetime.strptime(arrival_date, "%Y-%m-%d")
        end = datetime.strptime(departure_date, "%Y-%m-%d")
        duration = (end - start).days + 1

        if duration < 1:
            return "❌ Error: departure date must be after arrival date."

        season = start.strftime("%B")  # month of travel for seasonal suggestions
    except ValueError:
        return "❌ Error: date format must be YYYY-MM-DD."

    print(f"✈️ Planning a {duration}-day trip to {destination}...")

    # 2. Prepare search queries for RAG
    search_queries = [
        f"top tourist attractions and landmarks in {destination}",  # attractions
        f"famous local food and best restaurants in {destination}",  # food
        f"cultural activities and history of {destination}",  # culture
    ]

    if interests:
        search_queries.append(f"best {interests} activities in {destination}")

    rag_results = []

    try:
        # connect to local LanceDB
        db = lancedb.connect("./lancedb_data")
        if "book_table" in db.table_names():
            tbl = db.open_table("book_table")

            # run searches in LanceDB
            for q in search_queries:
                q_vec = embedding_model.encode([q])[0]
                # distance threshold check to avoid low-quality matches
                results = tbl.search(q_vec).limit(2).to_pandas()

                for _, row in results.iterrows():
                    # ignore low similarity results (e.g., distance > 0.6)
                    if row['_distance'] < 0.6:
                        rag_results.append(f"- {row['text']}")

    except Exception as e:
        print(f"⚠️ Error accessing LanceDB: {e}")

    # 3. If RAG results are insufficient, fallback to Tavily
    sources = []
    if len(rag_results) < 3:
        print("⚠️ Not enough information found in the book. Using Tavily Search...")
        try:
            # Example: run web searches for missing info (this is a placeholder)
            for q in search_queries:
                search_resp = tavily_client.search(query=q, max_results=2)
                for r in search_resp.get('results', []):
                    sources.append(r.get('snippet') or r.get('content'))
        except Exception as e:
            print(f"⚠️ Error using Tavily Search: {e}")

    # Build a simple itinerary text combining RAG results and web sources
    itinerary = f"Trip itinerary for {destination} ({arrival_date} to {departure_date}):\n\n"
    if rag_results:
        itinerary += "Notes from Tourism Book:\n" + "\n".join(rag_results[:10]) + "\n\n"
    if sources:
        itinerary += "Supplemental Web Sources:\n" + "\n".join(sources[:10])

    return itinerary
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;">Creating Graph Nodes</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
Reasoner node را پیاده‌سازی کنید که:
- System message و تاریخچه مکالمه را ترکیب کند
- LLM را invoke کند
- پاسخ را برگرداند
</p>
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
تعریف موفق reasoner node
</p>


````python
from typing import TypedDict, Annotated
from langgraph.graph.message import add_messages
from langchain_core.messages import HumanMessage, AIMessage

# 1. تعریف وضعیت (State)
# این کلاس وضعیت کلی مکالمه را نگهداری می‌کند (لیست پیام‌ها)
class AgentState(TypedDict):
    messages: Annotated[list, add_messages]

# 2. پیاده‌سازی Reasoner Node
def reasoner_node(state: AgentState):
    """
    این نود تاریخچه پیام‌ها را می‌خواند، پیام سیستم را به آن اضافه می‌کند
    و مدل را صدا می‌زند تا پاسخ دهد (یا ابزاری را انتخاب کند).
    """
    # دریافت پیام‌های فعلی از وضعیت
    messages = state["messages"]

    # اضافه کردن پیام سیستم به ابتدای لیست (اگر مدل از SystemMessage پشتیبانی می‌کند)
    # نکته: در LangGraph معمولاً پیام سیستم را یا در state اولیه می‌گذاریم یا اینجا prepend می‌کنیم.
    # برای اطمینان، اینجا یک لیست موقت می‌سازیم:
    full_history = [SystemMessage(content=system_prompt_text)] + messages

    # فراخوانی مدل
    response = llm_with_tools.invoke(full_history)

    # بازگرداندن پاسخ جدید برای آپدیت شدن در State
    return {"messages": [response]}

print("✅ Reasoner node defined successfully.")
````

<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
• Graph compiled موفق<br>
• تصویر visualization گراف<br>
</p>


````python
from langgraph.graph import StateGraph, END
from langgraph.prebuilt import ToolNode, tools_condition
from IPython.display import Image, display

# 1. ساخت گراف
workflow = StateGraph(AgentState)

# 2. اضافه کردن نودها
workflow.add_node("reasoner", reasoner_node)
workflow.add_node("tools", ToolNode(tools)) # نود اجرایی ابزارها

# 3. تعیین نقطه شروع
workflow.set_entry_point("reasoner")

# 4. اضافه کردن یال‌های شرطی (Conditional Edges)
# بعد از reasoner، چک می‌کنیم: اگر مدل خواست ابزار بزند -> tools، اگر نه -> END
workflow.add_conditional_edges(
    "reasoner",
    tools_condition,
)

# 5. اضافه کردن یال بازگشتی
# بعد از اجرای ابزار، نتیجه باید دوباره به reasoner برگردد تا مدل پاسخ نهایی را بسازد
workflow.add_edge("tools", "reasoner")

# 6. کامپایل کردن گراف
app = workflow.compile()

print("✅ Graph compiled successfully.")

# 7. رسم گراف (Visualization)
try:
    display(Image(app.get_graph().draw_mermaid_png()))
except Exception as e:
    print("Could not render graph (pygraphviz may need to be installed).")
    # print(app.get_graph().draw_ascii()) # fallback textual representation
````

## <div style="text-align: center; direction: rtl; font-family: Vazir;"> ساخت رابط کاربری CLI</div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
یک تابع chat_cli بنویسید که:
- در یک حلقه ورودی کاربر را بگیرد
- graph را invoke کند
- نام tool های استفاده شده را نمایش دهد
- پاسخ را چاپ کند
- دستورات exit، clear را پشتیبانی کند
</p>

<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
رابط CLI تعاملی که نمونه مکالمات را پشتیبانی کند
</p>


````python
import uuid

def chat_cli():
    print("🤖 **Travel Bot CLI**")
    print("Type 'exit' to quit, 'clear' to reset chat.")
    print("-" * 40)

    # ایجاد یک thread_id یکتا برای مدیریت حافظه در این نشست
    config = {"configurable": {"thread_id": str(uuid.uuid4())}}

    # تاریخچه محلی برای نمایش در CLI (اختیاری، چون گراف خودش state دارد)
    # اما چون ما هر بار invoke می‌کنیم، state در memory گراف ذخیره می‌شود اگر checkpointer داشته باشیم.
    # در اینجا برای سادگی، ورودی‌ها را به گراف پاس می‌دهیم.

    current_chat_history = []

    while True:
        try:
            user_input = input("\n👤 You: ").strip()

            if user_input.lower() in ["exit", "quit"]:
                print("👋 Goodbye!")
                break

            if user_input.lower() == "clear":
                current_chat_history = []
                print("🧹 Memory cleared.")
                continue

            if not user_input:
                continue

            print("⏳ Bot is thinking...", end="\r")

            # آماده‌سازی ورودی برای گراف
            # نکته: ما فقط پیام جدید کاربر را می‌فرستیم، LangGraph (اگر checkpointer داشت) تاریخچه را نگه می‌داشت.
            # در اینجا چون checkpointer ست نکردیم، باید کل تاریخچه را پاس بدهیم یا بپذیریم که هر بار state جدید است.
            # برای این تمرین، پیام کاربر را به state اضافه می‌کنیم.

            inputs = {"messages": [HumanMessage(content=user_input)]}

            # استفاده از stream برای دیدن مراحل (یا invoke برای نتیجه نهایی)
            # ما از invoke استفاده می‌کنیم و خروجی نهایی را پردازش می‌کنیم
            # اما برای دیدن ابزارها بهتر است آخرین مرحله را چک کنیم یا لاگ بگیریم.

            final_state = app.invoke(inputs, config=config)

            # استخراج آخرین پیام
            all_messages = final_state["messages"]
            last_message = all_messages[-1]

            # بررسی اینکه آیا ابزاری استفاده شده است؟
            # در تاریخچه پیام‌ها، اگر ToolMessage وجود داشته باشد یعنی ابزار صدا زده شده.
            # ما پیام‌های جدید تولید شده در این دور (turn) را بررسی می‌کنیم.

            # نمایش ابزارهای استفاده شده (از روی ToolMessage های موجود در پاسخ)
            # (این یک روش ساده‌سازی شده برای نمایش است)
            tool_calls = [m.name for m in all_messages if hasattr(m, 'tool_call_id') and m.type == 'tool']
            if tool_calls:
                print(f"\n🛠️ Tools used: {', '.join(tool_calls)}")

            print(f"\n🤖 Bot: {last_message.content}")

        except Exception as e:
            print(f"\n❌ Error: {e}")

# اجرای CLI (برای تست، خط زیر را از کامنت خارج کنید)
# chat_cli()
````

# <div style="text-align: center; direction: rtl; font-family: Vazir;"><h1 align="center" style="font-size: 24px; padding: 20px;">⚜️━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━⚜️<br>(۱۰نمره)Testing Agent With Different Scenarios<br>⚜️━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━⚜️</h1></div>


<div dir="rtl" style="text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">
<p style="line-height: 1.8; text-align: right;">
چت بات خود را برای هرابزار با حداقل دو سناریو مختلف تست کنید

برای هر سناریو:

- Query کاربر را نمایش دهید
- Tool های استفاده شده را مشخص کنید
- خروجی نهایی را نمایش دهید
- کیفیت پاسخ را تحلیل کنید
</p>
</div>


<p dir='rtl' style="line-height: 2.0; text-align: right; font-family: Vazir; font-size: 16px; margin-top: 20px; color: white; background-color:rgb(0, 40, 30); padding: 30px; border-radius: 8px;">
🎯 <b>خروجی مورد انتظار:</b><br>
۱۴ تست موفق با خروجی کامل و تحلیل نتایج
</p>


````python
import time
from langchain_core.messages import HumanMessage

# لیست کامل سناریوها (Full List)
test_scenarios = [
    # --- 1. Flight Search ---
    "Find a flight from Tehran to Dubai for next Friday.",
    "I need to fly from London to New York on 2024-05-20. Return flights?",
    "Check for available flights from Mashhad to Istanbul for tomorrow.",

    # --- 2. Hotel Search ---
    "Find a hotel in Paris for 3 nights starting tomorrow.",
    "Looking for a budget hotel in Istanbul under 100 dollars for next weekend.",
    "I need a luxury 5-star hotel in Rome near the city center for next Monday.",

    # --- 3. Restaurant Search ---
    "What are the best Italian restaurants in Rome?",
    "Where can I eat authentic Sushi in Tokyo? include price range.",
    "Find the best bakery for Croissants in Paris.",

    # --- 4. Weather Info ---
    "What is the weather like in Barcelona today?",
    "Should I pack an umbrella for my trip to London next Monday?",
    "What will be the temperature in Dubai for the next 3 days?",

    # --- 5. Currency Info ---
    "Convert 100 USD to Euros.",
    "What is the exchange rate between Iran and UAE today?",
    "How much is 1000 Canadian Dollars in Australian Dollars?",

    # --- 6. FAQ Search ---
    "Do I need a visa for Thailand?",
    "What are the rules for carrying liquids on the plane?",
    "How much baggage allowance do I usually get for international flights?",

    # --- 7. Trip Planning ---
    "Plan a 3-day trip to Istanbul focusing on history and food starting May 10th.",
    "I have a layover in Dubai for 2 days next week. What should I do?"
]

print(f"🚀 Starting Sequential Testing for {len(test_scenarios)} scenarios...\n")

for i, query in enumerate(test_scenarios, 1):
    print(f"\n🔹 **Test Scenario {i}:**")
    print(f"👤 User Query: {query}")

    # Create a unique thread ID for each scenario to keep memory clean
    config = {"configurable": {"thread_id": f"test_thread_{i}"}}
    inputs = {"messages": [HumanMessage(content=query)]}

    try:
        # app.invoke is BLOCKING (Synchronous), so it waits for the result before moving on
        final_state = app.invoke(inputs, config=config)
        last_msg = final_state["messages"][-1]

        # Extract tool usage
        tool_calls = set()
        for m in final_state["messages"]:
            if hasattr(m, 'tool_calls') and m.tool_calls:
                for tc in m.tool_calls:
                    tool_calls.add(tc['name'])
            if m.type == 'tool':
                tool_calls.add(m.name)

        print(f"🛠️ Tool(s) Used: {', '.join(tool_calls) if tool_calls else 'None (Direct Answer)'}")
        print(f"🤖 Bot Response: {last_msg.content[:250]}...")

    except Exception as e:
        print(f"❌ Error: {e}")

    print("-" * 50)

    # === CRITICAL DELAY ===
    # This prevents the loop from starting the next request too soon.
    # 15 seconds is usually safe for the Free Tier.
    if i < len(test_scenarios):
        print("⏳ Cooling down for 15 seconds to avoid API Rate Limit...")
        time.sleep(15)

print("\n✅ All tests completed.")
````
