### <div style="direction: rtl; font-family: Vazir;">مقایسه توکنایزرهای مختلف</div>

<div dir="rtl" style="width: 98%; text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

قبل از شروع کار با مدل، مهم است که نحوه کار توکنایزر (Tokenizer) را درک کنیم. توکنایزر ها نقش حیاتی در عملکرد مدل دارند و توکنایزر های مختلف می‌توانند متن یکسان را به روش‌های متفاوتی تجزیه کنند.

مراحل زیر را انجام دهید:

1. ایتدا یک توکن دسترسی Huggingface از آدرس زیر ایجاد کنید:
https://huggingface.co/settings/tokens

1. توکنایزر را برای مدل‌های زیر بارگذاری کنید:
    - `TinyLlama/TinyLlama-1.1B-Chat-v1.0`
    - `google/gemma-3-4b-it`
    - `meta-llama/Llama-3.1-8B`

2. متن `sample_text1` را بازای هر سه توکنایزر پردازش کرده و نمایش دهید.

3. یک جمله فارسی با **حداقل ۱۰ کلمه** در نظر گرفته و این کار را تکرار کنید.

هنگام استخراج توکن‌ها، به‌گونه‌ای عمل کنید که نشانه‌های
   ویژه‌ حفظ شود.

</div>

```python
from IPython.display import display, HTML
from transformers import AutoTokenizer

TOKEN_BG_COLORS = [
    "#58a6ff", "#f78166", "#85e89d", "#ffab70",
    "#bca0ff", "#79c0ff", "#ffb4a2", "#d0d0d0",
]

def show_raw_tokens(model_name, token_list):
    html_parts = [
        '<div style="margin:20px 0; padding:15px; background:#1e1e1e; border-radius:10px; '
        'border-left:6px solid #6272a4; font-family: monospace;">'
        f'<strong style="color:#ff79c6; font-size:18px;">{model_name}</strong><br><br>'
        '<div style="line-height: 1.6; white-space: pre-wrap;">'
    ]

    for i, token in enumerate(token_list):
        color = TOKEN_BG_COLORS[i % len(TOKEN_BG_COLORS)]
        style = (
            f"background:{color}; color:black; padding:2px 2px; margin:2px; "
            f"border-radius:2px; display:inline-block; font-size:13px;"
        )

        safe_token = token.replace("<", "&lt;").replace(">", "&gt;")

        html_parts.append(f'<span style="{style}">{safe_token}</span>')

    html_parts.append('</div></div>')
    display(HTML("".join(html_parts)))
```

```python
sample_text1 = """English and CAPITALIZATION 🎵 漢字
show_tokens False None elif == >= else:
two tabs:"	" Three tabs: "			"
3.25 * 12 = 39"""

sample_text2 = "این یک جمله فارسی است که برای آزمایش توکنایزر نوشته شده و شامل بیش از ده کلمه می‌باشد."
```

```python

models_to_compare = [
    "TinyLlama/TinyLlama-1.1B-Chat-v1.0",
    "google/gemma-2-2b-it",
    "meta-llama/Llama-3.2-1B"
]

```

```python

tokenizers = {}
for model_name in models_to_compare:
    try:
        tokenizer = AutoTokenizer.from_pretrained(model_name, token="YOUR_HF_TOKEN_HERE")
        tokenizers[model_name] = tokenizer
        print(f"✅ {model_name}")
    except Exception as e:
        print(f"Error loading {model_name}: {e}")
```

```python


for model_name, tokenizer in tokenizers.items():
    tokens = tokenizer.tokenize(sample_text1)
    token_ids = tokenizer.encode(sample_text1, add_special_tokens=True)

    print(f"\n{'='*100}")
    print(f"   Model: {model_name}")
    print(f"   Number of tokens: {len(tokens)}")
    print(f"   Number of IDs (with special tokens): {len(token_ids)}")

    show_raw_tokens(model_name, tokens)
```

```python


for model_name, tokenizer in tokenizers.items():
    tokens = tokenizer.tokenize(sample_text2)
    token_ids = tokenizer.encode(sample_text2, add_special_tokens=True)

    print(f"\n{'='*100}")
    print(f"🔹 Model: {model_name}")
    print(f"   Number of tokens: {len(tokens)}")
    print(f"   Number of IDs (with special tokens): {len(token_ids)}")

    show_raw_tokens(model_name, tokens)
```

```python
import pandas as pd

comparison_data = []
for model_name, tokenizer in tokenizers.items():
    en_tokens = len(tokenizer.tokenize(sample_text1))
    fa_tokens = len(tokenizer.tokenize(sample_text2))
    comparison_data.append({
        'Model': model_name.split('/')[-1],
        'English Tokens': en_tokens,
        'Persian Tokens': fa_tokens,
        'Ratio (FA/EN)': f"{fa_tokens/en_tokens:.2f}x"
    })

df_comparison = pd.DataFrame(comparison_data)
print(df_comparison.to_string(index=False))
```

<ul style='line-height: 1.8;'>

<ol style='line-height: 1.8;'>
<li><b>واژگان (Vocabulary) متفاوت:</b> هر مدل با corpus خاصی آموزش دیده است. مدل‌هایی که بر روی داده‌های چندزبانه‌تر آموزش دیده‌اند، واژگان غنی‌تری برای زبان‌های غیرانگلیسی دارند</li>

</li>

<li><b>اندازه واژگان:</b> مدل‌های مختلف واژگان با اندازه‌های متفاوت دارند (30K، 50K، 100K+ توکن). واژگان بزرگتر معمولاً پوشش بهتری از زبان‌های مختلف دارد</li>

<li><b>استراتژی normalization:</b> برخی توکنایزرها قبل از پردازش، متن را normalize می‌کنند (lowercase، unicode normalization) که می‌تواند نتیجه را تغییر دهد</li>
</ol>

### <div style="direction: rtl; font-family: Vazir;">بارگذاری مدل</div>

<div dir="rtl" style="width: 98%; text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

در این تمرین از مدل `TinyLlama-1.1B-Chat-v1.0` استفاده می‌کنیم.

این مدل یک LLM سبک با حدود 1.1 میلیارد پارامتر است و برای کاربردهای مکالمه‌ای تنظیم شده است. مدل و tokenizer آن را بارگذاری کنید.

</div>


```python
model_name = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"

import torch
from transformers import AutoModelForCausalLM, AutoTokenizer
```

```python

if torch.cuda.is_available():
    device = torch.device("cuda")
    torch_dtype = torch.float16
    map_device = "auto"
else:
    device = torch.device("cpu")
    torch_dtype = torch.float32
    map_device = None
```

```python
torch_dtype
```

```python
device
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
```

```python
print(f"   - Vocab size: {len(tokenizer)}")
print(f"   - PAD token: {tokenizer.pad_token}")
print(f"   - EOS token: {tokenizer.eos_token}")
print(f"   - BOS token: {tokenizer.bos_token}")

```

```python
if map_device:
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        torch_dtype=torch_dtype,
        device_map=map_device,
        low_cpu_mem_usage=True,
    )
else:
    model = AutoModelForCausalLM.from_pretrained(
        model_name,
        torch_dtype=torch_dtype,
        low_cpu_mem_usage=True,
    ).to(device)
```

```python
print(f"   - Number of parameters: {model.num_parameters():,}")
print(f"   - Device: {next(model.parameters()).device}")
print(f"   - Dtype: {model.dtype}")

```

```python

bytes_per_param = 2 if torch_dtype == torch.float16 else 4
param_memory_mb = (model.num_parameters() * bytes_per_param) / (1024**2)
print(f"   - Estimated VRAM memory: {param_memory_mb:.2f} MB (~{param_memory_mb/1024:.2f} GB)")
```

```python
test_prompt = "Hello, my name is"
inputs = tokenizer(test_prompt, return_tensors="pt").to(device)
```

```python

with torch.no_grad():
    outputs = model.generate(
        **inputs,
        max_new_tokens=20,
        do_sample=False
    )
```

```python
generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)
```

<li><b>FP32:</b> حافظه = N × 4 bytes</li>
<li><b>FP16:</b> حافظه = N × 2 bytes (نصف FP32)</li>
</ul>

<p style='line-height: 1.8; text-align: justify;'>
<b>مثال برای TinyLlama با 1.1 میلیارد پارامتر:</b>
</p>
<ul style='line-height: 1.8;'>
<li>FP32: 1.1B × 4 = 4.4 GB</li>
<li>FP16: 1.1B × 2 = 2.2 GB</li>
<li><b>صرفه‌جویی: 50% کاهش مصرف حافظه</b></li>
</ul>

<h4 style='color:#5bc0de;'>مزایا و معایب:</h4>

<p style='line-height: 1.8;'><b>FP16 (float16):</b></p>
<ul style='line-height: 1.8;'>
<li>مصرف حافظه نصف می‌شود</li>
<li>سرعت محاسبات بیشتر (GPU optimization)</li>
<li>امکان استفاده از batch size بزرگتر</li>
<li>دقت کمتر (ممکن است در آموزش مشکل ایجاد کند)</li>
<li>محدودیت range (underflow/overflow)</li>
</ul>

<p style='line-height: 1.8;'><b>FP32 (float32):</b></p>
<ul style='line-height: 1.8;'>
<li>دقت بالاتر</li>
<li>✅ پایداری عددی بهتر در آموزش</li>
<li>❌ مصرف حافظه دو برابر</li>
<li>❌ سرعت کمتر</li>
</ul>

<p style='line-height: 1.8; text-align: justify; background-color: #e8f4f8; padding: 10px; border-radius: 5px; margin-top: 15px;'>
💡 <b>توصیه:</b> برای inference و fine-tuning معمولاً FP16 یا BF16 کافی است و مصرف حافظه را نصف می‌کند. برای pre-training از صفر معمولاً Mixed Precision (ترکیب FP16 و FP32) استفاده می‌شود.
</p>
</div>

روش های مختلفی برای فاین تیون کردن مدل های شبکه عصبی وجود دارد. LoRA یک روش فاین‌تیون‌ سبک و کم‌هزینه است که با افزودن لایه‌های کوچک آموزش‌پذیر به برخی ماژول‌های مدل، امکان تنظیم دقیق LLMها را بدون آموزش کامل تمام وزن‌ها فراهم می‌کند.

در این بخش لازم است پیکربندی LoRA را برای مدل پایه مشخص کنید و سپس مدل پایه را با استفاده از آن به یک مدل قابل‌فاین‌تیون تبدیل کنید. در پایان، تعداد پارامترهای آموزش‌پذیر را گزارش کنید.

✍️ <b>پاسخ تشریحی:</b><br>

<p style='line-height: 1.8; text-align: justify;'>
<b>انتخاب هایپرپارامترهای LoRA بر اساس تعادل بین کارایی و عملکرد انجام شده است:</b>
</p>

<h4 style='color:#5bc0de;'>1️⃣ Rank (r = 16):</h4>
<ul style='line-height: 1.8;'>
<li><b>دلیل انتخاب:</b> rank تعیین‌کننده ظرفیت (capacity) ماتریس‌های low-rank است</li>
<li><b>r=16:</b> یک مقدار متوسط که تعادل خوبی بین کارایی و قدرت مدل‌سازی دارد</li>
<li><b>مقادیر رایج:</b> 8، 16، 32، 64</li>
<li><b>تحلیل:</b>
  - r کوچک (4-8): سریع‌تر اما ممکن است ظرفیت کافی نداشته باشد
  - r متوسط (16-32): تعادل مناسب (انتخاب ما)
  - r بزرگ (64+): عملکرد بهتر اما نزدیک به full fine-tuning
</li>
<li><b>نتیجه:</b> با r=16، تنها ~0.5% پارامترها آموزش‌پذیر می‌شوند</li>
</ul>

<h4 style='color:#5bc0de;'>2️⃣ Alpha (α = 32):</h4>
<ul style='line-height: 1.8;'>
<li><b>دلیل انتخاب:</b> alpha یک scaling factor است که تاثیر LoRA را کنترل می‌کند</li>
<li><b>فرمول:</b> scaling = α / r</li>
<li><b>α=32 با r=16:</b> scaling = 2.0</li>
<li><b>قاعده کلی:</b> معمولاً α = 2×r یا α = r انتخاب می‌شود</li>
<li><b>تحلیل:</b>
  - α/r بزرگ (>2): تاثیر قوی‌تر LoRA، ریسک instability
  - α/r متوسط (1-2): تعادل مناسب (انتخاب ما)
  - α/r کوچک (<1): تاثیر ضعیف‌تر
</li>
<li><b>نتیجه:</b> با scaling=2، وزن‌های LoRA تاثیر معنی‌داری دارند بدون اینکه مدل را ناپایدار کنند</li>
</ul>

<h4 style='color:#5bc0de;'>3️⃣ Target Modules:</h4>
<p style='line-height: 1.8; text-align: justify;'>
<b>ماژول‌های انتخاب‌شده:</b> q_proj, k_proj, v_proj, o_proj, gate_proj, up_proj, down_proj
</p>

<ul style='line-height: 1.8;'>
<li><b>Attention Projections (q, k, v, o):</b>
  - اصلی‌ترین بخش transformer که الگوهای attention را می‌آموزد
  - برای task‌های instruction-following حیاتی هستند
  - باید حتماً در LoRA شامل شوند
</li>

<li><b>FFN Projections (gate, up, down):</b>
  - بخش Feed-Forward Network که transformations غیرخطی انجام می‌دهد
  - به یادگیری الگوهای پیچیده‌تر کمک می‌کند
  - افزودن آن‌ها عملکرد را بهبود می‌بخشد
</li>

<li><b>چرا همه؟</b>
  - پوشش کامل‌تر → یادگیری بهتر
  - برای instruction-following، نیاز به تغییرات در تمام لایه‌ها داریم
  - هزینه: فقط کمی افزایش پارامترها (هنوز <1%)
</li>
</ul>

<h4 style='color:#5bc0de;'>4️⃣ سایر تنظیمات:</h4>
<ul style='line-height: 1.8;'>
<li><b>lora_dropout=0.05:</b> کمی regularization برای جلوگیری از overfitting</li>
<li><b>bias="none":</b> عدم آموزش bias برای کاهش پارامترها</li>
<li><b>task_type="CAUSAL_LM":</b> مشخص‌کننده نوع task (language modeling)</li>
</ul>

با این تنظیمات:

```python
prompts = [
    "Create a slogan for a new vintage blues record collection app.",
    "Create a slogan for a new vintage blues record collection app. The slogan should written in all capital letters.",
    "Create a slogan for a new vintage blues record collection app. The slogan should written in all capital letters and end with the word soul.",
    "Create a slogan for a new vintage blues record collection app. The slogan should written in all capital letters, end with the word soul, and contain exactly seven words.",
    "Create a slogan for a new vintage blues record collection app. The slogan should written in all capital letters, end with the word soul, and contain exactly seven words. Wrap the entire answer in JSON format."
]
```

```python

def generate_response(model, tokenizer, prompt, max_new_tokens=100):
    messages = [
        {"role": "user", "content": prompt}
    ]

    formatted_prompt = tokenizer.apply_chat_template(
        messages,
        tokenize=False,
        add_generation_prompt=True
    )

    inputs = tokenizer(formatted_prompt, return_tensors="pt").to(device)

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

    if "<|assistant|>" in generated_text:
        response = generated_text.split("<|assistant|>")[-1].strip()
    else:
        response = generated_text.replace(formatted_prompt, "").strip()

    return response
```

```python


for i, prompt in enumerate(prompts, 1):
    print(f"\n{'='*100}")
    print(f" Prompt {i}:")
    print(f"{prompt}\n")

    print("-"*100)
    print(" Base Model:")
    print("-"*100)
    base_response = generate_response(base_model, tokenizer, prompt)
    print(base_response)

    print("\n" + "-"*100)
    print(" Fine-tuned Model:")
    print("-"*100)
    finetuned_response = generate_response(merged_model, tokenizer, prompt)
    print(finetuned_response)

    print("\n" + "-"*100)
    print(" Analysis:")
    print("-"*100)

    if "capital" in prompt.lower():
        base_caps = base_response.isupper() or any(word.isupper() for word in base_response.split())
        ft_caps = finetuned_response.isupper() or any(word.isupper() for word in finetuned_response.split())
        print(f"   Capital letters: Base={base_caps}, Fine-tuned={ft_caps}")

    if "end with" in prompt.lower() and "soul" in prompt.lower():
        base_ends = base_response.strip().lower().endswith("soul")
        ft_ends = finetuned_response.strip().lower().endswith("soul")
        print(f"   Ends with 'soul': Base={base_ends}, Fine-tuned={ft_ends}")

    if "seven words" in prompt.lower():
        base_words = len(base_response.split())
        ft_words = len(finetuned_response.split())
        print(f"   Word count: Base={base_words}, Fine-tuned={ft_words} (target=7)")

    if "JSON" in prompt:
        base_json = "{" in base_response and "}" in base_response
        ft_json = "{" in finetuned_response and "}" in finetuned_response
        print(f"   JSON format: Base={base_json}, Fine-tuned={ft_json}")

    print("")

```

### <div style="direction: rtl; font-family: Vazir;">ارزیابی کمی با معیار `ifeval`</div>

<div dir="rtl" style="width: 98%; text-align: right; padding:10px; background-color:#6B7280;  border-radius: 12px; border: 2px solid rgb(2, 34, 22); font-family: Vazir;">

برای ارزیابی دقیق قابلیت instruction-following مدل، از بنچمارک **IFEval** استفاده می‌کنیم.

- مقاله: https://arxiv.org/abs/2311.07911
</div>

<div dir='rtl' style='width: 98%; background:#fffbe6; font-family: Vazir; border:1px dashed #f0ad4e; padding:12px; border-radius:8px; color:#111'>

⭐ <b>سؤال:</b>
 بصورت مختصر بیان کنید که بنچمارک <b>ifeval</b> چه نوع دستورالعمل‌هایی را ارزیابی می‌کند و معیارهای ارزیابی آن کدامند؟<br>

✍️ <b>پاسخ تشریحی:</b>

<p style='line-height: 1.8; text-align: justify;'>
<b>IFEval (Instruction-Following Evaluation)</b> یک بنچمارک جامع برای سنجش توانایی مدل‌های زبانی در پیروی دقیق از دستورالعمل‌های پیچیده و چندبخشی است.
</p>

<h4 style='color:#5bc0de;'>🎯 انواع دستورالعمل‌های ارزیابی‌شده:</h4>

<p style='line-height: 1.8;'><b>1. محدودیت‌های زبانی (Language Constraints):</b></p>
<ul style='line-height: 1.8;'>
<li><b>کلمات ممنوعه:</b> استفاده نکردن از کلمات خاص</li>
<li><b>کلمات الزامی:</b> حتماً شامل کلمات معین باشد</li>
<li><b>تکرار:</b> تکرار کلمات یا عبارات مشخص</li>
<li><b>حروف خاص:</b> محدودیت روی حروف استفاده‌شده</li>
</ul>

<p style='line-height: 1.8;'><b>2. محدودیت‌های ساختاری (Format Constraints):</b></p>
<ul style='line-height: 1.8;'>
<li><b>تعداد بخش‌ها:</b> پاسخ باید N بخش داشته باشد</li>
<li><b>تعداد پاراگراف:</b> شامل تعداد مشخصی پاراگراف</li>
<li><b>فرمت خاص:</b> JSON، bullet points، شماره‌گذاری</li>
<li><b>Title/Heading:</b> داشتن عنوان با قالب خاص</li>
</ul>

<p style='line-height: 1.8;'><b>3. محدودیت‌های طول (Length Constraints):</b></p>
<ul style='line-height: 1.8;'>
<li><b>تعداد کلمات:</b> دقیقاً N کلمه یا کمتر/بیشتر از N</li>
<li><b>تعداد جملات:</b> محدودیت روی تعداد جملات</li>
<li><b>تعداد پاراگراف:</b> حداقل/حداکثر پاراگراف</li>
</ul>

<p style='line-height: 1.8;'><b>4. محدودیت‌های محتوایی (Content Constraints):</b></p>
<ul style='line-height: 1.8;'>
<li><b>موضوع خاص:</b> پاسخ باید در مورد موضوع معینی باشد</li>
<li><b>سبک نوشتاری:</b> رسمی، غیررسمی، شاعرانه</li>
<li><b>مخاطب:</b> برای کودکان، متخصصان، عموم</li>
<li><b>لحن:</b> مثبت، منفی، خنثی</li>
</ul>

<p style='line-height: 1.8;'><b>5. محدودیت‌های پیچیده (Combination Constraints):</b></p>
<ul style='line-height: 1.8;'>
<li>ترکیب چند محدودیت همزمان</li>
<li>دستورات متناقض یا چالش‌برانگیز</li>
<li>محدودیت‌های چندسطحی</li>
</ul>

<h4 style='color:#5bc0de;'>📊 معیارهای ارزیابی:</h4>

<p style='line-height: 1.8;'><b>1. Prompt-Level Accuracy:</b></p>
<ul style='line-height: 1.8;'>
<li><b>تعریف:</b> درصد promptهایی که تمام محدودیت‌های آن‌ها رعایت شده است</li>
<li><b>محاسبه:</b> (تعداد promptهای کامل درست) / (کل promptها) × 100</li>
<li><b>سخت‌گیرانه:</b> اگر حتی یک محدودیت نقض شود، کل prompt نادرست است</li>
</ul>

<p style='line-height: 1.8;'><b>2. Instruction-Level Accuracy:</b></p>
<ul style='line-height: 1.8;'>
<li><b>تعریف:</b> درصد محدودیت‌های فردی که رعایت شده‌اند</li>
<li><b>محاسبه:</b> (تعداد محدودیت‌های رعایت‌شده) / (کل محدودیت‌ها) × 100</li>
<li><b>ملایم‌تر:</b> هر محدودیت به‌صورت مستقل ارزیابی می‌شود</li>
</ul>

<p style='line-height: 1.8;'><b>3. Strict vs Loose Evaluation:</b></p>
<ul style='line-height: 1.8;'>
<li><b>Strict:</b> رعایت دقیق محدودیت (مثلاً دقیقاً 7 کلمه)</li>
<li><b>Loose:</b> رعایت تقریبی (مثلاً 6-8 کلمه)</li>
</ul>

<h4 style='color:#5bc0de;'>🔬 مثال عملی:</h4>

<div style='background-color: #f8f9fa; padding: 15px; border-radius: 5px; border-left: 4px solid #3498db; margin: 10px 0;'>
<b>Prompt:</b> "Write a poem about cats. Use exactly 4 lines. Each line must start with 'Cat'. End with the word 'purr'."<br><br>
<b>محدودیت‌ها:</b>
<ul style='margin: 5px 0;'>
<li>✓ موضوع: cats</li>
<li>✓ تعداد خطوط: 4</li>
<li>✓ شروع هر خط: "Cat"</li>
<li>✓ کلمه پایانی: "purr"</li>
</ul>
<b>Instruction-Level:</b> 4 محدودیت<br>
<b>Prompt-Level:</b> باید همه 4 محدودیت رعایت شوند
</div>

<h4 style='color:#5bc0de;'>📈 ویژگی‌های کلیدی:</h4>
<ul style='line-height: 1.8;'>
<li>✅ <b>Verifiable:</b> محدودیت‌ها قابل بررسی خودکار هستند</li>
<li>✅ <b>Diverse:</b> طیف وسیعی از انواع محدودیت‌ها</li>
<li>✅ <b>Challenging:</b> برای مدل‌های پیشرفته هم چالش‌برانگیز است</li>
<li>✅ <b>Objective:</b> ارزیابی عینی بدون نیاز به قضاوت انسانی</li>
</ul>

<p style='line-height: 1.8; text-align: justify; background-color: #e8f4f8; padding: 10px; border-radius: 5px; margin-top: 15px;'>
💡 <b>اهمیت:</b> IFEval به ما نشان می‌دهد که مدل تا چه حد می‌تواند دستورات پیچیده را درک و اجرا کند - مهارتی حیاتی برای کاربردهای واقعی AI assistants.
